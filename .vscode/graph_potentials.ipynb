```python
## Pothan Tang 8/15/25
import numpy as np
import matplotlib.pyplot as plt

original_shape_3d = (201,201,201)

## Region of interest
um = 1e-6;
cm = 1e-2;
mm = 1e-3; 
res = 1*um; # resolution of potential
x_max = 100*um;
y_max = 100*um;
z_max = 200*um;
x = np.linspace(-1*x_max,x_max, int(2*x_max/res+1), endpoint=True); # -1cm<=x<=1cm 
y = np.linspace(-1*y_max,y_max, int(2*y_max/res+1), endpoint=True);  # -1cm<=y<=1cm
z = np.linspace(0,z_max, int(z_max/res+1), endpoint=True);  # 0cm<=z<=1cm

## Graph of DC potential
phi_dc_flat = np.loadtxt('dc_potential_zoomed')
phi_dc = phi_dc_flat.reshape(original_shape_3d)

# plane x=0
i = 100  # middle x-slice
y_mask = (y >= -100e-6) & (y <= 100e-6)
z_mask = (z >= 0) & (z <= 200e-6)
y_subset = y[y_mask]
z_subset = z[z_mask]
phi_dc_subset = phi_dc[i, :, :][y_mask, :][:, z_mask]

Y, Z = np.meshgrid(y_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(Y/um, Z/um, phi_dc_subset, levels=50, cmap='plasma')
plt.contour(Y/um, Z/um, phi_dc_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlabel('y (μm)')
plt.ylabel('z (μm)')
plt.title(f'DC Potential in YZ plane at x = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane y=0
j = 100  # middle y-slice
x_mask = (x >= -100e-6) & (x <= 100e-6)
z_mask = (z >= 0) & (z <= 200e-6)
x_subset = x[x_mask]
z_subset = z[z_mask]
phi_dc_subset = phi_dc[:, j, :][x_mask, :][:, z_mask]

X, Z = np.meshgrid(x_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(X/um, Z/um, phi_dc_subset, levels=50, cmap='plasma')
plt.contour(X/um, Z/um, phi_dc_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlabel('x (μm)')
plt.ylabel('z (μm)')
plt.title(f'DC Potential in XZ plane at y = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane z=0
k = 0  # check boundary for z-slice at M4 surface
x_mask = (x >= -100e-6) & (x <= 100e-6)
y_mask = (y >= -100e-6) & (y <= 100e-6)
x_subset = x[x_mask]
y_subset = y[y_mask]
phi_dc_subset = phi_dc[:, :, k][x_mask, :][:, y_mask]

X, Y = np.meshgrid(x_subset, y_subset, indexing='ij')  

plt.figure()
cf = plt.contourf(X/um, Y/um, phi_dc_subset, levels=50, cmap='viridis')
plt.contour(X/um, Y/um, phi_dc_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title(f'DC Potential in XY plane at z = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane z=70 um
k = 70  # z-slice at ion height
x_mask = (x >= -100e-6) & (x <= 100e-6)
y_mask = (y >= -100e-6) & (y <= 100e-6)
x_subset = x[x_mask]
y_subset = y[y_mask]
phi_dc_subset = phi_dc[:, :, k][x_mask, :][:, y_mask]

X, Y = np.meshgrid(x_subset, y_subset, indexing='ij')  # Match phi[x, y, z]

plt.figure()
cf = plt.contourf(X/um, Y/um, phi_dc_subset, levels=50, cmap='viridis')
plt.contour(X/um, Y/um, phi_dc_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlim(-1*x_max, x_max)
plt.ylim(-1*y_max, y_max)
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title(f'DC Potential in XY plane at z = 70 μm')
plt.axis('equal')
plt.tight_layout()
plt.show()

## Graph of RF potential
phi_rf_flat = np.loadtxt('rf_potential_max_zoomed')
phi_rf = phi_rf_flat.reshape(original_shape_3d)

# plane x=0
i = 100  # middle x-slice
y_mask = (y >= -100e-6) & (y <= 100e-6)
z_mask = (z >= 0) & (z <= 200e-6)
y_subset = y[y_mask]
z_subset = z[z_mask]
phi_rf_subset = phi_rf[i, :, :][y_mask, :][:, z_mask]

Y, Z = np.meshgrid(y_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(Y/um, Z/um, phi_rf_subset, levels=50, cmap='inferno')
plt.contour(Y/um, Z/um, phi_rf_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlim(-1*y_max, y_max)
plt.ylim(0, z_max)
plt.xlabel('y (μm)')
plt.ylabel('z (μm)')
plt.title(f'Max RF Potential in YZ plane at x = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane y=0
j = 100  # middle y-slice
x_mask = (x >= -100e-6) & (x <= 100e-6)
z_mask = (z >= 0) & (z <= 200e-6)
x_subset = x[x_mask]
z_subset = z[z_mask]
phi_rf_subset = phi_rf[:, j, :][x_mask, :][:, z_mask]

X, Z = np.meshgrid(x_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(X/um, Z/um, phi_rf_subset, levels=50, cmap='plasma')
plt.contour(X/um, Z/um, phi_rf_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlim(-1*x_max, x_max)
plt.ylim(0, z_max)
plt.xlabel('x (μm)')
plt.ylabel('z (μm)')
plt.title(f'Max RF Potential in XZ plane at y = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane z=0
k = 0  # check boundary for z-slice at M4 surface
x_mask = (x >= -100e-6) & (x <= 100e-6)
y_mask = (y >= -100e-6) & (y <= 100e-6)
x_subset = x[x_mask]
y_subset = y[y_mask]
phi_rf_subset = phi_rf[:, :, k][x_mask, :][:, y_mask]
X, Y = np.meshgrid(x_subset, y_subset, indexing='ij')  

plt.figure()
cf = plt.contourf(X/um, Y/um, phi_rf_subset, levels=50, cmap='viridis')
plt.contour(X/um, Y/um, phi_rf_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title(f'Max RF Potential in XY plane at z = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane z=70 um
k = 70  # z-slice at ion height
x_mask = (x >= -50e-6) & (x <= 50e-6)
y_mask = (y >= -50e-6) & (y <= 50e-6)
x_subset = x[x_mask]
y_subset = y[y_mask]
phi_rf_subset = phi_rf[:, :, k][x_mask, :][:, y_mask]

X, Y = np.meshgrid(x_subset, y_subset, indexing='ij')  

plt.figure()
cf = plt.contourf(X/um, Y/um, phi_rf_subset, levels=50, cmap='viridis')
plt.contour(X/um, Y/um, phi_rf_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential (V)')
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title(f'Max RF Potential in XY plane at z = 70 μm')
plt.axis('equal')
plt.tight_layout()
plt.show()

## Graph of RF pseudo-potential
pseudo_flat = np.loadtxt('pseudo_potential_zoomed')
pseudo = pseudo_flat.reshape(original_shape_3d)

# plane x=0
i = 100  # middle x-slice
y_mask = (y >= -60e-6) & (y <= 60e-6)
z_mask = (z >= 50e-6) & (z <= 170e-6)
y_subset = y[y_mask]
z_subset = z[z_mask]
pseudo_subset = pseudo[i, :, :][y_mask, :][:, z_mask]  # shape [len(y), len(z_subset)]

Y, Z = np.meshgrid(y_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(Y/um, Z/um, pseudo_subset, levels=50, cmap='plasma')
plt.contour(Y/um, Z/um, pseudo_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential Energy (J)')
plt.xlabel('y (μm)')
plt.ylabel('z (μm)')
plt.title(f'Pseudo-potential in YZ plane at x = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane y=0
j = 100  # middle y-slice
x_mask = (x >= -10e-6) & (x <= 10e-6)
z_mask = (z >= 60e-6) & (z <= 80e-6)
x_subset = x[x_mask]
z_subset = z[z_mask]
pseudo_subset = pseudo[:, j, :][x_mask, :][:, z_mask]

X, Z = np.meshgrid(x_subset, z_subset, indexing='ij')

plt.figure()
cf = plt.contourf(X/um, Z/um, pseudo_subset, levels=50, cmap='plasma')
plt.contour(X/um, Z/um, pseudo_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential Energy (J)')
plt.xlim(-1*x_max, x_max)
plt.ylim(0, z_max)
plt.xlabel('x (μm)')
plt.ylabel('z (μm)')
plt.title(f'Pseudo-potential in XZ plane at y = 0')
plt.axis('equal')
plt.tight_layout()
plt.show()

# plane z=70 um
k = 70  # z-slice at ion height
x_mask = (x >= -50e-6) & (x <= 50e-6)
y_mask = (y >= -50e-6) & (y <= 50e-6)
x_subset = x[x_mask]
y_subset = y[y_mask]
pseudo_subset = pseudo[:, :, k][x_mask, :][:, y_mask]

X, Y = np.meshgrid(x_subset, y_subset, indexing='ij')  

plt.figure()
cf = plt.contourf(X/um, Y/um, pseudo_subset, levels=50, cmap='viridis')
plt.contour(X/um, Y/um, pseudo_subset, levels=50, colors='k', linewidths=0.5)
plt.colorbar(cf,label='Potential Energy (J)')
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title(f'Pseudo-potential in XY plane at z = 70 μm')
plt.axis('equal')
plt.tight_layout()
plt.show()
```
