# Potential/Force Computation
## Compute DC Potential

```python
## From dc_potential.ipynb
## Functions
# (x,y,z) = coordinate of sample
# (xik,yik,z0) = ith corner coordinates of kth electrode
# vk = voltage applied to kth electrode
def potential_term(x,y,z,xik,yik,z0):
    num = (xik-x)*(yik-y); # numerator
    den = (z-z0)*np.sqrt((z-z0)**2+(xik-x)**2+(yik-y)**2); # denominator
    return num/den

def dc_potential_single_electrode(x,y,z,x1k,y1k,x2k,y2k,z0,vk):
    term1 = np.arctan(potential_term(x,y,z,x2k,y2k,z0))
    term2 = np.arctan(potential_term(x,y,z,x1k,y2k,z0))
    term3 = np.arctan(potential_term(x,y,z,x2k,y1k,z0))
    term4 = np.arctan(potential_term(x,y,z,x1k,y1k,z0))
    return (vk/(2*math.pi))*(term1-term2-term3+term4)

def dc_potential_total(x,y,z,xy1k=c.xy1k,xy2k=c.xy2k,z0=c.z0,vk=c.vk):
    # extract individual (xik,yik),vk values from array
    # all arrays must be same size
    # sum potentials from all electrodes
    # Allow x, y, and z to be either scalars or same-shaped arrays.
    x, y, z = np.broadcast_arrays(
        np.asarray(x, dtype=float),
        np.asarray(y, dtype=float),
        np.asarray(z, dtype=float),
    )
    z = z+c.ion_height # account for ion height
    pot = np.zeros_like(x, dtype=float)
    for i in range (0,len(c.vk)):
        x1k = xy1k[i][0]
        y1k = xy1k[i][1]
        x2k = xy2k[i][0]
        y2k = xy2k[i][1]
        v = vk[i]
        if(i==19 or i==39): # make sure Q39-40 are on the M4 layer
            new_pot = dc_potential_single_electrode(x,y,z,x1k,y1k,x2k,y2k,0,v); # z0=0 for Q39-40
        else:
            new_pot = dc_potential_single_electrode(x,y,z,x1k,y1k,x2k,y2k,z0,v); # z0 = -11.8um for Q1-38
        pot+=new_pot
    return pot
```

## Compute RF Potential

```python
## From rf_potential.ipynb
## Functions
## Compute Pseudopotential (J)
"""
# Method 1: numerical RF gradient
Ex,Ey,Ez = np.gradient(-1*phi_rf,res,res,res); # Electric field components
E_squared = Ex**2+Ey**2+Ez**2; # Electric field strength squared
pseudo2 = (c.Z**2*c.e**2)/(4*c.m*c.omega**2)*E_squared
# pseudo_reshaped = pseudo.reshape(int(2*x_max/res+1),-1)
"""
# Method 2: analytical RF gradient
def divatan(up,down):
    #This is d(arctan2(up,down))/ddown up to a minus sign. It's useful for the pseudo-potential
    return up/(up**2+down**2)
def divatanup(up,down):
    #This is d(divatan(up,down))/dup
    return (down**2-up**2)/(up**2+down**2)**2
def divatandown(up,down):
    #This is d(divatan(up,down))/ddown
    return -2*up*down/(up**2+down**2)**2
def pseudopotential(x,y,z,m,Z,q=c.e,omrf=c.omega,VRF=c.vrf,ymin=c.y11,yedge1=c.y21,yedge2=c.y12,ymax=c.y22):
    #this is the pseudo potential, which is Z^2*(Div[PhiRF]/cos(om*t))^2/(4m*omega^2)
    x, y, z = np.broadcast_arrays(
        np.asarray(x, dtype=float),
        np.asarray(y, dtype=float),
        np.asarray(z, dtype=float),
    )
    z = z+c.ion_height # account for ion height
    divypart=divatan(z,yedge2-y)-divatan(z,yedge1-y)+divatan(z,ymin-y)-divatan(z,ymax-y)
    divzpart=divatan(yedge2-y,z)-divatan(yedge1-y,z)+divatan(ymin-y,z)-divatan(ymax-y,z)
    return ((Z**2*q**2)/(4*m*omrf**2)) * (VRF/math.pi)**2 * (divypart**2+divzpart**2)
```

```python
## Compute Full Potential Energy
def coulomb_pot(x,y,z,eps,Z=c.Z):
    x, y, z = np.broadcast_arrays(
        np.asarray(x, dtype=float),
        np.asarray(y, dtype=float),
        np.asarray(z, dtype=float),
    )
    return c.K*eps*Z*c.e**2/np.sqrt(x**2+y**2+z**2)
    
def potential_energy(xyz, m_dm, eps, component="total"):
    """
    Compute potential energy for either one point or many points.
    Accepted input shapes
    ---------------------
    xyz.shape == (3,)
        Returns a scalar for a single component.
    xyz.shape == (n, 3)
        Returns an array of shape (n,) for a single component.
    """
    xyz = np.asarray(xyz, dtype=float)
    single_input = False
    if xyz.ndim == 1:
        if xyz.shape != (3,):
            raise ValueError(
                "A single position must have shape (3,)"
            )
        xyz = xyz.reshape(1, 3)
        single_input = True
    elif xyz.ndim != 2 or xyz.shape[1] != 3:
        raise ValueError(
            "xyz must have shape (3,) or (n, 3)"
        )
    x = xyz[:, 0]
    y = xyz[:, 1]
    z = xyz[:, 2]
    dc = eps * c.e * dc_potential_total(x, y, z)
    pseudo_rf = pseudopotential(
        x,
        y,
        z,
        m_dm,
        eps,
    )
    coulomb = coulomb_pot(
        x,
        y,
        z,
        eps,
    )
    if component == "dc":
        out = dc
    elif component == "pseudo_rf":
        out = pseudo_rf
    elif component == "coulomb":
        out = coulomb
    elif component == "trap":
        out = dc + pseudo_rf
    elif component == "total":
        out = dc + pseudo_rf + coulomb
    elif component == "all":
        if single_input:
            return (
                float(dc[0]),
                float(pseudo_rf[0]),
                float(coulomb[0]),
            )
        return dc, pseudo_rf, coulomb
    else:
        raise ValueError(
            f"Unknown potential component: {component}"
        )
    if single_input:
        return float(out[0])
    return np.asarray(out, dtype=float)
```

## Compute RF/DC Force

```python
"""
xyz : ndarray, shape (n, 3)
        Particle positions.
"""
# RF pseudo-force
def FRF(xyz,m,Z,q=c.e,omrf=c.omega,VRF=c.vrf,ymin=c.y11,yedge1=c.y21,yedge2=c.y12,ymax=c.y22):
    xyz = np.asarray(xyz, dtype=float)
    if xyz.ndim != 2 or xyz.shape[1] != 3:
        raise ValueError("xyz must have shape (n, 3)")
    x = xyz[:, 0]
    y = xyz[:, 1]
    z = xyz[:, 2] + c.ion_height # Take into account ion height
    m = float(np.asarray(m).ravel()[0])
    Z = float(np.asarray(Z).ravel()[0])
    q = float(np.asarray(q).ravel()[0])
    omrf = float(np.asarray(omrf).ravel()[0])
    VRF = float(np.asarray(VRF).ravel()[0])
    #this is the pseudo potential, which is Z^2*(Div[PhiRF]/cos(om*t))^2/(4m*omega^2)
    divypart=divatan(z,yedge2-y)-divatan(z,yedge1-y)+divatan(z,ymin-y)-divatan(z,ymax-y)
    divzpart=divatan(yedge2-y,z)-divatan(yedge1-y,z)+divatan(ymin-y,z)-divatan(ymax-y,z) 
    divyparty=-2*divypart*(divatandown(z,yedge2-y)-divatandown(z,yedge1-y)+divatandown(z,ymin-y)-divatandown(z,ymax-y)) 
    divypartz=2*divypart*(divatanup(z,yedge2-y)-divatanup(z,yedge1-y)+divatanup(z,ymin-y)-divatanup(z,ymax-y)) 
    divzparty=-2*divzpart*(divatanup(yedge2-y,z)-divatanup(yedge1-y,z)+divatanup(ymin-y,z)-divatanup(ymax-y,z)) 
    divzpartz=2*divzpart*(divatandown(yedge2-y,z)-divatandown(yedge1-y,z)+divatandown(ymin-y,z)-divatandown(ymax-y,z)) 
    return -1*((VRF*Z*q)/(2*pi*np.sqrt(m)*omrf))**2*np.column_stack([divypartz*0, divzparty+divyparty, divzpartz+divypartz])

# DC force
def divatan(up,down):
    #This is d(arctan2(up,down))/ddown up to a minus sign. It's useful for the pseudo-potential
    return up/(up**2+down**2)
def divatanup(up,down):
    #This is d(divatan(up,down))/dup
    return (down**2-up**2)/(up**2+down**2)**2
def divatandown(up,down):
    #This is d(divatan(up,down))/ddown
    return -2*up*down/(up**2+down**2)**2

def anatangrad(xi,yi,xyz,v): # gradient term of DC potential
    xyz = np.asarray(xyz, dtype=float)
    x = xyz[:, 0]
    y = xyz[:, 1]
    z = xyz[:, 2]
    dy=y-yi
    dx=x-xi
    r = np.sqrt(dx**2+dy**2+z**2); # added distance
    dry2=z**2+dy**2
    drx2=z**2+dx**2
    divy=z*dx/(r*dry2); # divide by factor r
    divz=-dy*dx*(1/dry2+1/drx2)/r; # divide by factor r
    divx=z*dy/(r*drx2); # divide by factor r
    return (v/(2*np.pi))*np.column_stack([divx, divy, divz])

def FDC_single(x1,y1,x2,y2,xyz,v,Z,q=c.e):
    return -Z*q * (anatangrad(x2,y2,xyz,v)-anatangrad(x2,y1,xyz,v)-anatangrad(x1,y2,xyz,v)+anatangrad(x1,y1,xyz,v))

def FDC(xyz, m, Z, q=c.e, omrf=c.omega,
        ymin=c.y11, yedge1=c.y21, yedge2=c.y12, ymax=c.y22):
    xyz = np.asarray(xyz, dtype=float)
    if xyz.ndim != 2 or xyz.shape[1] != 3:
        raise ValueError("xyz must have shape (n, 3)")
    z_original = xyz[:, 2].copy() + c.ion_height # Take into account ion height
    force = np.zeros_like(xyz)
    for k in range(0, 40):
        xyz_k = xyz.copy()
        if (k != 19 and k != 39):
            xyz_k[:, 2] = z_original - z_offset  # keep scalar math
        else:
            xyz_k[:, 2] = z_original
        (x1k, y1k) = xy1k[k]
        (x2k, y2k) = xy2k[k]
        # enforce scalar on vk[k]
        v_k = float(np.asarray(vk[k]).ravel()[0])
        force_single = FDC_single(x1k, y1k, x2k, y2k, xyz_k, v_k, Z, q=q)
        # check shape of force_single 
        force_single = np.asarray(force_single, dtype=float)
        if force_single.shape != xyz.shape:
            raise ValueError(
                f"FDC_single returned unexpected shape {force_single.shape}; "
                f"expected {xyz.shape} at k={k}"
            )
        force += force_single
    return force

def force(xyz, m_dm, eps, component="total"):
    xyz = np.asarray(xyz, dtype=float)

    single_input = False
    if xyz.ndim == 1:
        xyz = xyz.reshape(1, 3)
        single_input = True

    if xyz.ndim != 2 or xyz.shape[1] != 3:
        raise ValueError("xyz must have shape (n, 3) or (3,)")

    dc_force = FDC(xyz, m_dm, eps)
    pseudo_force = FRF(xyz, m_dm, eps)

    r = np.linalg.norm(xyz, axis=1)

    coulomb_force = np.full_like(xyz, np.nan, dtype=float)
    good = r > 0

    coulomb_force[good] = (
        c.K * c.Z * eps * c.e**2
        * xyz[good]
        / r[good, None]**3
    )

    if component == "dc":
        out = dc_force
    elif component == "pseudo_rf":
        out = pseudo_force
    elif component == "coulomb":
        out = coulomb_force
    elif component == "trap":
        out = dc_force + pseudo_force
    elif component == "total":
        out = dc_force + pseudo_force + coulomb_force
    elif component == "all":
        return dc_force, pseudo_force, coulomb_force
    else:
        raise ValueError(f"Unknown force component: {component}")

    if single_input:
        return out[0]

    return out
```

```python
def xyz_on_sphere(n, radius, center=(0.0, 0.0, 0.0)):
    """
    Generate positions evenly distributed on the surface of a sphere.

    Parameters
    ----------
    n_dm : int
        Number of coordinate points.
    radius : float
        Sphere radius in meters.
    center : array-like, shape (3,)
        Sphere center in meters.
    
    Returns
    -------
    x0_dms : ndarray, shape (n_dm, 3)
        Positions on the sphere surface.
    """

    center = np.asarray(center, dtype=float)
    if n <= 0:
        raise ValueError("n_dm must be positive")
    indices = np.arange(n)
    golden_angle = np.pi * (3.0 - np.sqrt(5.0))
    z = 1.0 - 2.0 * (indices + 0.5) / n
    r_xy = np.sqrt(1.0 - z**2)
    theta = golden_angle * indices
    x = r_xy * np.cos(theta)
    y = r_xy * np.sin(theta)
    points = np.column_stack([x, y, z])
    x0_dms = center + radius * points
    return x0_dms

def valid_xyz_mask(xyz):
    xyz = np.asarray(xyz, dtype=float)
    return xyz[..., 2] > (z_min_valid + z_margin)

def xyz_on_valid_sphere(n, radius, center=(0.0, 0.0, 0.0)):
    xyz = xyz_on_sphere(n, radius, center=center)
    mask = valid_xyz_mask(xyz)
    return xyz[mask]

def valid_fraction_on_sphere(n, radius):
    xyz = xyz_on_sphere(n, radius)
    return np.mean(valid_xyz_mask(xyz))

def PE_on_sphere(
    n,
    radius,
    m_dm,
    eps,
    component="total",
):
    xyz = xyz_on_valid_sphere(n, radius)
    if xyz.shape[0] == 0:
        return np.array([], dtype=float)
    return np.asarray(
        potential_energy(
            xyz,
            m_dm,
            eps,
            component=component,
        ),
        dtype=float,
    )

def force_on_sphere(
    n,
    radius,
    m_dm,
    eps,
    component="total",
):
    xyz = xyz_on_valid_sphere(n, radius)
    if xyz.shape[0] == 0:
        return np.empty((0, 3), dtype=float)
    return np.asarray(
        force(
            xyz,
            m_dm,
            eps,
            component=component,
        ),
        dtype=float,
    )

def deterministic_MB_speeds(T, m_dm, quantile):
    k_B = 1.380649e-23
    scale = np.sqrt(k_B * T / m_dm)
    speeds = maxwell.ppf(quantile, scale=scale)
    return speeds
```

## Consistency of Potential vs Force

```python
def numerical_force_from_potential(xyz, m_dm, eps, component, h=0.05e-6):
    xyz = np.asarray(xyz, dtype=float)

    if xyz.shape != (3,):
        raise ValueError("xyz must have shape (3,)")

    F = np.zeros(3)

    for k in range(3):
        step = np.zeros(3)
        step[k] = h

        xyz_plus = xyz + step
        xyz_minus = xyz - step

        # Avoid finite-difference points outside valid potential domain.
        if xyz_plus[2] <= z_min_valid + z_margin:
            F[k] = np.nan
            continue

        if xyz_minus[2] <= z_min_valid + z_margin:
            F[k] = np.nan
            continue

        U_plus = potential_energy(xyz_plus, m_dm, eps, component)
        U_minus = potential_energy(xyz_minus, m_dm, eps, component)

        F[k] = -(U_plus - U_minus) / (2.0 * h)

    return F


test_points = [
    np.array([0.0, 0.0, 1e-6]),
    np.array([0.0, 0.0, 10e-6]),
    np.array([10e-6, 0.0, 10e-6]),
    np.array([0.0, 10e-6, 10e-6]),
    np.array([0.0, 0.0, 100e-6]),
    np.array([100e-6, 0.0, 300e-6]),
]

for xyz in test_points:
    if xyz[2] <= z_min_valid + z_margin:
        continue

    print()
    print("xyz =", xyz)

    for component in ["dc", "pseudo_rf", "coulomb", "trap", "total"]:
        F_analytic = np.asarray(force(xyz, m_dm, eps, component), dtype=float).reshape(3)
        F_numeric = numerical_force_from_potential(xyz, m_dm, eps, component)

        diff = F_analytic - F_numeric

        denom = max(
            np.linalg.norm(F_analytic),
            np.linalg.norm(F_numeric),
            1e-300,
        )

        rel_err = np.linalg.norm(diff) / denom

        print(component)
        print("  analytic:", F_analytic)
        print("  numeric :", F_numeric)
        print("  diff    :", diff)
        print("  rel_err :", rel_err)
```

# Full Simulation Code

## Original

```python
def run_simulation_many_dm_single_ion(
    x0_dms, v0_dms, x0_ion, v0_ion,
    t_span, rtol, atol, dt, m_dm, eps, run_rutherford
):
    """
    Run a single-ion / many-DM trajectory simulation.

    This version includes:
        - one trapped ion near the origin
        - many DM particles
        - ion-DM Coulomb forces
        - no DM-DM Coulomb forces

    Parameters
    ----------
    x0_dms : array-like, shape (n_dm, 3)
        Initial positions of the DM particles.

    v0_dms : array-like, shape (n_dm, 3)
        Initial velocities of the DM particles.

    x0_ion : array-like, shape (3,)
        Initial position of the single ion.

    v0_ion : array-like, shape (3,)
        Initial velocity of the single ion.

    t_span : tuple (t_start, t_end)
        Start and end time of the integration.

    dt : float
        Time increment used to build the evaluation grid.

    m_dm : float
        Mass of DM particles in kg.
    
    eps : float
        Charge of DM particles in units of elementary charge.

    run_rutherford : bool
        If True, set trap fields to zero for Rutherford scattering comparison.

    Returns
    -------
    t_eval : ndarray, shape (N,)
        Times at which the solution was evaluated.

    x_ion_sol : ndarray, shape (3, N)
        Ion position trajectory.

    v_ion_sol : ndarray, shape (3, N)
        Ion velocity trajectory.

    x_dms_sol : ndarray, shape (n_dm, 3, N)
        DM position trajectories.

    v_dms_sol : ndarray, shape (n_dm, 3, N)
        DM velocity trajectories.
    """

    x0_dms = np.asarray(x0_dms, dtype=float)
    v0_dms = np.asarray(v0_dms, dtype=float)
    x0_ion = np.asarray(x0_ion, dtype=float)
    v0_ion = np.asarray(v0_ion, dtype=float)

    if x0_dms.ndim != 2 or x0_dms.shape[1] != 3:
        raise ValueError("x0_dms must have shape (n_dm, 3)")

    if v0_dms.shape != x0_dms.shape:
        raise ValueError("v0_dms must have the same shape as x0_dms")

    if x0_ion.shape != (3,):
        raise ValueError("x0_ion must have shape (3,)")

    if v0_ion.shape != (3,):
        raise ValueError("v0_ion must have shape (3,)")

    n_dm = x0_dms.shape[0]

    # The single ion is trapped about the origin.
    r_ion_eq = np.zeros(3, dtype=float)

    def rhs_unified(t, U, r_ion_eq, use_harmonic_ion=True):
        """
        State vector layout:

        U = [
            x_ion(3),
            x_dms(3*n_dm),
            v_ion(3),
            v_dms(3*n_dm)
        ]
        """

        idx0 = 0
        idx1 = idx0 + 3
        idx2 = idx1 + 3 * n_dm
        idx3 = idx2 + 3
        idx4 = idx3 + 3 * n_dm

        x_ion = U[idx0:idx1]
        x_dms = U[idx1:idx2].reshape((n_dm, 3))

        v_ion = U[idx2:idx3]
        v_dms = U[idx3:idx4].reshape((n_dm, 3))

        dx_iondt = v_ion
        dx_dmsdt = v_dms

        dv_iondt = np.zeros(3)
        dv_dmsdt = np.zeros((n_dm, 3))

        # --------------------------------------------------
        # Trap force acting on ion
        # --------------------------------------------------
        if use_harmonic_ion:
            if run_rutherford:
                dv_iondt += np.zeros(3)
            else:
                dv_iondt += -omega_vec**2 * (x_ion - r_ion_eq)
        else:
            if not run_rutherford:
                x_ion_arr = x_ion[None, :]
                F_dc_ion = FDC(x_ion_arr, m=m_ion, Z=Z_ion)[0]
                F_rf_ion = FRF(x_ion_arr, m=m_ion, Z=Z_ion)[0]
                F_ion = F_dc_ion + F_rf_ion
                dv_iondt += F_ion / m_ion

        # --------------------------------------------------
        # Trap force acting on each DM particle
        # --------------------------------------------------
        if not run_rutherford:
            F_dc_dms = FDC(x_dms, m_dm, eps)
            F_rf_dms = FRF(x_dms, m_dm, eps)

            dv_dmsdt += (F_dc_dms + F_rf_dms) / m_dm

        # --------------------------------------------------
        # Coulomb interactions: ion <-> each DM particle
        #
        # No DM-DM Coulomb forces are included.
        # --------------------------------------------------
        delta = x_ion[None, :] - x_dms
        r = np.linalg.norm(delta, axis=1)

        forces = K * Z_ion * eps * e**2 * delta / r[:, None]**3

        # Force on ion due to DM k
        dv_iondt += forces.sum(axis=0) / m_ion

        # Equal and opposite force on DM k due to ion
        dv_dmsdt -= forces / m_dm

        return np.concatenate([
            dx_iondt,
            dx_dmsdt.flatten(),
            dv_iondt,
            dv_dmsdt.flatten()
        ])

    print("Number of DM particles:", n_dm)
    print("Initial ion position:", x0_ion)
    print("Initial ion velocity:", v0_ion)
    if(x0_dms.shape==(1,3)):
        print("Initial DM position:", x0_dms)
        print("Initial DM velocity:", v0_dms)

    # Initial state
    U0 = np.concatenate([
        x0_ion,
        x0_dms.flatten(),
        v0_ion,
        v0_dms.flatten()
    ]).astype(float)

    # Time grid
    t_eval = np.arange(t_span[0], t_span[1], dt)

    # Solve
    sol = solve_ivp(
        rhs_unified,
        t_span,
        U0,
        method="DOP853",
        t_eval=t_eval,
        dense_output=True,
        rtol=rtol,
        atol=atol,
        max_step=dt,
        args=(r_ion_eq,)
    )

    print("Success?", sol.success)
    print("Message:", sol.message)

    if sol.t.size > 0:
        print("Final time:", sol.t[-1])

    # Extract solution
    idx0 = 0
    idx1 = idx0 + 3
    idx2 = idx1 + 3 * n_dm
    idx3 = idx2 + 3
    idx4 = idx3 + 3 * n_dm

    x_ion_sol = sol.y[idx0:idx1]
    x_dms_flat_sol = sol.y[idx1:idx2]

    v_ion_sol = sol.y[idx2:idx3]
    v_dms_flat_sol = sol.y[idx3:idx4]

    # Convert flattened DM solution into shape (n_dm, 3, N)
    n_t = sol.y.shape[1]

    x_dms_sol = x_dms_flat_sol.reshape((n_dm, 3, n_t))
    v_dms_sol = v_dms_flat_sol.reshape((n_dm, 3, n_t))

    return t_eval, x_ion_sol, v_ion_sol, x_dms_sol, v_dms_sol
```

```python
def analyze_simulation_results(x0_dms, v0_dms, x0_ion, v0_ion, t_span, dt, t_min, m_dm, eps, fit_curve, rutherford, show_plots,
                                rtol, atol):
    # default rtol=1e-13, atol=1e-16
    if(rutherford):
        t00, xion00, vion00, xdms00, vdms00 = run_simulation_many_dm_single_ion(x0_dms, v0_dms, x0_ion, v0_ion, t_span, rtol,atol,dt, m_dm, eps, run_rutherford=True)
    else:
        t00, xion00, vion00, xdms00, vdms00 = run_simulation_many_dm_single_ion(x0_dms, v0_dms, x0_ion, v0_ion, t_span, rtol,atol,dt, m_dm, eps, run_rutherford=False)
    t0 = t00/um
    xion0 = xion00/um
    vion0 = vion00
    xdms0 = xdms00/um
    vdms0 = vdms00
    x0_ion1, y0_ion1, z0_ion1 = xion0[0], xion0[1], xion0[2] # motion of center ion
    d_dms = np.linalg.norm(xdms0, axis=1) # for DM

    ## Plots 
    if(show_plots):

        # -------------------------
        # Ion coordinates vs time
        # -------------------------
        plt.figure(figsize=(18, 5))

        plt.subplot(1, 3, 1)
        plt.plot(t0, x0_ion1, label="x ion")
        plt.xlabel("t (us)")
        plt.ylabel("x (um)")
        plt.title("Ion x-position vs time")
        plt.legend()

        plt.subplot(1, 3, 2)
        plt.plot(t0, y0_ion1, label="y ion")
        plt.xlabel("t (us)")
        plt.ylabel("y (um)")
        plt.title("Ion y-position vs time")
        plt.legend()

        plt.subplot(1, 3, 3)
        plt.plot(t0, z0_ion1, label="z ion")
        plt.xlabel("t (us)")
        plt.ylabel("z (um)")
        plt.title("Ion z-position vs time")
        plt.legend()

        plt.tight_layout()
        plt.show()

        # print for single DM
        if(np.array(x0_dms).shape==(1,3)):
            fig = plt.figure(figsize=(18,5))
            x0,y0,z0 = xdms0[0][0],xdms0[0][1],xdms0[0][2]

            ax = fig.add_subplot(projection='3d')
            ax.set_xlabel('x (um)')
            ax.set_ylabel('y (um)')
            ax.set_zlabel('z (um)')
            ax.plot3D(x0,y0,z0,'red')
            ax.scatter(0, 0, 0, color='blue', s=50) 
            plt.title("DM Trajectory")
            plt.legend()

            plt.figure(figsize=(18,5))
            plt.plot(t0,d_dms[0])
            plt.xlabel("t (us)")
            plt.ylabel("distance (um)")
            plt.title("DM distance vs time")

            plt.tight_layout()
            plt.show()

    return t00, xion00, vion00, xdms0, vdms0, d_dms
```

```python
def sample_maxwell_boltzmann(n, T, m, plot, seed, bins=100):
    """
    Sample n velocities from a Maxwell-Boltzmann distribution at temperature T for particles of mass m.

    Parameters
    ----------
    n : int
        Number of samples to draw.
    T : float
        Temperature of the distribution (Kelvin).
    m : float
        Mass of the particles.
    seed : int
        Seed for random number generator.

    Returns
    -------
    v : array_like
        Sampled velocities.
    """
    k_B = 1.380649e-23  # Boltzmann constant in J/K
    sigma = np.sqrt(k_B * T / m)
    rng = np.random.default_rng(seed)
    velocities = rng.normal(0, sigma, size=(n, 3))
    speeds = np.linalg.norm(velocities, axis=1)

    if plot:
        v_max = max(np.max(speeds), 5 * sigma)

        bin_edges = np.linspace(0.0, v_max, bins + 1)
        bin_width = bin_edges[1] - bin_edges[0]

        v_grid = np.linspace(0.0, v_max, 1000)

        mb_pdf = (
            4.0 * np.pi * v_grid**2
            * (m / (2.0 * np.pi * k_B * T))**1.5
            * np.exp(-m * v_grid**2 / (2.0 * k_B * T))
        )

        # Convert probability density to approximate probability per bin.
        mb_prob_per_bin = mb_pdf * bin_width

        weights = np.ones_like(speeds) / speeds.size

        plt.figure(figsize=(7, 5))

        plt.hist(
            speeds,
            bins=bin_edges,
            weights=weights,
            alpha=0.5,
            label="Sampled fraction per bin"
        )

        plt.plot(
            v_grid,
            mb_prob_per_bin,
            linewidth=2,
            label="Maxwell-Boltzmann prediction per bin"
        )

        plt.xlabel("Speed [m/s]")
        plt.ylabel("Proportion of particles")
        plt.title("Maxwell-Boltzmann speed distribution")
        plt.legend()
        plt.tight_layout()
        plt.show()
    
    return velocities, speeds
```

```python
def make_trap_anisotropic_trajectory(theta, alpha, psi, b, speed, R_start):
    """
    One-DM initial condition for an anisotropic trap.

    Trap convention:
        x = axial direction
        y,z = radial directions, with omega_y != omega_z

    Parameters
    ----------
    theta : float
        Polar angle from the axial x-axis.

    alpha : float
        Azimuthal angle around the x-axis.
        alpha = 0 means radial projection along y.
        alpha = pi/2 means radial projection along z.

    psi : float
        Impact-parameter angle in the plane perpendicular to u_hat.

    b : float
        Impact parameter magnitude.

    speed : float
        DM speed.

    R_start : float
        Starting distance upstream along -u_hat.

    Returns
    -------
    r0_dm : ndarray, shape (3,)
        Initial DM position.

    v0_dm : ndarray, shape (3,)
        Initial DM velocity.

    b_vec : ndarray, shape (3,)
        Impact vector.

    u_hat : ndarray, shape (3,)
        Incoming velocity direction.
    """
    u_hat = np.array([
        np.cos(theta),
        np.sin(theta) * np.cos(alpha),
        np.sin(theta) * np.sin(alpha)
    ], dtype=float)

    e_theta = np.array([
        -np.sin(theta),
        np.cos(theta) * np.cos(alpha),
        np.cos(theta) * np.sin(alpha)
    ], dtype=float)

    e_alpha = np.array([
        0.0,
        -np.sin(alpha),
        np.cos(alpha)
    ], dtype=float)

    u_hat = u_hat / np.linalg.norm(u_hat)
    e_theta = e_theta / np.linalg.norm(e_theta)
    e_alpha = e_alpha / np.linalg.norm(e_alpha)

    b_vec = b * (np.cos(psi) * e_theta + np.sin(psi) * e_alpha)

    r0_dm = b_vec - R_start * u_hat
    v0_dm = speed * u_hat

    return r0_dm, v0_dm, b_vec, u_hat, e_theta, e_alpha
```

## Staged/Optimized

```python
#!/usr/bin/env python3
"""Three-stage single-DM / single-ion trajectory simulation.

This module combines the handoff logic used by the Test 1 and Test 2 scripts:

1. Test 1 outer-tail straight-line diagnostic
   The nominal incoming line is sampled between ``R_full`` and ``R_max``.
   The smallest safe outer-tail handoff point is called ``R_far``.  The DM
   speed at that point is corrected by conservation of energy in the static
   trap potential.

2. Test 2 DM-only trap propagation
   The DM is propagated from the Test 1 ``R_far`` state until it enters the
   sphere ``R_switch``.  The ion is not included in this stage.

3. Full ion-DM coupled propagation and Fourier quadratures
   At the incoming ``R_switch`` crossing, the ion is introduced and the full
   ion-trap + DM-trap + ion-DM Coulomb system is integrated until the pair
   escapes the local interaction region or the integration times out.

4. Weak incoming and outgoing Fourier tails
   The Coulomb-force quadratures are completed coherently from incoming
   ``R_outer`` to outgoing ``R_outer``.  The canonical final ion energy is
   ``sum_j |int F_j(t) exp(i omega_j t) dt|^2 / (2 m_ion)``.

The main public entry point is ``single_simulation``.  With ``plot=True``
the module also displays ion motion in the xy, yz and xz planes.  The
printed summary reports the three handoff radii, their reach times, and
the initial/final ion and DM phase-space states.

Required trap-model objects
---------------------------
The easiest notebook workflow is to run the trap-model cells first and then:

    %run -i modified_staged_single_simulation.py

The following names are then resolved from the notebook namespace:

    potential_energy, force, FDC, FRF,
    m_ion, Z_ion, omega_vec, K, e

Alternatively, construct a ``SimulationEnvironment`` and pass it through the
``environment=`` argument.

Important convention
--------------------
The incoming direction is

    u = [cos(theta), sin(theta) cos(alpha), sin(theta) sin(alpha)]

and the nominal incoming line is

    r(s) = b_vec - s u,

with velocity directed along ``+u``.  Consequently, the initial radial
velocity is negative for an incoming trajectory.
"""

from __future__ import annotations

from dataclasses import dataclass
import math
import time
from typing import Any, Callable

import numpy as np
from scipy.integrate import quad, solve_ivp


Array = np.ndarray


# =============================================================================
# CONFIGURATION DATA CLASSES
# =============================================================================


@dataclass
class SimulationEnvironment:
    """Trap-model functions and constants required by the staged solver."""

    potential_energy: Callable[..., Any]
    force: Callable[..., Any] | None
    FDC: Callable[..., Any]
    FRF: Callable[..., Any]
    m_ion: float
    Z_ion: float
    omega_vec: Array
    K: float
    e: float
    ion_height_m: float | None = None

    @classmethod
    def from_namespace(cls, namespace: dict[str, Any]) -> "SimulationEnvironment":
        required = (
            "potential_energy",
            "FDC",
            "FRF",
            "m_ion",
            "Z_ion",
            "omega_vec",
            "K",
            "e",
        )
        missing = [name for name in required if name not in namespace]
        if missing:
            raise RuntimeError(
                "Missing trap-model names: " + ", ".join(missing)
            )

        ion_height_m = None
        c_obj = namespace.get("c")
        if c_obj is not None and hasattr(c_obj, "ion_height"):
            ion_height_m = float(c_obj.ion_height)

        return cls(
            potential_energy=namespace["potential_energy"],
            force=namespace.get("force"),
            FDC=namespace["FDC"],
            FRF=namespace["FRF"],
            m_ion=float(namespace["m_ion"]),
            Z_ion=float(namespace["Z_ion"]),
            omega_vec=np.asarray(namespace["omega_vec"], dtype=float).reshape(3),
            K=float(namespace["K"]),
            e=float(namespace["e"]),
            ion_height_m=ion_height_m,
        )


@dataclass
class OuterTailSettings:
    R_max_m: float = 20.0e-3
    n_path: int = 800
    energy_tol: float = 1.0e-2
    angle_tol: float = 1.0e-3
    displacement_tol_m: float = 1.0e-6
    U_reference_J: float = 0.0
    z_margin_m: float = 1.0e-6
    allow_R_far_lower_bound: bool = False


@dataclass
class DMOnlySettings:
    rtol: float = 1.0e-6
    atol: float = 1.0e-9
    max_step_s: float = 5.0e-7
    time_factor: float = 3.0
    minimum_after_arrival_s: float = 20.0e-6
    escape_factor_from_R_far: float = 1.05
    local_escape_factor_from_R_switch: float = 1.5


@dataclass
class FullSettings:
    rtol: float = 1.0e-6
    atol: float = 1.0e-9
    sample_dt_s: float = 5.0e-9
    max_step_s: float = 5.0e-9
    time_factor: float = 8.0
    minimum_time_s: float = 20.0e-6
    escape_factor_from_R_switch: float = 1.5
    energy_average_window_s: float = 2.0e-6
    use_harmonic_ion: bool = True
    coulomb_softening_m: float = 0.0


@dataclass
class FourierSettings:
    """Controls the full-pulse secular-mode Fourier-energy calculation.

    The Fourier time origin is the incoming R_switch crossing.  The complete
    force history is assembled coherently from

        incoming R_outer -> incoming R_far,
        incoming R_far -> incoming R_switch,
        the fully coupled central encounter,
        outgoing coupled escape -> outgoing R_far,
        outgoing R_far -> outgoing R_outer.

    ``outer_radius_m`` defaults to the straight-stage ``R_max`` supplied to
    ``single_simulation``.
    """

    outer_radius_m: float | None = None
    require_complete_tails: bool = True
    steps_per_secular_period: int = 20

    tail_rtol: float = 1.0e-6
    tail_atol: float = 1.0e-9
    tail_max_step_s: float = 5.0e-7
    tail_time_factor: float = 4.0
    tail_minimum_time_s: float = 50.0e-6

    quad_epsrel: float = 1.0e-8
    quad_epsabs_dimensionless: float = 1.0e-11
    quad_limit: int = 1000
    force_scale_samples: int = 21


# =============================================================================
# GENERAL HELPERS
# =============================================================================


def _resolve_environment(
    environment: SimulationEnvironment | None,
) -> SimulationEnvironment:
    if environment is not None:
        return environment
    return SimulationEnvironment.from_namespace(globals())


def _as_vector(value: Any, name: str) -> Array:
    vector = np.asarray(value, dtype=float).reshape(-1)
    if vector.shape != (3,) or not np.all(np.isfinite(vector)):
        raise ValueError(f"{name} must be a finite length-3 vector")
    return vector


def radial_velocity(position: Array, velocity: Array) -> float:
    position = _as_vector(position, "position")
    velocity = _as_vector(velocity, "velocity")
    radius = float(np.linalg.norm(position))
    if radius <= 0.0:
        return 0.0
    return float(np.dot(position, velocity) / radius)


def _append_event_state(
    solution: Any,
    event_index: int,
) -> tuple[Array, Array]:
    times = np.asarray(solution.t, dtype=float)
    states = np.asarray(solution.y, dtype=float)
    if (
        solution.t_events is None
        or len(solution.t_events) <= event_index
        or len(solution.t_events[event_index]) == 0
    ):
        return times, states

    event_time = float(solution.t_events[event_index][0])
    event_state = np.asarray(solution.y_events[event_index][0], dtype=float)
    if times.size == 0 or not np.isclose(times[-1], event_time):
        times = np.append(times, event_time)
        states = np.column_stack((states, event_state))
    return times, states


def _sampled_min_radius(states: Array, position_slice: slice) -> float:
    if states.size == 0:
        return float("nan")
    radius = np.linalg.norm(np.asarray(states[position_slice], dtype=float), axis=0)
    finite = radius[np.isfinite(radius)]
    return float(np.min(finite)) if finite.size else float("nan")


def _time_inside(times: Array, mask: Array) -> float:
    times = np.asarray(times, dtype=float)
    mask = np.asarray(mask, dtype=bool)
    if times.size < 2:
        return 0.0
    dt = np.diff(times)
    active = mask[:-1] | mask[1:]
    return float(np.sum(dt[active]))


def _format_vector(vector: Any, *, scale: float = 1.0, precision: int = 6) -> str:
    values = np.asarray(vector, dtype=float).reshape(3) * float(scale)
    return np.array2string(
        values,
        precision=precision,
        suppress_small=False,
        separator=", ",
    )


def _first_event_time(solution: Any, event_index: int) -> float:
    if (
        solution.t_events is None
        or len(solution.t_events) <= event_index
        or len(solution.t_events[event_index]) == 0
    ):
        return float("nan")
    return float(solution.t_events[event_index][0])


# =============================================================================
# TRAJECTORY GEOMETRY
# =============================================================================


def direction_basis(
    theta: float,
    alpha: float,
    psi: float,
) -> tuple[Array, Array, Array, Array]:
    """Return incoming direction and the perpendicular impact basis."""

    u_hat = np.array(
        [
            math.cos(theta),
            math.sin(theta) * math.cos(alpha),
            math.sin(theta) * math.sin(alpha),
        ],
        dtype=float,
    )
    e_theta = np.array(
        [
            -math.sin(theta),
            math.cos(theta) * math.cos(alpha),
            math.cos(theta) * math.sin(alpha),
        ],
        dtype=float,
    )
    e_alpha = np.array(
        [0.0, -math.sin(alpha), math.cos(alpha)],
        dtype=float,
    )

    u_hat /= np.linalg.norm(u_hat)
    e_theta /= np.linalg.norm(e_theta)
    e_alpha /= np.linalg.norm(e_alpha)

    b_hat = math.cos(psi) * e_theta + math.sin(psi) * e_alpha
    b_hat /= np.linalg.norm(b_hat)
    return u_hat, b_hat, e_theta, e_alpha


def build_straight_path(
    *,
    theta: float,
    alpha: float,
    psi: float,
    b_m: float,
    R_inner_m: float,
    R_outer_m: float,
    n_path: int,
) -> dict[str, Array | float]:
    """Construct the Test 1 nominal incoming line.

    ``s`` increases outward.  The particle moves inward along ``+u_hat``.
    For ``b < R_inner``, the innermost point is the line's intersection with
    the ``R_inner`` sphere.  If the line misses that sphere, the diagnostic is
    extended close to nominal closest approach, matching the Test 1 logic.
    """

    b_m = float(b_m)
    R_inner_m = float(R_inner_m)
    R_outer_m = float(R_outer_m)
    n_path = int(n_path)

    if b_m < 0.0:
        raise ValueError("b_m must be nonnegative")
    if R_inner_m <= 0.0:
        raise ValueError("R_inner_m must be positive")
    if R_outer_m <= R_inner_m:
        raise ValueError("R_outer_m must exceed R_inner_m")
    if b_m >= R_outer_m:
        raise ValueError("b_m must be smaller than R_outer_m")
    if n_path < 3:
        raise ValueError("n_path must be at least 3")

    u_hat, b_hat, e_theta, e_alpha = direction_basis(theta, alpha, psi)
    b_vec = b_m * b_hat

    s_outer = math.sqrt(max(R_outer_m**2 - b_m**2, 0.0))
    if b_m < R_inner_m:
        s_inner = math.sqrt(max(R_inner_m**2 - b_m**2, 0.0))
    else:
        s_inner = max(1.0e-9, 0.02 * R_inner_m)

    if not (s_outer > s_inner > 0.0):
        raise ValueError(
            "Invalid straight-path coordinate interval: "
            f"s_inner={s_inner:.6e}, s_outer={s_outer:.6e}, b={b_m:.6e}"
        )

    s_grid = np.geomspace(s_inner, s_outer, n_path)
    xyz = b_vec[None, :] - s_grid[:, None] * u_hat[None, :]
    return {
        "s_grid_m": s_grid,
        "xyz_m": xyz,
        "u_hat": u_hat,
        "b_hat": b_hat,
        "b_vec_m": b_vec,
        "e_theta": e_theta,
        "e_alpha": e_alpha,
    }


# =============================================================================
# TRAP MODEL ADAPTERS
# =============================================================================


def _valid_xyz_mask(
    xyz: Array,
    *,
    invalid_z_floor_m: float | None,
) -> Array:
    xyz = np.asarray(xyz, dtype=float).reshape(-1, 3)
    finite = np.all(np.isfinite(xyz), axis=1)
    if invalid_z_floor_m is None:
        return finite
    return finite & (xyz[:, 2] > float(invalid_z_floor_m))


def _evaluate_potential(
    points: Array,
    *,
    m_dm: float,
    eps: float,
    environment: SimulationEnvironment,
) -> Array:
    points = np.asarray(points, dtype=float).reshape(-1, 3)
    try:
        values = np.asarray(
            environment.potential_energy(
                points,
                m_dm,
                eps,
                component="trap",
            ),
            dtype=float,
        ).reshape(-1)
        if values.shape[0] != points.shape[0]:
            raise ValueError("Unexpected vectorized potential shape")
        return values
    except Exception:
        return np.array(
            [
                float(
                    np.asarray(
                        environment.potential_energy(
                            point,
                            m_dm,
                            eps,
                            component="trap",
                        )
                    ).reshape(-1)[0]
                )
                for point in points
            ],
            dtype=float,
        )


def _trap_force_from_FDC_FRF(
    points: Array,
    *,
    mass: float,
    charge: float,
    environment: SimulationEnvironment,
) -> Array:
    points = np.asarray(points, dtype=float).reshape(-1, 3)
    return (
        np.asarray(environment.FDC(points, mass, charge), dtype=float).reshape(-1, 3)
        + np.asarray(environment.FRF(points, mass, charge), dtype=float).reshape(-1, 3)
    )


def _evaluate_force(
    points: Array,
    *,
    mass: float,
    charge: float,
    environment: SimulationEnvironment,
) -> Array:
    points = np.asarray(points, dtype=float).reshape(-1, 3)

    if environment.force is not None:
        try:
            values = np.asarray(
                environment.force(
                    points,
                    mass,
                    charge,
                    component="trap",
                ),
                dtype=float,
            ).reshape(-1, 3)
            if values.shape[0] != points.shape[0]:
                raise ValueError("Unexpected vectorized force shape")
            return values
        except Exception:
            try:
                return np.vstack(
                    [
                        np.asarray(
                            environment.force(
                                point,
                                mass,
                                charge,
                                component="trap",
                            ),
                            dtype=float,
                        ).reshape(3)
                        for point in points
                    ]
                )
            except Exception:
                pass

    return _trap_force_from_FDC_FRF(
        points,
        mass=mass,
        charge=charge,
        environment=environment,
    )


def evaluate_straight_trap_path(
    xyz_m: Array,
    *,
    m_dm: float,
    eps: float,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
) -> tuple[Array, Array, Array]:
    """Evaluate trap-only potential and force along the nominal line."""

    xyz_m = np.asarray(xyz_m, dtype=float).reshape(-1, 3)
    n_points = xyz_m.shape[0]
    potential = np.full(n_points, np.nan, dtype=float)
    force = np.full((n_points, 3), np.nan, dtype=float)

    geometry_valid = _valid_xyz_mask(
        xyz_m,
        invalid_z_floor_m=invalid_z_floor_m,
    )
    indices = np.flatnonzero(geometry_valid)
    if indices.size == 0:
        return potential, force, np.zeros(n_points, dtype=bool)

    valid_points = xyz_m[indices]
    potential[indices] = _evaluate_potential(
        valid_points,
        m_dm=m_dm,
        eps=eps,
        environment=environment,
    )
    force[indices] = _evaluate_force(
        valid_points,
        mass=m_dm,
        charge=eps,
        environment=environment,
    )

    valid = (
        geometry_valid
        & np.isfinite(potential)
        & np.all(np.isfinite(force), axis=1)
    )
    return potential, force, valid


# =============================================================================
# TEST 1 OUTER-TAIL R_FAR DIAGNOSTIC
# =============================================================================


def _reverse_cumulative_sum(segment_values: Array) -> Array:
    values = np.asarray(segment_values)
    if values.ndim == 1:
        output = np.zeros(values.shape[0] + 1, dtype=float)
        output[:-1] = np.cumsum(values[::-1])[::-1]
        return output

    output = np.zeros((values.shape[0] + 1,) + values.shape[1:], dtype=float)
    output[:-1] = np.cumsum(values[::-1], axis=0)[::-1]
    return output


def analyze_R_far(
    *,
    s_grid_m: Array,
    xyz_m: Array,
    u_hat: Array,
    potential_J: Array,
    force_N: Array,
    valid: Array,
    m_dm: float,
    v_inf_m_s: float,
    R_full_m: float,
    R_max_m: float,
    energy_tol: float,
    angle_tol: float,
    displacement_tol_m: float,
    U_reference_J: float = 0.0,
) -> dict[str, Any]:
    """Find the smallest Test 1 handoff point with a negligible outer tail."""

    s_grid = np.asarray(s_grid_m, dtype=float)
    xyz = np.asarray(xyz_m, dtype=float)
    u = _as_vector(u_hat, "u_hat")
    potential = np.asarray(potential_J, dtype=float)
    force = np.asarray(force_N, dtype=float)
    valid = np.asarray(valid, dtype=bool)

    kinetic_inf = 0.5 * float(m_dm) * float(v_inf_m_s) ** 2
    if not np.isfinite(kinetic_inf) or kinetic_inf <= 0.0:
        return {
            "outer_status": "invalid_speed",
            "R_far_m": np.nan,
            "R_far_actual_m": np.nan,
            "R_far_is_lower_bound": False,
            "v_far_m_s": np.nan,
            "limiting_criterion": "not_applicable",
        }

    if xyz.shape[0] == 0 or not np.any(valid):
        return {
            "outer_status": "no_valid_points",
            "R_far_m": np.nan,
            "R_far_actual_m": np.nan,
            "R_far_is_lower_bound": False,
            "v_far_m_s": np.nan,
            "limiting_criterion": "not_applicable",
        }
    if not valid[-1]:
        return {
            "outer_status": "invalid_far_start_below_electrode",
            "R_far_m": np.nan,
            "R_far_actual_m": np.nan,
            "R_far_is_lower_bound": False,
            "v_far_m_s": np.nan,
            "limiting_criterion": "not_applicable",
            "xyz_far_m": xyz[-1].copy(),
        }

    kinetic_local = kinetic_inf + float(U_reference_J) - potential
    point_ok = valid & np.isfinite(kinetic_local) & (kinetic_local > 0.0)

    local_speed_sq = np.full(s_grid.shape[0], np.nan, dtype=float)
    local_speed_sq[point_ok] = 2.0 * kinetic_local[point_ok] / float(m_dm)

    force_parallel_scalar = force @ u
    force_parallel = force_parallel_scalar[:, None] * u[None, :]
    force_perpendicular = force - force_parallel

    q_vector = np.zeros_like(force_perpendicular)
    q_vector[point_ok] = force_perpendicular[point_ok] / (
        float(m_dm) * local_speed_sq[point_ok, None]
    )
    q_abs = np.linalg.norm(q_vector, axis=1)

    ds = np.diff(s_grid)
    s_mid = 0.5 * (s_grid[:-1] + s_grid[1:])
    segment_ok = point_ok[:-1] & point_ok[1:]

    q_mid_vector = 0.5 * (q_vector[:-1] + q_vector[1:])
    q_mid_abs = 0.5 * (q_abs[:-1] + q_abs[1:])
    q_mid_vector[~segment_ok] = 0.0
    q_mid_abs[~segment_ok] = 0.0

    theta_vector_tail = _reverse_cumulative_sum(q_mid_vector * ds[:, None])
    theta_signed_tail = np.linalg.norm(theta_vector_tail, axis=1)
    theta_abs_tail = _reverse_cumulative_sum(q_mid_abs * ds)

    moment_vector_tail = _reverse_cumulative_sum(
        s_mid[:, None] * q_mid_vector * ds[:, None]
    )
    displacement_vector_tail = (
        moment_vector_tail - s_grid[:, None] * theta_vector_tail
    )
    displacement_signed_tail = np.linalg.norm(displacement_vector_tail, axis=1)

    moment_abs_tail = _reverse_cumulative_sum(s_mid * q_mid_abs * ds)
    displacement_abs_tail = np.maximum(
        moment_abs_tail - s_grid * theta_abs_tail,
        0.0,
    )

    valid_tail = np.logical_and.accumulate(valid[::-1])[::-1]
    kinetic_for_tail = np.where(valid, kinetic_local, -np.inf)
    kinetic_tail_min = np.minimum.accumulate(kinetic_for_tail[::-1])[::-1]

    energy_ratio_point = np.where(
        valid,
        np.abs(potential - float(U_reference_J)) / kinetic_inf,
        np.inf,
    )
    energy_ratio_tail_max = np.maximum.accumulate(
        energy_ratio_point[::-1]
    )[::-1]

    safe = (
        valid_tail
        & (kinetic_tail_min > 0.0)
        & (energy_ratio_tail_max <= float(energy_tol))
        & (theta_abs_tail <= float(angle_tol))
        & (displacement_abs_tail <= float(displacement_tol_m))
    )
    safe_indices = np.flatnonzero(safe)

    nonpositive = np.flatnonzero(
        valid & np.isfinite(kinetic_local) & (kinetic_local <= 0.0)
    )
    turning_radius_m = (
        float(np.max(s_grid[nonpositive])) if nonpositive.size else np.nan
    )

    if safe_indices.size == 0:
        limiting = (
            "R_max_nonpositive_K" if not point_ok[-1] else "R_max_energy"
        )
        v_at_max = (
            math.sqrt(2.0 * kinetic_local[-1] / float(m_dm))
            if point_ok[-1]
            else np.nan
        )
        return {
            "outer_status": "needs_larger_R_max",
            "R_far_m": float(R_max_m),
            "R_far_actual_m": float(np.linalg.norm(xyz[-1])),
            "R_far_is_lower_bound": True,
            "v_far_m_s": v_at_max,
            "U_far_J": float(potential[-1]),
            "K_inf_J": kinetic_inf,
            "energy_tail_max": (
                float(energy_ratio_tail_max[-1])
                if np.isfinite(energy_ratio_tail_max[-1])
                else np.nan
            ),
            "theta_signed_tail": 0.0,
            "theta_abs_tail": 0.0,
            "displacement_signed_tail_m": 0.0,
            "displacement_abs_tail_m": 0.0,
            "limiting_criterion": limiting,
            "turning_radius_m": turning_radius_m,
            "xyz_far_m": xyz[-1].copy(),
            "velocity_far_m_s": (
                v_at_max * u if np.isfinite(v_at_max) else np.full(3, np.nan)
            ),
            "safe_mask": safe,
            "energy_ratio_tail_max_array": energy_ratio_tail_max,
            "theta_abs_tail_array": theta_abs_tail,
            "displacement_abs_tail_array_m": displacement_abs_tail,
        }

    index = int(safe_indices[0])
    K_far = float(kinetic_local[index])
    if K_far <= 0.0:
        return {
            "outer_status": "nonpositive_K_at_R_far",
            "R_far_m": float(s_grid[index]),
            "R_far_actual_m": float(np.linalg.norm(xyz[index])),
            "R_far_is_lower_bound": False,
            "v_far_m_s": np.nan,
            "limiting_criterion": "not_applicable",
            "turning_radius_m": turning_radius_m,
        }

    v_far = math.sqrt(2.0 * K_far / float(m_dm))
    normalized_metrics = {
        "energy": float(energy_ratio_tail_max[index]) / float(energy_tol),
        "angle": float(theta_abs_tail[index]) / float(angle_tol),
        "displacement": (
            float(displacement_abs_tail[index]) / float(displacement_tol_m)
        ),
    }
    limiting = max(normalized_metrics, key=normalized_metrics.get)
    outer_status = (
        "straight_safe_to_full_sphere" if index == 0 else "straight_then_dm_only"
    )

    return {
        "outer_status": outer_status,
        "R_far_m": float(s_grid[index]),
        "R_far_actual_m": float(np.linalg.norm(xyz[index])),
        "R_far_is_lower_bound": False,
        "v_far_m_s": v_far,
        "U_far_J": float(potential[index]),
        "K_inf_J": kinetic_inf,
        "energy_tail_max": float(energy_ratio_tail_max[index]),
        "theta_signed_tail": float(theta_signed_tail[index]),
        "theta_abs_tail": float(theta_abs_tail[index]),
        "displacement_signed_tail_m": float(displacement_signed_tail[index]),
        "displacement_abs_tail_m": float(displacement_abs_tail[index]),
        "limiting_criterion": limiting,
        "turning_radius_m": turning_radius_m,
        "xyz_far_m": xyz[index].copy(),
        "velocity_far_m_s": v_far * u,
        "safe_index": index,
        "safe_mask": safe,
        "energy_ratio_tail_max_array": energy_ratio_tail_max,
        "theta_abs_tail_array": theta_abs_tail,
        "displacement_abs_tail_array_m": displacement_abs_tail,
    }


def run_straight_outer_tail_test(
    *,
    theta: float,
    alpha: float,
    psi: float,
    b_m: float,
    v_inf_m_s: float,
    m_dm: float,
    eps: float,
    R_full_m: float,
    settings: OuterTailSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
) -> dict[str, Any]:
    """Run the complete Test 1 straight-line outer-tail stage."""

    geometry = build_straight_path(
        theta=theta,
        alpha=alpha,
        psi=psi,
        b_m=b_m,
        R_inner_m=R_full_m,
        R_outer_m=settings.R_max_m,
        n_path=settings.n_path,
    )
    potential, force, valid = evaluate_straight_trap_path(
        geometry["xyz_m"],
        m_dm=m_dm,
        eps=eps,
        environment=environment,
        invalid_z_floor_m=invalid_z_floor_m,
    )
    outer = analyze_R_far(
        s_grid_m=geometry["s_grid_m"],
        xyz_m=geometry["xyz_m"],
        u_hat=geometry["u_hat"],
        potential_J=potential,
        force_N=force,
        valid=valid,
        m_dm=m_dm,
        v_inf_m_s=v_inf_m_s,
        R_full_m=R_full_m,
        R_max_m=settings.R_max_m,
        energy_tol=settings.energy_tol,
        angle_tol=settings.angle_tol,
        displacement_tol_m=settings.displacement_tol_m,
        U_reference_J=settings.U_reference_J,
    )
    outer.update(
        {
            "s_grid_m": geometry["s_grid_m"],
            "xyz_path_m": geometry["xyz_m"],
            "u_hat": geometry["u_hat"],
            "b_hat": geometry["b_hat"],
            "b_vec_m": geometry["b_vec_m"],
            "potential_path_J": potential,
            "force_path_N": force,
            "valid_path": valid,
        }
    )
    return outer


# =============================================================================
# TEST 2 DM-ONLY PROPAGATION TO R_SWITCH
# =============================================================================


def _dm_only_rhs(
    *,
    m_dm: float,
    eps: float,
    environment: SimulationEnvironment,
) -> Callable[[float, Array], Array]:
    def rhs(_time: float, state: Array) -> Array:
        position = state[:3]
        velocity = state[3:]
        force = _trap_force_from_FDC_FRF(
            position[None, :],
            mass=m_dm,
            charge=eps,
            environment=environment,
        )[0]
        acceleration = force / float(m_dm)
        return np.concatenate((velocity, acceleration))

    return rhs


def _reconstruct_incoming_switch_state(
    x_inside_m: Array,
    v_inside_m_s: Array,
    *,
    m_dm: float,
    eps: float,
    R_switch_m: float,
    invalid_z_floor_m: float | None,
    settings: DMOnlySettings,
    environment: SimulationEnvironment,
) -> dict[str, Any]:
    """Integrate backward from an inside point to the incoming switch crossing."""

    speed = max(float(np.linalg.norm(v_inside_m_s)), 1.0e-12)
    time_scale = max(
        settings.time_factor * R_switch_m / speed,
        settings.minimum_after_arrival_s,
    )

    def switch_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[:3]) - R_switch_m)

    switch_event.terminal = True
    switch_event.direction = 0
    events: list[Callable] = [switch_event]

    if invalid_z_floor_m is not None:
        def invalid_event(_time: float, state: Array) -> float:
            return float(state[2] - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = 0
        events.append(invalid_event)

    solution = solve_ivp(
        _dm_only_rhs(m_dm=m_dm, eps=eps, environment=environment),
        (0.0, -time_scale),
        np.concatenate((x_inside_m, v_inside_m_s)),
        method="DOP853",
        rtol=settings.rtol,
        atol=settings.atol,
        max_step=settings.max_step_s,
        events=events,
    )
    entered = len(solution.t_events[0]) > 0
    switch_time_relative = _first_event_time(solution, 0)
    if not entered:
        return {
            "success": False,
            "status": "dm_only_inside_requires_rebuild",
            "message": str(solution.message),
            "x_switch_m": np.full(3, np.nan),
            "v_switch_m_s": np.full(3, np.nan),
            "switch_time_relative_s": float("nan"),
            "solution": solution,
        }

    state = np.asarray(solution.y_events[0][0], dtype=float)
    return {
        "success": True,
        "status": "dm_only_reconstructed_incoming_switch",
        "message": str(solution.message),
        "x_switch_m": state[:3],
        "v_switch_m_s": state[3:],
        "switch_time_relative_s": switch_time_relative,
        "solution": solution,
    }


def run_dm_trap_only_to_switch(
    x0_m: Array,
    v0_m_s: Array,
    *,
    m_dm: float,
    eps: float,
    R_switch_m: float,
    settings: DMOnlySettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
    keep_trajectory: bool = True,
) -> dict[str, Any]:
    """Propagate the DM under trap forces alone to the incoming switch sphere."""

    started = time.perf_counter()
    x0 = _as_vector(x0_m, "x0_m")
    v0 = _as_vector(v0_m_s, "v0_m_s")
    R_switch_m = float(R_switch_m)
    r0 = float(np.linalg.norm(x0))
    speed0 = max(float(np.linalg.norm(v0)), 1.0e-12)
    initial_radial_velocity = radial_velocity(x0, v0)

    if r0 < R_switch_m * (1.0 - 1.0e-10):
        rebuilt = _reconstruct_incoming_switch_state(
            x0,
            v0,
            m_dm=m_dm,
            eps=eps,
            R_switch_m=R_switch_m,
            invalid_z_floor_m=invalid_z_floor_m,
            settings=settings,
            environment=environment,
        )
        solution = rebuilt.pop("solution")
        switch_time_relative = float(rebuilt.pop("switch_time_relative_s", np.nan))
        x_switch = np.asarray(rebuilt["x_switch_m"], dtype=float)
        v_switch = np.asarray(rebuilt["v_switch_m_s"], dtype=float)
        result = {
            "dm_only_success": bool(rebuilt["success"]),
            "dm_only_status": str(rebuilt["status"]),
            "dm_only_message": str(rebuilt["message"]),
            "entered_switch": bool(rebuilt["success"]),
            "dm_only_runtime_s": time.perf_counter() - started,
            "dm_only_t_final_s": float(solution.t[-1]) if solution.t.size else 0.0,
            "dm_only_t_reach_R_switch_s": switch_time_relative,
            "dm_only_t_arrival_estimate_s": r0 / speed0,
            "dm_only_t_max_s": abs(float(solution.t[-1])) if solution.t.size else np.nan,
            "dm_only_initial_radius_m": r0,
            "dm_only_initial_radial_velocity_m_s": initial_radial_velocity,
            "dm_only_final_radius_m": (
                float(np.linalg.norm(solution.y[:3, -1])) if solution.y.size else r0
            ),
            "dm_only_min_radius_m": _sampled_min_radius(solution.y, slice(0, 3)),
            "dm_only_final_radial_velocity_m_s": (
                radial_velocity(solution.y[:3, -1], solution.y[3:, -1])
                if solution.y.size
                else np.nan
            ),
            "x_dm_initial_m": x0.copy(),
            "v_dm_initial_m_s": v0.copy(),
            "x_dm_final_m": (
                solution.y[:3, -1].copy() if solution.y.size else x0.copy()
            ),
            "v_dm_final_m_s": (
                solution.y[3:, -1].copy() if solution.y.size else v0.copy()
            ),
            "x_switch_m": x_switch,
            "v_switch_m_s": v_switch,
        }
        if keep_trajectory:
            result["times_s"] = np.asarray(solution.t, dtype=float)
            result["states"] = np.asarray(solution.y, dtype=float)
        return result

    if np.isclose(r0, R_switch_m, rtol=1.0e-10, atol=1.0e-15):
        return {
            "dm_only_success": True,
            "dm_only_status": "dm_only_started_on_switch",
            "dm_only_message": "Initial state lies on R_switch",
            "entered_switch": True,
            "dm_only_runtime_s": time.perf_counter() - started,
            "dm_only_t_final_s": 0.0,
            "dm_only_t_reach_R_switch_s": 0.0,
            "dm_only_t_arrival_estimate_s": 0.0,
            "dm_only_t_max_s": 0.0,
            "dm_only_initial_radius_m": r0,
            "dm_only_initial_radial_velocity_m_s": initial_radial_velocity,
            "dm_only_final_radius_m": r0,
            "dm_only_min_radius_m": r0,
            "dm_only_final_radial_velocity_m_s": initial_radial_velocity,
            "x_dm_initial_m": x0.copy(),
            "v_dm_initial_m_s": v0.copy(),
            "x_dm_final_m": x0.copy(),
            "v_dm_final_m_s": v0.copy(),
            "x_switch_m": x0.copy(),
            "v_switch_m_s": v0.copy(),
            "times_s": np.array([0.0]),
            "states": np.concatenate((x0, v0))[:, None],
        }

    arrival_estimate = r0 / speed0
    t_max = max(
        settings.time_factor * arrival_estimate,
        arrival_estimate + settings.minimum_after_arrival_s,
    )
    escape_radius = max(
        settings.escape_factor_from_R_far * r0,
        settings.local_escape_factor_from_R_switch * R_switch_m,
    )

    def switch_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[:3]) - R_switch_m)

    switch_event.terminal = True
    switch_event.direction = -1

    def escape_event(_time: float, state: Array) -> float:
        radius = float(np.linalg.norm(state[:3]))
        if radial_velocity(state[:3], state[3:]) <= 0.0:
            return -abs(radius - escape_radius) - 1.0e-30
        return radius - escape_radius

    escape_event.terminal = True
    escape_event.direction = 1

    def closest_approach_event(_time: float, state: Array) -> float:
        return radial_velocity(state[:3], state[3:])

    closest_approach_event.terminal = False
    closest_approach_event.direction = 1
    events: list[Callable] = [switch_event, escape_event, closest_approach_event]

    if invalid_z_floor_m is not None:
        def invalid_event(_time: float, state: Array) -> float:
            return float(state[2] - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = -1
        events.append(invalid_event)

    solution = solve_ivp(
        _dm_only_rhs(m_dm=m_dm, eps=eps, environment=environment),
        (0.0, t_max),
        np.concatenate((x0, v0)),
        method="DOP853",
        rtol=settings.rtol,
        atol=settings.atol,
        max_step=settings.max_step_s,
        events=events,
    )

    entered = len(solution.t_events[0]) > 0
    escaped = len(solution.t_events[1]) > 0
    closest_states = (
        np.asarray(solution.y_events[2], dtype=float)
        if len(solution.y_events) > 2 and len(solution.y_events[2])
        else np.empty((0, 6), dtype=float)
    )
    invalid_index = 3
    invalid = len(events) > invalid_index and len(solution.t_events[invalid_index]) > 0
    reconstructed_message = ""
    switch_time_s = _first_event_time(solution, 0)

    if entered:
        status = "dm_only_entered_switch"
        event_state = np.asarray(solution.y_events[0][0], dtype=float)
    elif closest_states.size:
        closest_radii = np.linalg.norm(closest_states[:, :3], axis=1)
        closest_index = int(np.argmin(closest_radii))
        closest_state = closest_states[closest_index]
        if closest_radii[closest_index] <= R_switch_m * (1.0 + 1.0e-10):
            closest_time_s = float(solution.t_events[2][closest_index])
            rebuilt = _reconstruct_incoming_switch_state(
                closest_state[:3],
                closest_state[3:],
                m_dm=m_dm,
                eps=eps,
                R_switch_m=R_switch_m,
                invalid_z_floor_m=invalid_z_floor_m,
                settings=settings,
                environment=environment,
            )
            entered = bool(rebuilt["success"])
            reconstructed_message = str(rebuilt["message"])
            if entered:
                switch_time_s = closest_time_s + float(
                    rebuilt.get("switch_time_relative_s", np.nan)
                )
                status = "dm_only_entered_switch_via_closest_approach"
                event_state = np.concatenate(
                    (rebuilt["x_switch_m"], rebuilt["v_switch_m_s"])
                )
            else:
                status = "dm_only_closest_approach_rebuild_failure"
                event_state = np.full(6, np.nan)
        elif escaped:
            status = "dm_only_escaped_without_switch"
            event_state = np.full(6, np.nan)
        elif invalid:
            status = "dm_only_invalid_domain"
            event_state = np.full(6, np.nan)
        elif not solution.success:
            status = "dm_only_solver_failure"
            event_state = np.full(6, np.nan)
        else:
            status = "dm_only_timeout"
            event_state = np.full(6, np.nan)
    elif escaped:
        status = "dm_only_escaped_without_switch"
        event_state = np.full(6, np.nan)
    elif invalid:
        status = "dm_only_invalid_domain"
        event_state = np.full(6, np.nan)
    elif not solution.success:
        status = "dm_only_solver_failure"
        event_state = np.full(6, np.nan)
    else:
        status = "dm_only_timeout"
        event_state = np.full(6, np.nan)

    event_to_append = 0 if len(solution.t_events[0]) else 1 if escaped else 0
    times, states = _append_event_state(solution, event_to_append)
    final_state = states[:, -1]
    minimum_radius = _sampled_min_radius(states, slice(0, 3))
    if closest_states.size:
        minimum_radius = min(
            minimum_radius,
            float(np.min(np.linalg.norm(closest_states[:, :3], axis=1))),
        )

    message = str(solution.message)
    if reconstructed_message:
        message += "; switch reconstructed from closest approach: " + reconstructed_message

    result = {
        "dm_only_success": bool(solution.success) and not status.endswith("failure"),
        "dm_only_status": status,
        "dm_only_message": message,
        "entered_switch": bool(entered),
        "dm_only_runtime_s": time.perf_counter() - started,
        "dm_only_t_final_s": float(times[-1]),
        "dm_only_t_reach_R_switch_s": switch_time_s,
        "dm_only_t_arrival_estimate_s": arrival_estimate,
        "dm_only_t_max_s": t_max,
        "dm_only_initial_radius_m": r0,
        "dm_only_initial_radial_velocity_m_s": initial_radial_velocity,
        "dm_only_final_radius_m": float(np.linalg.norm(final_state[:3])),
        "dm_only_min_radius_m": minimum_radius,
        "dm_only_final_radial_velocity_m_s": radial_velocity(
            final_state[:3], final_state[3:]
        ),
        "x_dm_initial_m": x0.copy(),
        "v_dm_initial_m_s": v0.copy(),
        "x_dm_final_m": final_state[:3].copy(),
        "v_dm_final_m_s": final_state[3:].copy(),
        "x_switch_m": np.asarray(event_state[:3], dtype=float),
        "v_switch_m_s": np.asarray(event_state[3:], dtype=float),
    }
    if keep_trajectory:
        result["times_s"] = times
        result["states"] = states
    return result


# =============================================================================
# FULL ION-DM COUPLED PROPAGATION
# =============================================================================


def _ion_energy_modes(
    position_m: Array,
    velocity_m_s: Array,
    *,
    settings: FullSettings,
    environment: SimulationEnvironment,
) -> Array:
    position = np.asarray(position_m, dtype=float)
    velocity = np.asarray(velocity_m_s, dtype=float)
    kinetic = 0.5 * environment.m_ion * velocity**2
    if settings.use_harmonic_ion:
        omega = environment.omega_vec.reshape(3, 1)
        return kinetic + 0.5 * environment.m_ion * (omega * position) ** 2
    return kinetic


def run_full_ion_dm_after_switch(
    x_switch_m: Array,
    v_switch_m_s: Array,
    *,
    m_dm: float,
    eps: float,
    R_full_m: float,
    R_switch_m: float,
    settings: FullSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
    x0_ion_m: Array | None = None,
    v0_ion_m_s: Array | None = None,
    target_energy_J: float = 1.0e-27,
    keep_trajectory: bool = True,
) -> dict[str, Any]:
    """Run the full single-ion / single-DM coupled ODE after switch entry."""

    started = time.perf_counter()
    x_switch = _as_vector(x_switch_m, "x_switch_m")
    v_switch = _as_vector(v_switch_m_s, "v_switch_m_s")
    x_ion0 = np.zeros(3) if x0_ion_m is None else _as_vector(x0_ion_m, "x0_ion_m")
    v_ion0 = np.zeros(3) if v0_ion_m_s is None else _as_vector(v0_ion_m_s, "v0_ion_m_s")

    initial_state = np.concatenate((x_ion0, x_switch, v_ion0, v_switch))
    softening_sq = float(settings.coulomb_softening_m) ** 2

    def rhs(_time: float, state: Array) -> Array:
        x_ion = state[0:3]
        x_dm = state[3:6]
        v_ion = state[6:9]
        v_dm = state[9:12]

        if settings.use_harmonic_ion:
            a_ion = -environment.omega_vec**2 * x_ion
        else:
            ion_force = _trap_force_from_FDC_FRF(
                x_ion[None, :],
                mass=environment.m_ion,
                charge=environment.Z_ion,
                environment=environment,
            )[0]
            a_ion = ion_force / environment.m_ion

        dm_force = _trap_force_from_FDC_FRF(
            x_dm[None, :],
            mass=m_dm,
            charge=eps,
            environment=environment,
        )[0]
        a_dm = dm_force / float(m_dm)

        delta = x_ion - x_dm
        separation_sq = max(
            float(np.dot(delta, delta)) + softening_sq,
            np.finfo(float).tiny,
        )
        coulomb_force = (
            environment.K
            * environment.Z_ion
            * float(eps)
            * environment.e**2
            * delta
            * separation_sq ** (-1.5)
        )
        a_ion = a_ion + coulomb_force / environment.m_ion
        a_dm = a_dm - coulomb_force / float(m_dm)
        return np.concatenate((v_ion, v_dm, a_ion, a_dm))

    escape_radius = float(settings.escape_factor_from_R_switch) * float(R_switch_m)

    def escape_event(_time: float, state: Array) -> float:
        relative_position = state[3:6] - state[0:3]
        relative_velocity = state[9:12] - state[6:9]
        separation = float(np.linalg.norm(relative_position))
        if radial_velocity(relative_position, relative_velocity) <= 0.0:
            return -abs(separation - escape_radius) - 1.0e-30
        return separation - escape_radius

    escape_event.terminal = True
    escape_event.direction = 1

    def R_full_entry_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[3:6] - state[0:3]) - R_full_m)

    R_full_entry_event.terminal = False
    R_full_entry_event.direction = -1
    events: list[Callable] = [escape_event, R_full_entry_event]

    if invalid_z_floor_m is not None:
        def invalid_event(_time: float, state: Array) -> float:
            return float(min(state[2], state[5]) - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = -1
        events.append(invalid_event)

    speed = max(float(np.linalg.norm(v_switch)), 1.0e-12)
    t_max = max(
        settings.time_factor * float(R_switch_m) / speed,
        settings.minimum_time_s,
    )
    t_eval = np.arange(0.0, t_max, settings.sample_dt_s, dtype=float)
    if t_eval.size == 0 or not np.isclose(t_eval[0], 0.0):
        t_eval = np.insert(t_eval, 0, 0.0)
    if not np.isclose(t_eval[-1], t_max):
        t_eval = np.append(t_eval, t_max)

    solution = solve_ivp(
        rhs,
        (0.0, t_max),
        initial_state,
        method="DOP853",
        t_eval=t_eval,
        rtol=settings.rtol,
        atol=settings.atol,
        max_step=settings.max_step_s,
        events=events,
    )

    escaped = len(solution.t_events[0]) > 0
    reached_R_full_event = len(solution.t_events[1]) > 0
    invalid_index = 2
    invalid = len(events) > invalid_index and len(solution.t_events[invalid_index]) > 0
    event_index = 0 if escaped else invalid_index if invalid else 0
    times, states = _append_event_state(solution, event_index)

    x_ion = states[0:3]
    x_dm = states[3:6]
    v_ion = states[6:9]
    v_dm = states[9:12]
    relative = x_dm - x_ion
    separations = np.linalg.norm(relative, axis=0)
    d_min = float(np.min(separations)) if separations.size else np.nan
    reached_R_full = bool(
        reached_R_full_event
        or (np.isfinite(d_min) and d_min <= float(R_full_m))
    )
    full_t_reach_R_full_s = _first_event_time(solution, 1)

    energy_modes = _ion_energy_modes(
        x_ion,
        v_ion,
        settings=settings,
        environment=environment,
    )
    total_energy = np.sum(energy_modes, axis=0)
    final_time = float(times[-1]) if times.size else 0.0
    average_mask = times >= max(
        0.0,
        final_time - settings.energy_average_window_s,
    )
    if not np.any(average_mask):
        average_mask = np.ones_like(times, dtype=bool)

    final_modes = np.mean(energy_modes[:, average_mask], axis=1)
    final_energy = float(np.sum(final_modes))
    initial_energy = float(total_energy[0]) if total_energy.size else 0.0
    delta_energy = final_energy - initial_energy

    interaction_switch = _time_inside(times, separations <= float(R_switch_m))
    interaction_full = _time_inside(times, separations <= float(R_full_m))

    if invalid:
        status = "full_invalid_domain"
    elif not solution.success:
        status = "full_solver_failure"
    elif not escaped:
        status = "full_timeout"
    elif reached_R_full:
        status = "reached_R_full_and_escaped"
    else:
        status = "entered_switch_but_missed_R_full"

    result = {
        "full_success": bool(solution.success),
        "full_status": status,
        "full_message": str(solution.message),
        "reached_R_full": reached_R_full,
        "above_energy_threshold": bool(
            status == "reached_R_full_and_escaped"
            and delta_energy >= float(target_energy_J)
        ),
        "d_min_m": d_min,
        "d_min_um": d_min * 1.0e6 if np.isfinite(d_min) else np.nan,
        "E_initial_J": initial_energy,
        "E_final_J": final_energy,
        "delta_E_ion_J": delta_energy,
        "E_x_J": float(final_modes[0]),
        "E_y_J": float(final_modes[1]),
        "E_z_J": float(final_modes[2]),
        "interaction_time_inside_R_switch_s": interaction_switch,
        "interaction_time_inside_R_full_s": interaction_full,
        "full_t_final_s": final_time,
        "full_t_reach_R_full_s": full_t_reach_R_full_s,
        "full_t_max_s": t_max,
        "full_runtime_s": time.perf_counter() - started,
        "x_ion_initial_m": x_ion0.copy(),
        "v_ion_initial_m_s": v_ion0.copy(),
        "x_dm_initial_m": x_switch.copy(),
        "v_dm_initial_m_s": v_switch.copy(),
        "x_ion_final_m": x_ion[:, -1].copy(),
        "v_ion_final_m_s": v_ion[:, -1].copy(),
        "x_dm_final_m": x_dm[:, -1].copy(),
        "v_dm_final_m_s": v_dm[:, -1].copy(),
    }
    if keep_trajectory:
        result.update(
            {
                "times_s": times,
                "states": states,
                "x_ion_m": x_ion,
                "v_ion_m_s": v_ion,
                "x_dm_m": x_dm,
                "v_dm_m_s": v_dm,
                "separation_m": separations,
                "ion_energy_modes_J": energy_modes,
                "ion_total_energy_J": total_energy,
            }
        )
    return result



# =============================================================================
# FULL-PULSE FOURIER ENERGY WITH INCOMING AND OUTGOING TAILS
# =============================================================================


def _coulomb_force_on_ion(
    ion_position_m: Array,
    dm_position_m: Array,
    *,
    eps: float,
    settings: FullSettings,
    environment: SimulationEnvironment,
) -> Array:
    """Return the instantaneous Coulomb force exerted on the ion."""

    x_ion = _as_vector(ion_position_m, "ion_position_m")
    x_dm = _as_vector(dm_position_m, "dm_position_m")
    delta = x_ion - x_dm
    radius_sq = max(
        float(np.dot(delta, delta)) + float(settings.coulomb_softening_m) ** 2,
        np.finfo(float).tiny,
    )
    return (
        environment.K
        * environment.Z_ion
        * float(eps)
        * environment.e**2
        * delta
        * radius_sq ** (-1.5)
    )


def _free_harmonic_ion_state(
    local_time_s: float,
    x0_m: Array,
    v0_m_s: Array,
    *,
    environment: SimulationEnvironment,
) -> tuple[Array, Array]:
    """Exact unforced harmonic-ion state after a tail handoff."""

    t = float(local_time_s)
    x0 = _as_vector(x0_m, "x0_m")
    v0 = _as_vector(v0_m_s, "v0_m_s")
    omega = np.asarray(environment.omega_vec, dtype=float).reshape(3)

    position = np.empty(3, dtype=float)
    velocity = np.empty(3, dtype=float)
    oscillatory = np.abs(omega) > 0.0

    phase = omega[oscillatory] * t
    position[oscillatory] = (
        x0[oscillatory] * np.cos(phase)
        + v0[oscillatory] * np.sin(phase) / omega[oscillatory]
    )
    velocity[oscillatory] = (
        -x0[oscillatory] * omega[oscillatory] * np.sin(phase)
        + v0[oscillatory] * np.cos(phase)
    )

    position[~oscillatory] = x0[~oscillatory] + v0[~oscillatory] * t
    velocity[~oscillatory] = v0[~oscillatory]
    return position, velocity


def _empty_fourier_segment(
    *,
    status: str,
    complete: bool,
    duration_s: float = 0.0,
    global_start_time_s: float = np.nan,
    start_radius_m: float = np.nan,
    end_radius_m: float = np.nan,
) -> dict[str, Any]:
    return {
        "complete": bool(complete),
        "status": str(status),
        "duration_s": float(duration_s),
        "global_start_time_s": float(global_start_time_s),
        "start_radius_m": float(start_radius_m),
        "end_radius_m": float(end_radius_m),
        "amplitude_N_s": np.zeros(3, dtype=complex),
        "quadrature_error_N_s": np.full(3, np.nan, dtype=float),
        "energy_modes_J": np.zeros(3, dtype=float),
        "energy_J": 0.0,
        "force_at_start_N": np.nan,
        "force_at_end_N": np.nan,
    }


def _integrate_fourier_force_segment(
    force_at_local_time: Callable[[float], Array],
    *,
    duration_s: float,
    global_start_time_s: float,
    status: str,
    start_radius_m: float,
    end_radius_m: float,
    fourier_settings: FourierSettings,
    environment: SimulationEnvironment,
) -> dict[str, Any]:
    """Integrate one force segment as A_j = int F_j exp(i omega_j t) dt."""

    duration = float(duration_s)
    start_time = float(global_start_time_s)
    if not np.isfinite(duration) or duration < 0.0:
        return _empty_fourier_segment(
            status="invalid_fourier_segment_duration",
            complete=False,
            duration_s=duration,
            global_start_time_s=start_time,
            start_radius_m=start_radius_m,
            end_radius_m=end_radius_m,
        )
    if duration == 0.0:
        return _empty_fourier_segment(
            status=status + "_zero_length",
            complete=True,
            duration_s=0.0,
            global_start_time_s=start_time,
            start_radius_m=start_radius_m,
            end_radius_m=end_radius_m,
        )

    n_scale = max(int(fourier_settings.force_scale_samples), 3)
    sample_times = np.linspace(0.0, duration, n_scale)
    sampled_norms: list[float] = []
    for sample_time in sample_times:
        sampled_force = _as_vector(
            force_at_local_time(float(sample_time)),
            "sampled Fourier force",
        )
        sampled_norms.append(float(np.linalg.norm(sampled_force)))

    force_scale = max(max(sampled_norms, default=0.0), np.finfo(float).tiny)
    omega = np.asarray(environment.omega_vec, dtype=float).reshape(3)
    amplitude = np.zeros(3, dtype=complex)
    absolute_error = np.zeros(3, dtype=float)

    for mode_index, angular_frequency in enumerate(omega):

        def normalized_component(unit_time: float) -> float:
            local_time = float(unit_time) * duration
            force_value = _as_vector(
                force_at_local_time(local_time),
                "Fourier force",
            )
            return float(force_value[mode_index] / force_scale)

        scaled_frequency = float(angular_frequency) * duration
        if abs(scaled_frequency) <= 1.0e-14:
            cosine_value, cosine_error = quad(
                normalized_component,
                0.0,
                1.0,
                epsabs=float(fourier_settings.quad_epsabs_dimensionless),
                epsrel=float(fourier_settings.quad_epsrel),
                limit=int(fourier_settings.quad_limit),
            )
            sine_value, sine_error = 0.0, 0.0
        else:
            cosine_value, cosine_error = quad(
                normalized_component,
                0.0,
                1.0,
                weight="cos",
                wvar=scaled_frequency,
                epsabs=float(fourier_settings.quad_epsabs_dimensionless),
                epsrel=float(fourier_settings.quad_epsrel),
                limit=int(fourier_settings.quad_limit),
            )
            sine_value, sine_error = quad(
                normalized_component,
                0.0,
                1.0,
                weight="sin",
                wvar=scaled_frequency,
                epsabs=float(fourier_settings.quad_epsabs_dimensionless),
                epsrel=float(fourier_settings.quad_epsrel),
                limit=int(fourier_settings.quad_limit),
            )

        local_amplitude = (
            complex(cosine_value, sine_value) * force_scale * duration
        )
        amplitude[mode_index] = local_amplitude * np.exp(
            1j * float(angular_frequency) * start_time
        )
        absolute_error[mode_index] = (
            math.hypot(float(cosine_error), float(sine_error))
            * force_scale
            * duration
        )

    energy_modes = np.abs(amplitude) ** 2 / (2.0 * environment.m_ion)
    return {
        "complete": True,
        "status": str(status),
        "duration_s": duration,
        "global_start_time_s": start_time,
        "start_radius_m": float(start_radius_m),
        "end_radius_m": float(end_radius_m),
        "amplitude_N_s": amplitude,
        "quadrature_error_N_s": absolute_error,
        "energy_modes_J": energy_modes,
        "energy_J": float(np.sum(energy_modes)),
        "force_at_start_N": float(
            np.linalg.norm(force_at_local_time(0.0))
        ),
        "force_at_end_N": float(
            np.linalg.norm(force_at_local_time(duration))
        ),
    }


def _flatten_fourier_segment(
    segment: dict[str, Any],
    *,
    prefix: str,
) -> dict[str, Any]:
    """Convert an internal segment dictionary into scalar result fields."""

    amplitude = np.asarray(
        segment.get("amplitude_N_s", np.zeros(3, dtype=complex)),
        dtype=complex,
    ).reshape(3)
    error = np.asarray(
        segment.get("quadrature_error_N_s", np.full(3, np.nan)),
        dtype=float,
    ).reshape(3)
    modes = np.asarray(
        segment.get("energy_modes_J", np.zeros(3)),
        dtype=float,
    ).reshape(3)

    return {
        f"{prefix}_complete": bool(segment.get("complete", False)),
        f"{prefix}_status": str(segment.get("status", "")),
        f"{prefix}_duration_s": float(segment.get("duration_s", np.nan)),
        f"{prefix}_global_start_time_s": float(
            segment.get("global_start_time_s", np.nan)
        ),
        f"{prefix}_start_radius_m": float(
            segment.get("start_radius_m", np.nan)
        ),
        f"{prefix}_end_radius_m": float(
            segment.get("end_radius_m", np.nan)
        ),
        f"{prefix}_cos_x_N_s": float(amplitude[0].real),
        f"{prefix}_cos_y_N_s": float(amplitude[1].real),
        f"{prefix}_cos_z_N_s": float(amplitude[2].real),
        f"{prefix}_sin_x_N_s": float(amplitude[0].imag),
        f"{prefix}_sin_y_N_s": float(amplitude[1].imag),
        f"{prefix}_sin_z_N_s": float(amplitude[2].imag),
        f"{prefix}_quad_error_x_N_s": float(error[0]),
        f"{prefix}_quad_error_y_N_s": float(error[1]),
        f"{prefix}_quad_error_z_N_s": float(error[2]),
        f"{prefix}_energy_x_J": float(modes[0]),
        f"{prefix}_energy_y_J": float(modes[1]),
        f"{prefix}_energy_z_J": float(modes[2]),
        f"{prefix}_energy_J": float(segment.get("energy_J", np.nan)),
        f"{prefix}_force_at_start_N": float(
            segment.get("force_at_start_N", np.nan)
        ),
        f"{prefix}_force_at_end_N": float(
            segment.get("force_at_end_N", np.nan)
        ),
    }


def run_fourier_prefix_R_far_to_switch(
    x_far_m: Array,
    v_far_m_s: Array,
    *,
    m_dm: float,
    eps: float,
    R_switch_m: float,
    dm_settings: DMOnlySettings,
    full_settings: FullSettings,
    fourier_settings: FourierSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
) -> dict[str, Any]:
    """Reintegrate the incoming R_far -> R_switch prefix with dense output."""

    x0 = _as_vector(x_far_m, "x_far_m")
    v0 = _as_vector(v_far_m_s, "v_far_m_s")
    r0 = float(np.linalg.norm(x0))
    target = float(R_switch_m)

    if np.isclose(r0, target, rtol=1.0e-10, atol=1.0e-15):
        return _empty_fourier_segment(
            status="incoming_R_far_to_R_switch_zero_length",
            complete=True,
            duration_s=0.0,
            global_start_time_s=0.0,
            start_radius_m=r0,
            end_radius_m=target,
        )
    if r0 < target:
        return _empty_fourier_segment(
            status="incoming_prefix_started_inside_R_switch",
            complete=False,
            start_radius_m=r0,
            end_radius_m=target,
        )

    speed = max(float(np.linalg.norm(v0)), 1.0e-12)
    arrival_estimate = r0 / speed
    t_max = max(
        dm_settings.time_factor * arrival_estimate,
        arrival_estimate + dm_settings.minimum_after_arrival_s,
    )

    def switch_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[:3]) - target)

    switch_event.terminal = True
    switch_event.direction = -1
    events: list[Callable] = [switch_event]

    if invalid_z_floor_m is not None:

        def invalid_event(_time: float, state: Array) -> float:
            return float(state[2] - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = -1
        events.append(invalid_event)

    solution = solve_ivp(
        _dm_only_rhs(m_dm=m_dm, eps=eps, environment=environment),
        (0.0, t_max),
        np.concatenate((x0, v0)),
        method="DOP853",
        rtol=dm_settings.rtol,
        atol=dm_settings.atol,
        max_step=dm_settings.max_step_s,
        events=events,
        dense_output=True,
    )

    reached = len(solution.t_events[0]) > 0
    invalid = len(events) > 1 and len(solution.t_events[1]) > 0
    if not reached or solution.sol is None:
        if invalid:
            status = "incoming_prefix_invalid_domain"
        elif not solution.success:
            status = "incoming_prefix_solver_failure"
        else:
            status = "incoming_prefix_did_not_reach_R_switch"
        return _empty_fourier_segment(
            status=status,
            complete=False,
            duration_s=float(solution.t[-1]) if solution.t.size else 0.0,
            global_start_time_s=np.nan,
            start_radius_m=r0,
            end_radius_m=(
                float(np.linalg.norm(solution.y[:3, -1]))
                if solution.y.size
                else np.nan
            ),
        )

    duration = float(solution.t_events[0][0])

    def force_at(local_time_s: float) -> Array:
        dm_state = np.asarray(solution.sol(float(local_time_s)), dtype=float)
        return _coulomb_force_on_ion(
            np.zeros(3),
            dm_state[:3],
            eps=eps,
            settings=full_settings,
            environment=environment,
        )

    segment = _integrate_fourier_force_segment(
        force_at,
        duration_s=duration,
        global_start_time_s=-duration,
        status="integrated_incoming_R_far_to_R_switch",
        start_radius_m=r0,
        end_radius_m=target,
        fourier_settings=fourier_settings,
        environment=environment,
    )
    segment["dm_state_at_switch"] = np.asarray(
        solution.y_events[0][0],
        dtype=float,
    )
    return segment


def run_fourier_incoming_outer_tail(
    x_far_m: Array,
    v_far_m_s: Array,
    *,
    prefix_duration_s: float,
    outer_radius_m: float,
    m_dm: float,
    eps: float,
    full_settings: FullSettings,
    fourier_settings: FourierSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
) -> dict[str, Any]:
    """Reconstruct and integrate incoming R_outer -> R_far trap-only tail."""

    x_far = _as_vector(x_far_m, "x_far_m")
    v_far = _as_vector(v_far_m_s, "v_far_m_s")
    r_far = float(np.linalg.norm(x_far))
    target = float(outer_radius_m)

    if target <= r_far * (1.0 + 1.0e-10):
        return _empty_fourier_segment(
            status="incoming_outer_zero_length",
            complete=True,
            duration_s=0.0,
            global_start_time_s=-float(prefix_duration_s),
            start_radius_m=r_far,
            end_radius_m=r_far,
        )

    base_rhs = _dm_only_rhs(m_dm=m_dm, eps=eps, environment=environment)

    def backward_rhs(_time: float, state: Array) -> Array:
        return -np.asarray(base_rhs(0.0, state), dtype=float)

    def outer_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[:3]) - target)

    outer_event.terminal = True
    outer_event.direction = 1
    events: list[Callable] = [outer_event]

    if invalid_z_floor_m is not None:

        def invalid_event(_time: float, state: Array) -> float:
            return float(state[2] - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = 0
        events.append(invalid_event)

    speed = max(float(np.linalg.norm(v_far)), 1.0e-12)
    travel_estimate = max((target - r_far) / speed, 0.0)
    t_max = max(
        fourier_settings.tail_time_factor * travel_estimate,
        travel_estimate + fourier_settings.tail_minimum_time_s,
    )

    solution = solve_ivp(
        backward_rhs,
        (0.0, t_max),
        np.concatenate((x_far, v_far)),
        method="DOP853",
        rtol=fourier_settings.tail_rtol,
        atol=fourier_settings.tail_atol,
        max_step=fourier_settings.tail_max_step_s,
        events=events,
        dense_output=True,
    )

    reached = len(solution.t_events[0]) > 0
    invalid = len(events) > 1 and len(solution.t_events[1]) > 0
    if not reached or solution.sol is None:
        if invalid:
            status = "incoming_outer_invalid_domain"
        elif not solution.success:
            status = "incoming_outer_solver_failure"
        else:
            status = "incoming_outer_did_not_reach_R_outer"
        return _empty_fourier_segment(
            status=status,
            complete=False,
            duration_s=float(solution.t[-1]) if solution.t.size else 0.0,
            start_radius_m=(
                float(np.linalg.norm(solution.y[:3, -1]))
                if solution.y.size
                else np.nan
            ),
            end_radius_m=r_far,
        )

    duration = float(solution.t_events[0][0])

    def dm_position_forward(local_time_s: float) -> Array:
        backward_time = duration - float(local_time_s)
        return np.asarray(solution.sol(backward_time), dtype=float)[:3]

    def force_at(local_time_s: float) -> Array:
        return _coulomb_force_on_ion(
            np.zeros(3),
            dm_position_forward(local_time_s),
            eps=eps,
            settings=full_settings,
            environment=environment,
        )

    segment = _integrate_fourier_force_segment(
        force_at,
        duration_s=duration,
        global_start_time_s=-(duration + float(prefix_duration_s)),
        status="integrated_incoming_R_outer_to_R_far",
        start_radius_m=target,
        end_radius_m=r_far,
        fourier_settings=fourier_settings,
        environment=environment,
    )
    segment["dm_state_at_outer"] = np.asarray(
        solution.y_events[0][0],
        dtype=float,
    )
    return segment


def run_fourier_outgoing_tail_to_radius(
    x_ion_start_m: Array,
    v_ion_start_m_s: Array,
    x_dm_start_m: Array,
    v_dm_start_m_s: Array,
    *,
    target_radius_m: float,
    global_start_time_s: float,
    segment_status: str,
    m_dm: float,
    eps: float,
    full_settings: FullSettings,
    fourier_settings: FourierSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
) -> dict[str, Any]:
    """Propagate a weak outgoing tail and integrate its Fourier amplitude."""

    x_ion0 = _as_vector(x_ion_start_m, "x_ion_start_m")
    v_ion0 = _as_vector(v_ion_start_m_s, "v_ion_start_m_s")
    x_dm0 = _as_vector(x_dm_start_m, "x_dm_start_m")
    v_dm0 = _as_vector(v_dm_start_m_s, "v_dm_start_m_s")

    initial_radius = float(np.linalg.norm(x_dm0))
    target = float(target_radius_m)
    if target <= initial_radius * (1.0 + 1.0e-10):
        segment = _empty_fourier_segment(
            status=segment_status + "_zero_length",
            complete=True,
            duration_s=0.0,
            global_start_time_s=global_start_time_s,
            start_radius_m=initial_radius,
            end_radius_m=initial_radius,
        )
        segment.update(
            {
                "x_ion_end_m": x_ion0.copy(),
                "v_ion_end_m_s": v_ion0.copy(),
                "x_dm_end_m": x_dm0.copy(),
                "v_dm_end_m_s": v_dm0.copy(),
            }
        )
        return segment

    def rhs(_time: float, state: Array) -> Array:
        x_dm = state[:3]
        v_dm = state[3:]
        dm_force = _trap_force_from_FDC_FRF(
            x_dm[None, :],
            mass=m_dm,
            charge=eps,
            environment=environment,
        )[0]
        return np.concatenate((v_dm, dm_force / float(m_dm)))

    def target_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[:3]) - target)

    target_event.terminal = True
    target_event.direction = 1
    events: list[Callable] = [target_event]

    if invalid_z_floor_m is not None:

        def invalid_event(_time: float, state: Array) -> float:
            return float(state[2] - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = -1
        events.append(invalid_event)

    speed = max(float(np.linalg.norm(v_dm0)), 1.0e-12)
    travel_estimate = max((target - initial_radius) / speed, 0.0)
    t_max = max(
        fourier_settings.tail_time_factor * travel_estimate,
        travel_estimate + fourier_settings.tail_minimum_time_s,
    )

    solution = solve_ivp(
        rhs,
        (0.0, t_max),
        np.concatenate((x_dm0, v_dm0)),
        method="DOP853",
        rtol=fourier_settings.tail_rtol,
        atol=fourier_settings.tail_atol,
        max_step=fourier_settings.tail_max_step_s,
        events=events,
        dense_output=True,
    )

    reached = len(solution.t_events[0]) > 0
    invalid = len(events) > 1 and len(solution.t_events[1]) > 0
    if not reached or solution.sol is None:
        if invalid:
            status = segment_status + "_invalid_domain"
        elif not solution.success:
            status = segment_status + "_solver_failure"
        else:
            status = segment_status + "_did_not_reach_target"
        return _empty_fourier_segment(
            status=status,
            complete=False,
            duration_s=float(solution.t[-1]) if solution.t.size else 0.0,
            global_start_time_s=global_start_time_s,
            start_radius_m=initial_radius,
            end_radius_m=(
                float(np.linalg.norm(solution.y[:3, -1]))
                if solution.y.size
                else np.nan
            ),
        )

    duration = float(solution.t_events[0][0])

    def states_at(local_time_s: float) -> tuple[Array, Array, Array, Array]:
        dm_state = np.asarray(solution.sol(float(local_time_s)), dtype=float)
        x_ion, v_ion = _free_harmonic_ion_state(
            local_time_s,
            x_ion0,
            v_ion0,
            environment=environment,
        )
        return x_ion, v_ion, dm_state[:3], dm_state[3:]

    def force_at(local_time_s: float) -> Array:
        x_ion, _v_ion, x_dm, _v_dm = states_at(local_time_s)
        return _coulomb_force_on_ion(
            x_ion,
            x_dm,
            eps=eps,
            settings=full_settings,
            environment=environment,
        )

    segment = _integrate_fourier_force_segment(
        force_at,
        duration_s=duration,
        global_start_time_s=global_start_time_s,
        status=segment_status,
        start_radius_m=initial_radius,
        end_radius_m=target,
        fourier_settings=fourier_settings,
        environment=environment,
    )

    x_ion_end, v_ion_end, x_dm_end, v_dm_end = states_at(duration)
    segment.update(
        {
            "x_ion_end_m": x_ion_end,
            "v_ion_end_m_s": v_ion_end,
            "x_dm_end_m": x_dm_end,
            "v_dm_end_m_s": v_dm_end,
            "tail_t_max_s": t_max,
        }
    )
    return segment


def run_full_ion_dm_after_switch_fourier(
    x_switch_m: Array,
    v_switch_m_s: Array,
    *,
    m_dm: float,
    eps: float,
    R_full_m: float,
    R_switch_m: float,
    R_far_m: float,
    incoming_prefix: dict[str, Any],
    incoming_outer_tail: dict[str, Any],
    settings: FullSettings,
    fourier_settings: FourierSettings,
    environment: SimulationEnvironment,
    invalid_z_floor_m: float | None,
    x0_ion_m: Array | None = None,
    v0_ion_m_s: Array | None = None,
    target_energy_J: float = 1.0e-27,
    keep_trajectory: bool = True,
) -> dict[str, Any]:
    """Run the coupled core and calculate full-pulse Fourier ion energy.

    The canonical ``E_final_J`` and ``delta_E_ion_J`` returned by this
    function are the coherent Fourier energy from incoming R_outer through
    outgoing R_outer.  Mechanical energies from the explicitly propagated core
    are retained under ``E_mechanical_*`` fields for comparison.
    """

    if not settings.use_harmonic_ion:
        raise ValueError(
            "The secular-mode Fourier energy requires use_harmonic_ion=True"
        )

    started = time.perf_counter()
    x_switch = _as_vector(x_switch_m, "x_switch_m")
    v_switch = _as_vector(v_switch_m_s, "v_switch_m_s")
    x_ion0 = (
        np.zeros(3)
        if x0_ion_m is None
        else _as_vector(x0_ion_m, "x0_ion_m")
    )
    v_ion0 = (
        np.zeros(3)
        if v0_ion_m_s is None
        else _as_vector(v0_ion_m_s, "v0_ion_m_s")
    )

    omega = np.asarray(environment.omega_vec, dtype=float).reshape(3)
    positive_omega = np.abs(omega[np.isfinite(omega) & (np.abs(omega) > 0.0)])
    if positive_omega.size == 0:
        raise ValueError("omega_vec must contain a positive secular frequency")

    fastest_period = float(2.0 * np.pi / np.max(positive_omega))
    resolution_step = fastest_period / max(
        int(fourier_settings.steps_per_secular_period),
        1,
    )
    full_sample_dt = min(float(settings.sample_dt_s), resolution_step)
    full_max_step = min(float(settings.max_step_s), resolution_step)

    initial_state = np.concatenate(
        (
            x_ion0,
            x_switch,
            v_ion0,
            v_switch,
            np.zeros(3, dtype=float),
            np.zeros(3, dtype=float),
        )
    )
    softening_sq = float(settings.coulomb_softening_m) ** 2

    def rhs(time_s: float, state: Array) -> Array:
        x_ion = state[0:3]
        x_dm = state[3:6]
        v_ion = state[6:9]
        v_dm = state[9:12]

        a_ion = -omega**2 * x_ion
        dm_force = _trap_force_from_FDC_FRF(
            x_dm[None, :],
            mass=m_dm,
            charge=eps,
            environment=environment,
        )[0]
        a_dm = dm_force / float(m_dm)

        delta = x_ion - x_dm
        separation_sq = max(
            float(np.dot(delta, delta)) + softening_sq,
            np.finfo(float).tiny,
        )
        coulomb_force = (
            environment.K
            * environment.Z_ion
            * float(eps)
            * environment.e**2
            * delta
            * separation_sq ** (-1.5)
        )

        a_ion = a_ion + coulomb_force / environment.m_ion
        a_dm = a_dm - coulomb_force / float(m_dm)
        phase = omega * float(time_s)

        return np.concatenate(
            (
                v_ion,
                v_dm,
                a_ion,
                a_dm,
                coulomb_force * np.cos(phase),
                coulomb_force * np.sin(phase),
            )
        )

    escape_radius = float(settings.escape_factor_from_R_switch) * float(
        R_switch_m
    )

    def escape_event(_time: float, state: Array) -> float:
        relative_position = state[3:6] - state[0:3]
        relative_velocity = state[9:12] - state[6:9]
        separation = float(np.linalg.norm(relative_position))
        if radial_velocity(relative_position, relative_velocity) <= 0.0:
            return -abs(separation - escape_radius) - 1.0e-30
        return separation - escape_radius

    escape_event.terminal = True
    escape_event.direction = 1

    def R_full_entry_event(_time: float, state: Array) -> float:
        return float(np.linalg.norm(state[3:6] - state[0:3]) - R_full_m)

    R_full_entry_event.terminal = False
    R_full_entry_event.direction = -1
    events: list[Callable] = [escape_event, R_full_entry_event]

    if invalid_z_floor_m is not None:

        def invalid_event(_time: float, state: Array) -> float:
            return float(min(state[2], state[5]) - invalid_z_floor_m)

        invalid_event.terminal = True
        invalid_event.direction = -1
        events.append(invalid_event)

    speed = max(float(np.linalg.norm(v_switch)), 1.0e-12)
    t_max = max(
        settings.time_factor * float(R_switch_m) / speed,
        settings.minimum_time_s,
    )
    t_eval = np.arange(0.0, t_max, full_sample_dt, dtype=float)
    if t_eval.size == 0 or not np.isclose(t_eval[0], 0.0):
        t_eval = np.insert(t_eval, 0, 0.0)
    if not np.isclose(t_eval[-1], t_max):
        t_eval = np.append(t_eval, t_max)

    solution = solve_ivp(
        rhs,
        (0.0, t_max),
        initial_state,
        method="DOP853",
        t_eval=t_eval,
        rtol=settings.rtol,
        atol=settings.atol,
        max_step=full_max_step,
        events=events,
    )

    escaped = len(solution.t_events[0]) > 0
    reached_R_full_event = len(solution.t_events[1]) > 0
    invalid_index = 2
    invalid = (
        len(events) > invalid_index
        and len(solution.t_events[invalid_index]) > 0
    )
    event_index = 0 if escaped else invalid_index if invalid else 0
    times, states = _append_event_state(solution, event_index)

    x_ion = states[0:3]
    x_dm = states[3:6]
    v_ion = states[6:9]
    v_dm = states[9:12]
    separations = np.linalg.norm(x_dm - x_ion, axis=0)
    d_min = float(np.min(separations)) if separations.size else np.nan
    reached_R_full = bool(
        reached_R_full_event
        or (np.isfinite(d_min) and d_min <= float(R_full_m))
    )
    full_t_reach_R_full_s = _first_event_time(solution, 1)

    mechanical_modes_history = _ion_energy_modes(
        x_ion,
        v_ion,
        settings=settings,
        environment=environment,
    )
    mechanical_total_history = np.sum(mechanical_modes_history, axis=0)
    final_time = float(times[-1]) if times.size else 0.0
    average_mask = times >= max(
        0.0,
        final_time - settings.energy_average_window_s,
    )
    if not np.any(average_mask):
        average_mask = np.ones_like(times, dtype=bool)

    mechanical_average_modes = np.mean(
        mechanical_modes_history[:, average_mask],
        axis=1,
    )
    mechanical_average_energy = float(np.sum(mechanical_average_modes))
    mechanical_final_modes = mechanical_modes_history[:, -1]
    mechanical_final_energy = float(np.sum(mechanical_final_modes))
    mechanical_initial_energy = (
        float(mechanical_total_history[0])
        if mechanical_total_history.size
        else 0.0
    )

    central_amplitude = (
        np.asarray(states[12:15, -1], dtype=float)
        + 1j * np.asarray(states[15:18, -1], dtype=float)
    )
    central_modes = (
        np.abs(central_amplitude) ** 2
        / (2.0 * environment.m_ion)
    )

    if invalid:
        status = "full_invalid_domain"
    elif not solution.success:
        status = "full_solver_failure"
    elif not escaped:
        status = "full_timeout"
    elif reached_R_full:
        status = "reached_R_full_and_escaped"
    else:
        status = "entered_switch_but_missed_R_full"

    core_resolved = bool(solution.success and escaped and not invalid)

    if core_resolved:
        outgoing_inner = run_fourier_outgoing_tail_to_radius(
            x_ion[:, -1],
            v_ion[:, -1],
            x_dm[:, -1],
            v_dm[:, -1],
            target_radius_m=float(R_far_m),
            global_start_time_s=final_time,
            segment_status="integrated_outgoing_escape_to_R_far",
            m_dm=m_dm,
            eps=eps,
            full_settings=settings,
            fourier_settings=fourier_settings,
            environment=environment,
            invalid_z_floor_m=invalid_z_floor_m,
        )
    else:
        outgoing_inner = _empty_fourier_segment(
            status="outgoing_R_far_tail_not_run_unresolved_core",
            complete=False,
            global_start_time_s=final_time,
            start_radius_m=float(np.linalg.norm(x_dm[:, -1])),
            end_radius_m=float(R_far_m),
        )

    outer_radius = float(
        fourier_settings.outer_radius_m
        if fourier_settings.outer_radius_m is not None
        else R_far_m
    )

    if outgoing_inner.get("complete", False):
        outgoing_outer = run_fourier_outgoing_tail_to_radius(
            outgoing_inner["x_ion_end_m"],
            outgoing_inner["v_ion_end_m_s"],
            outgoing_inner["x_dm_end_m"],
            outgoing_inner["v_dm_end_m_s"],
            target_radius_m=outer_radius,
            global_start_time_s=(
                final_time + float(outgoing_inner["duration_s"])
            ),
            segment_status="integrated_outgoing_R_far_to_R_outer",
            m_dm=m_dm,
            eps=eps,
            full_settings=settings,
            fourier_settings=fourier_settings,
            environment=environment,
            invalid_z_floor_m=invalid_z_floor_m,
        )
    else:
        outgoing_outer = _empty_fourier_segment(
            status="outgoing_outer_tail_not_run_missing_R_far_state",
            complete=False,
            global_start_time_s=np.nan,
            start_radius_m=np.nan,
            end_radius_m=outer_radius,
        )

    prefix_amplitude = np.asarray(
        incoming_prefix.get(
            "amplitude_N_s",
            np.zeros(3, dtype=complex),
        ),
        dtype=complex,
    ).reshape(3)
    incoming_outer_amplitude = np.asarray(
        incoming_outer_tail.get(
            "amplitude_N_s",
            np.zeros(3, dtype=complex),
        ),
        dtype=complex,
    ).reshape(3)
    outgoing_inner_amplitude = np.asarray(
        outgoing_inner.get(
            "amplitude_N_s",
            np.zeros(3, dtype=complex),
        ),
        dtype=complex,
    ).reshape(3)
    outgoing_outer_amplitude = np.asarray(
        outgoing_outer.get(
            "amplitude_N_s",
            np.zeros(3, dtype=complex),
        ),
        dtype=complex,
    ).reshape(3)

    amplitude_R_far_to_escape = prefix_amplitude + central_amplitude
    amplitude_R_far_roundtrip = (
        amplitude_R_far_to_escape + outgoing_inner_amplitude
    )
    amplitude_full_pulse = (
        incoming_outer_amplitude
        + amplitude_R_far_roundtrip
        + outgoing_outer_amplitude
    )

    modes_R_far_to_escape = (
        np.abs(amplitude_R_far_to_escape) ** 2
        / (2.0 * environment.m_ion)
    )
    modes_R_far_roundtrip = (
        np.abs(amplitude_R_far_roundtrip) ** 2
        / (2.0 * environment.m_ion)
    )
    modes_full_pulse = (
        np.abs(amplitude_full_pulse) ** 2
        / (2.0 * environment.m_ion)
    )

    E_fourier_switch = float(np.sum(central_modes))
    E_fourier_R_far_to_escape = float(np.sum(modes_R_far_to_escape))
    E_fourier_R_far = float(np.sum(modes_R_far_roundtrip))
    E_fourier_R_outer = float(np.sum(modes_full_pulse))

    required_segments_complete = bool(
        incoming_prefix.get("complete", False)
        and incoming_outer_tail.get("complete", False)
        and outgoing_inner.get("complete", False)
        and outgoing_outer.get("complete", False)
    )
    fourier_resolved = bool(
        core_resolved
        and (
            required_segments_complete
            or not fourier_settings.require_complete_tails
        )
    )
    above_threshold = bool(
        fourier_resolved
        and np.isfinite(E_fourier_R_outer)
        and E_fourier_R_outer >= float(target_energy_J)
    )
    canonical_fourier_energy = (
        E_fourier_R_outer if fourier_resolved else np.nan
    )
    canonical_fourier_modes = (
        modes_full_pulse
        if fourier_resolved
        else np.full(3, np.nan, dtype=float)
    )

    interaction_switch = _time_inside(
        times,
        separations <= float(R_switch_m),
    )
    interaction_full = _time_inside(
        times,
        separations <= float(R_full_m),
    )

    result = {
        "full_success": bool(solution.success),
        "full_status": status,
        "full_message": str(solution.message),
        "reached_R_full": reached_R_full,
        "above_energy_threshold": above_threshold,
        "energy_method": "full_pulse_fourier",
        "fourier_detection_resolved": fourier_resolved,
        "fourier_all_required_segments_complete": required_segments_complete,
        "fourier_require_complete_tails": bool(
            fourier_settings.require_complete_tails
        ),
        "fourier_detection_history": (
            "incoming_R_outer_to_outgoing_R_outer"
        ),
        "fourier_time_origin": "incoming_R_switch_crossing",
        "fourier_outer_radius_m": outer_radius,
        "fourier_fastest_secular_period_s": fastest_period,
        "fourier_resolution_step_s": resolution_step,
        "d_min_m": d_min,
        "d_min_um": d_min * 1.0e6 if np.isfinite(d_min) else np.nan,

        # Canonical final energy requested by the user.
        "E_initial_J": 0.0,
        "E_final_J": canonical_fourier_energy,
        "delta_E_ion_J": canonical_fourier_energy,
        "E_x_J": float(canonical_fourier_modes[0]),
        "E_y_J": float(canonical_fourier_modes[1]),
        "E_z_J": float(canonical_fourier_modes[2]),

        # Explicit Fourier diagnostics.
        "E_fourier_total_J": E_fourier_R_outer,
        "E_fourier_from_R_outer_J": E_fourier_R_outer,
        "E_fourier_from_R_outer_x_J": float(modes_full_pulse[0]),
        "E_fourier_from_R_outer_y_J": float(modes_full_pulse[1]),
        "E_fourier_from_R_outer_z_J": float(modes_full_pulse[2]),
        "E_fourier_from_R_far_J": E_fourier_R_far,
        "E_fourier_from_R_far_x_J": float(modes_R_far_roundtrip[0]),
        "E_fourier_from_R_far_y_J": float(modes_R_far_roundtrip[1]),
        "E_fourier_from_R_far_z_J": float(modes_R_far_roundtrip[2]),
        "E_fourier_R_far_in_to_escape_J": E_fourier_R_far_to_escape,
        "E_fourier_from_R_switch_J": E_fourier_switch,
        "E_fourier_from_R_switch_x_J": float(central_modes[0]),
        "E_fourier_from_R_switch_y_J": float(central_modes[1]),
        "E_fourier_from_R_switch_z_J": float(central_modes[2]),
        "fourier_cos_x_N_s": float(amplitude_full_pulse[0].real),
        "fourier_cos_y_N_s": float(amplitude_full_pulse[1].real),
        "fourier_cos_z_N_s": float(amplitude_full_pulse[2].real),
        "fourier_sin_x_N_s": float(amplitude_full_pulse[0].imag),
        "fourier_sin_y_N_s": float(amplitude_full_pulse[1].imag),
        "fourier_sin_z_N_s": float(amplitude_full_pulse[2].imag),
        "fourier_history_t_initial_s": -(
            float(incoming_outer_tail.get("duration_s", 0.0))
            + float(incoming_prefix.get("duration_s", 0.0))
        ),
        "fourier_history_t_final_s": (
            final_time
            + float(outgoing_inner.get("duration_s", 0.0))
            + float(outgoing_outer.get("duration_s", 0.0))
        ),

        # Mechanical core values are retained only as diagnostics.
        "E_mechanical_initial_J": mechanical_initial_energy,
        "E_mechanical_final_J": mechanical_final_energy,
        "E_mechanical_average_J": mechanical_average_energy,
        "E_mechanical_delta_average_J": (
            mechanical_average_energy - mechanical_initial_energy
        ),
        "E_mechanical_x_average_J": float(mechanical_average_modes[0]),
        "E_mechanical_y_average_J": float(mechanical_average_modes[1]),
        "E_mechanical_z_average_J": float(mechanical_average_modes[2]),

        "interaction_time_inside_R_switch_s": interaction_switch,
        "interaction_time_inside_R_full_s": interaction_full,
        "full_t_final_s": final_time,
        "full_t_reach_R_full_s": full_t_reach_R_full_s,
        "full_t_max_s": t_max,
        "full_runtime_s": time.perf_counter() - started,
        "x_ion_initial_m": x_ion0.copy(),
        "v_ion_initial_m_s": v_ion0.copy(),
        "x_dm_initial_m": x_switch.copy(),
        "v_dm_initial_m_s": v_switch.copy(),
        "x_ion_final_m": x_ion[:, -1].copy(),
        "v_ion_final_m_s": v_ion[:, -1].copy(),
        "x_dm_final_m": x_dm[:, -1].copy(),
        "v_dm_final_m_s": v_dm[:, -1].copy(),
    }

    result.update(
        _flatten_fourier_segment(
            incoming_prefix,
            prefix="fourier_incoming_R_far_prefix",
        )
    )
    result.update(
        _flatten_fourier_segment(
            incoming_outer_tail,
            prefix="fourier_incoming_outer_tail",
        )
    )
    result.update(
        _flatten_fourier_segment(
            outgoing_inner,
            prefix="fourier_outgoing_R_far_tail",
        )
    )
    result.update(
        _flatten_fourier_segment(
            outgoing_outer,
            prefix="fourier_outgoing_outer_tail",
        )
    )

    if keep_trajectory:
        result.update(
            {
                "times_s": times,
                "states": states,
                "x_ion_m": x_ion,
                "v_ion_m_s": v_ion,
                "x_dm_m": x_dm,
                "v_dm_m_s": v_dm,
                "separation_m": separations,
                "ion_energy_modes_J": mechanical_modes_history,
                "ion_total_energy_J": mechanical_total_history,
            }
        )
    return result


# =============================================================================
# OPTIONAL RUTHERFORD R_FULL HELPER
# =============================================================================


def reduced_mass(m_dm: float, m_ion: float) -> float:
    if m_dm <= 0.0 or m_ion <= 0.0:
        raise ValueError("Masses must be positive")
    return float(m_dm * m_ion / (m_dm + m_ion))


def rutherford_r_min_threshold_m(
    *,
    speed_m_s: float,
    m_dm: float,
    eps: float,
    target_energy_J: float,
    environment: SimulationEnvironment,
) -> float:
    """Analytic Test 0/Test 1 Rutherford radius used as ``R_full``."""

    speed = float(speed_m_s)
    if speed <= 0.0 or target_energy_J <= 0.0:
        return float("nan")
    mu = reduced_mass(m_dm, environment.m_ion)
    minimum_speed = math.sqrt(
        target_energy_J * environment.m_ion / (2.0 * mu**2)
    )
    if speed < minimum_speed:
        return float("nan")

    coupling = abs(
        environment.K
        * environment.Z_ion
        * float(eps)
        * environment.e**2
    )
    if coupling <= 0.0:
        return float("nan")

    return (
        coupling / (mu * speed**2)
        + coupling / speed
        * math.sqrt(2.0 / (environment.m_ion * target_energy_J))
    )


# =============================================================================
# SUMMARY AND PLOTTING
# =============================================================================


def _print_stage_summary(result: dict[str, Any]) -> None:
    parameters = result["parameters"]
    straight = result["straight"]
    dm_only = result.get("dm_only")
    full = result.get("full")

    R_far = float(straight.get("R_far_actual_m", np.nan))
    R_switch = float(parameters.get("R_switch_m", np.nan))
    R_full = float(parameters.get("R_full_m", np.nan))
    t_far = float(parameters.get("t_reach_R_far_from_R_far_start_s", 0.0))
    t_switch = float(parameters.get("t_reach_R_switch_from_R_far_start_s", np.nan))
    t_full = float(parameters.get("t_reach_R_full_from_R_far_start_s", np.nan))

    print("\n-------- Staged Simulation Summary --------")
    print(
        "Radii:         "
        f"R_far={R_far * 1e6:.6f} um  "
        f"R_switch={R_switch * 1e6:.6f} um  "
        f"R_full={R_full * 1e6:.6f} um"
    )
    print(
        "Reach times:   "
        f"t(R_far)={t_far * 1e6:.6f} us  "
        f"t(R_switch)={t_switch * 1e6:.6f} us  "
        f"t(R_full)={t_full * 1e6:.6f} us  "
        "[time origin is the DM launch at R_far]"
    )
    outer_nominal = float(
        parameters.get("nominal_time_R_max_to_R_far_s", np.nan)
    )
    if np.isfinite(outer_nominal):
        print(
            "Outer-line diagnostic: nominal time from R_max to R_far="
            f"{outer_nominal * 1e6:.6f} us"
        )

    print(
        "Straight tail: "
        f"status={straight.get('outer_status')}  "
        f"v_far={straight.get('v_far_m_s', np.nan):.6g} m/s  "
        f"limiting={straight.get('limiting_criterion')}"
    )
    if dm_only is not None:
        print(
            "DM only:      "
            f"status={dm_only.get('dm_only_status')}  "
            f"entered_switch={dm_only.get('entered_switch')}  "
            f"r_min={dm_only.get('dm_only_min_radius_m', np.nan) * 1e6:.3f} um  "
            f"r_min/R_far={dm_only.get('dm_only_min_radius_over_R_far', np.nan):.6g}  "
            f"runtime={dm_only.get('dm_only_runtime_s', np.nan):.3f} s"
        )
    if full is not None:
        print(
            "Full coupled: "
            f"status={full.get('full_status')}  "
            f"reached_R_full={full.get('reached_R_full')}  "
            f"d_min={full.get('d_min_um', np.nan):.3f} um  "
            f"E_FT(R_outer)={full.get('E_fourier_from_R_outer_J', np.nan):.6e} J  "
            f"E_FT(R_far)={full.get('E_fourier_from_R_far_J', np.nan):.6e} J  "
            f"E_mech={full.get('E_mechanical_final_J', np.nan):.6e} J  "
            f"FT_resolved={full.get('fourier_detection_resolved')}  "
            f"runtime={full.get('full_runtime_s', np.nan):.3f} s"
        )

    print("\n-------- Initial and Final States --------")
    print("Positions are in um; velocities are in m/s.")
    print(
        "DM initial at R_far:  "
        f"x={_format_vector(parameters.get('x_dm_initial_at_R_far_m', np.full(3, np.nan)), scale=1e6)}  "
        f"v={_format_vector(parameters.get('v_dm_initial_at_R_far_m_s', np.full(3, np.nan)))}"
    )
    if dm_only is not None and dm_only.get("entered_switch", False):
        print(
            "DM at R_switch:     "
            f"x={_format_vector(dm_only.get('x_switch_m', np.full(3, np.nan)), scale=1e6)}  "
            f"v={_format_vector(dm_only.get('v_switch_m_s', np.full(3, np.nan)))}"
        )

    print(
        "Ion initial:          "
        f"x={_format_vector(parameters.get('x_ion_initial_m', np.zeros(3)), scale=1e6)}  "
        f"v={_format_vector(parameters.get('v_ion_initial_m_s', np.zeros(3)))}"
    )

    if full is not None:
        print(
            "Ion final:            "
            f"x={_format_vector(full.get('x_ion_final_m', np.full(3, np.nan)), scale=1e6)}  "
            f"v={_format_vector(full.get('v_ion_final_m_s', np.full(3, np.nan)))}"
        )
        print(
            "DM final:             "
            f"x={_format_vector(full.get('x_dm_final_m', np.full(3, np.nan)), scale=1e6)}  "
            f"v={_format_vector(full.get('v_dm_final_m_s', np.full(3, np.nan)))}"
        )
    elif dm_only is not None:
        print("Ion final:            not propagated because the full stage was not entered")
        print(
            "DM final:             "
            f"x={_format_vector(dm_only.get('x_dm_final_m', np.full(3, np.nan)), scale=1e6)}  "
            f"v={_format_vector(dm_only.get('v_dm_final_m_s', np.full(3, np.nan)))}"
        )
    else:
        print("Ion final:            not propagated")
        print("DM final:             not propagated beyond the straight-line diagnostic")

    print(f"\nOverall status: {result.get('overall_status')}")


def plot_simulation(result: dict[str, Any]) -> None:
    """Display diagnostic figures.  This function does not save files."""

    import matplotlib.pyplot as plt

    straight = result["straight"]
    xyz = np.asarray(straight["xyz_path_m"])
    radius = np.linalg.norm(xyz, axis=1)

    plt.figure(figsize=(8, 5))
    plt.semilogx(radius * 1.0e6, straight["energy_ratio_tail_max_array"], label="energy tail")
    plt.semilogx(radius * 1.0e6, straight["theta_abs_tail_array"], label="angle tail")
    plt.semilogx(
        radius * 1.0e6,
        np.asarray(straight["displacement_abs_tail_array_m"]) * 1.0e6,
        label="displacement tail [um]",
    )
    plt.axvline(straight["R_far_actual_m"] * 1.0e6, linestyle="--", label="R_far")
    plt.xlabel("Nominal path radius [um]")
    plt.ylabel("Diagnostic value")
    plt.title("Test 1 outer-tail diagnostics")
    plt.legend()
    plt.tight_layout()
    plt.show()

    dm_only = result.get("dm_only")
    if dm_only is not None and "times_s" in dm_only:
        states = np.asarray(dm_only["states"])
        dm_radius = np.linalg.norm(states[:3], axis=0)
        plt.figure(figsize=(8, 5))
        plt.plot(np.asarray(dm_only["times_s"]) * 1.0e6, dm_radius * 1.0e6)
        plt.axhline(result["parameters"]["R_switch_m"] * 1.0e6, linestyle="--", label="R_switch")
        plt.axhline(straight["R_far_actual_m"] * 1.0e6, linestyle=":", label="R_far")
        plt.xlabel("DM-only time [us]")
        plt.ylabel("DM radius [um]")
        plt.title("DM-only trap propagation")
        plt.legend()
        plt.tight_layout()
        plt.show()

    full = result.get("full")
    if full is not None and "times_s" in full:
        plt.figure(figsize=(8, 5))
        plt.plot(np.asarray(full["times_s"]) * 1.0e6, np.asarray(full["separation_m"]) * 1.0e6)
        plt.axhline(result["parameters"]["R_switch_m"] * 1.0e6, linestyle="--", label="R_switch")
        plt.axhline(result["parameters"]["R_full_m"] * 1.0e6, linestyle=":", label="R_full")
        plt.xlabel("Full-stage time [us]")
        plt.ylabel("Ion-DM separation [um]")
        plt.title("Full coupled ion-DM propagation")
        plt.legend()
        plt.tight_layout()
        plt.show()

        x_ion = np.asarray(full["x_ion_m"])
        energy = np.asarray(full["ion_total_energy_J"])
        times = np.asarray(full["times_s"])

        plt.figure(figsize=(8, 5))
        plt.plot(times * 1.0e6, energy, label="mechanical core energy")
        fourier_energy = float(full.get("E_fourier_from_R_outer_J", np.nan))
        if np.isfinite(fourier_energy):
            plt.axhline(
                fourier_energy,
                linestyle="--",
                label="full-pulse Fourier energy",
            )
        plt.xlabel("Full-stage time [us]")
        plt.ylabel("Ion energy [J]")
        plt.title("Mechanical core energy and full-pulse Fourier result")
        plt.legend()
        plt.tight_layout()
        plt.show()

        x_ion_um = x_ion * 1.0e6
        plane_specs = (
            (0, 1, "x [um]", "y [um]", "Ion motion in the xy plane"),
            (1, 2, "y [um]", "z [um]", "Ion motion in the yz plane"),
            (0, 2, "x [um]", "z [um]", "Ion motion in the xz plane"),
        )
        for first, second, xlabel, ylabel, title in plane_specs:
            plt.figure(figsize=(7, 6))
            plt.plot(x_ion_um[first], x_ion_um[second], label="ion trajectory")
            plt.scatter(
                [x_ion_um[first, 0]],
                [x_ion_um[second, 0]],
                marker="o",
                label="initial ion",
            )
            plt.scatter(
                [x_ion_um[first, -1]],
                [x_ion_um[second, -1]],
                marker="x",
                label="final ion",
            )
            plt.xlabel(xlabel)
            plt.ylabel(ylabel)
            plt.title(title)
            plt.axis("equal")
            plt.legend()
            plt.tight_layout()
            plt.show()

        fig = plt.figure(figsize=(8, 6))
        axis = fig.add_subplot(projection="3d")
        x_dm = np.asarray(full["x_dm_m"]) * 1.0e6
        axis.plot(x_dm[0], x_dm[1], x_dm[2], label="DM")
        axis.plot(x_ion_um[0], x_ion_um[1], x_ion_um[2], label="ion")
        axis.set_xlabel("x [um]")
        axis.set_ylabel("y [um]")
        axis.set_zlabel("z [um]")
        axis.set_title("Full coupled trajectories")
        axis.legend()
        plt.tight_layout()
        plt.show()


# =============================================================================
# MAIN PUBLIC ENTRY POINT
# =============================================================================


def single_simulation(
    m_dm: float,
    eps: float,
    theta: float,
    alpha: float,
    psi: float,
    b: float,
    *,
    speed: float | None = None,
    T: float = 300.0,
    seed: int = 1,
    R_full: float | None = None,
    R_switch: float | None = None,
    R_switch_factor: float = 2.0,
    target_E: float = 1.0e-27,
    R_max: float = 20.0e-3,
    n_path: int = 800,
    energy_tol: float = 1.0e-2,
    angle_tol: float = 1.0e-3,
    displacement_tol: float = 1.0e-6,
    dm_only_rtol: float = 1.0e-6,
    dm_only_atol: float = 1.0e-9,
    dm_only_max_step: float = 5.0e-7,
    full_rtol: float = 1.0e-6,
    full_atol: float = 1.0e-9,
    full_dt: float = 5.0e-9,
    full_max_step: float = 5.0e-9,
    fourier_outer_radius: float | None = None,
    fourier_require_complete_tails: bool = True,
    fourier_steps_per_secular_period: int = 20,
    fourier_tail_rtol: float = 1.0e-6,
    fourier_tail_atol: float = 1.0e-9,
    fourier_tail_max_step: float = 5.0e-7,
    fourier_tail_time_factor: float = 4.0,
    fourier_tail_minimum_time: float = 50.0e-6,
    fourier_quad_epsrel: float = 1.0e-8,
    fourier_quad_epsabs_dimensionless: float = 1.0e-11,
    fourier_quad_limit: int = 1000,
    fourier_force_scale_samples: int = 21,
    use_harmonic_ion: bool = True,
    coulomb_softening: float = 0.0,
    x0_ion: Array | None = None,
    v0_ion: Array | None = None,
    invalid_z_floor: float | None = None,
    z_margin: float = 1.0e-6,
    allow_R_far_lower_bound: bool = False,
    keep_trajectories: bool = True,
    print_summary: bool = True,
    plot: bool = False,
    environment: SimulationEnvironment | None = None,
) -> dict[str, Any]:
    """Run the complete straight-tail -> DM-only -> full coupled simulation.

    Parameters
    ----------
    m_dm, eps
        DM mass in kg and charge in units of the elementary charge.
    theta, alpha, psi, b
        Incoming direction and impact geometry.  Angles are radians and ``b``
        is in metres.
    speed
        Asymptotic speed ``v_inf`` in m/s.  If omitted, one Maxwell-Boltzmann
        3D velocity is sampled at temperature ``T`` and only its magnitude is
        used.
    R_full
        Full-interaction radius in metres.  If omitted, the analytic
        Rutherford threshold radius is calculated from ``speed`` and
        ``target_E``.
    R_switch
        DM-only/full-ODE handoff radius.  Defaults to
        ``R_switch_factor * R_full``.

    fourier_outer_radius
        Radius used to terminate the weak incoming and outgoing Fourier tails.
        The default is ``R_max``.  The Fourier time origin is the incoming
        R_switch crossing.
    fourier_require_complete_tails
        When true, the threshold flag is unresolved unless all four weak-tail
        segments and the coupled core complete successfully.

    Returns
    -------
    dict
        Nested dictionaries named ``straight``, ``dm_only`` and ``full``.
        In ``full``, ``E_final_J`` and ``delta_E_ion_J`` are the coherent
        full-pulse Fourier energy from incoming R_outer to outgoing R_outer.
        Mechanical core energies remain available as ``E_mechanical_*``.
        If the trajectory fails before a later stage, the corresponding later
        dictionary is ``None`` and ``overall_status`` explains why.
    """

    total_started = time.perf_counter()
    env = _resolve_environment(environment)

    m_dm = float(m_dm)
    eps = float(eps)
    b = float(b)
    if m_dm <= 0.0:
        raise ValueError("m_dm must be positive")
    if b < 0.0:
        raise ValueError("b must be nonnegative")

    if speed is None:
        if T <= 0.0:
            raise ValueError("T must be positive when speed is sampled")
        rng = np.random.default_rng(seed)
        sigma = math.sqrt(1.380649e-23 * float(T) / m_dm)
        velocity_sample = rng.normal(0.0, sigma, size=3)
        speed = float(np.linalg.norm(velocity_sample))
    else:
        speed = float(speed)
    if not np.isfinite(speed) or speed <= 0.0:
        raise ValueError("speed must be positive and finite")

    if R_full is None:
        R_full = rutherford_r_min_threshold_m(
            speed_m_s=speed,
            m_dm=m_dm,
            eps=eps,
            target_energy_J=target_E,
            environment=env,
        )
        if not np.isfinite(R_full):
            raise ValueError(
                "Could not derive R_full from the Rutherford threshold. "
                "Provide R_full explicitly or choose a speed above the "
                "kinematic threshold."
            )
    R_full = float(R_full)
    if R_full <= 0.0:
        raise ValueError("R_full must be positive")

    if R_switch is None:
        R_switch = float(R_switch_factor) * R_full
    R_switch = float(R_switch)
    if R_switch <= R_full:
        raise ValueError("R_switch must be greater than R_full")

    if invalid_z_floor is None and env.ion_height_m is not None:
        invalid_z_floor = -float(env.ion_height_m) + float(z_margin)

    outer_settings = OuterTailSettings(
        R_max_m=float(R_max),
        n_path=int(n_path),
        energy_tol=float(energy_tol),
        angle_tol=float(angle_tol),
        displacement_tol_m=float(displacement_tol),
        z_margin_m=float(z_margin),
        allow_R_far_lower_bound=bool(allow_R_far_lower_bound),
    )
    dm_settings = DMOnlySettings(
        rtol=float(dm_only_rtol),
        atol=float(dm_only_atol),
        max_step_s=float(dm_only_max_step),
    )
    full_settings = FullSettings(
        rtol=float(full_rtol),
        atol=float(full_atol),
        sample_dt_s=float(full_dt),
        max_step_s=float(full_max_step),
        use_harmonic_ion=bool(use_harmonic_ion),
        coulomb_softening_m=float(coulomb_softening),
    )
    resolved_fourier_outer_radius = (
        float(R_max)
        if fourier_outer_radius is None
        else float(fourier_outer_radius)
    )
    if not np.isfinite(resolved_fourier_outer_radius) or resolved_fourier_outer_radius <= 0.0:
        raise ValueError("fourier_outer_radius must be finite and positive")
    fourier_settings = FourierSettings(
        outer_radius_m=resolved_fourier_outer_radius,
        require_complete_tails=bool(fourier_require_complete_tails),
        steps_per_secular_period=int(fourier_steps_per_secular_period),
        tail_rtol=float(fourier_tail_rtol),
        tail_atol=float(fourier_tail_atol),
        tail_max_step_s=float(fourier_tail_max_step),
        tail_time_factor=float(fourier_tail_time_factor),
        tail_minimum_time_s=float(fourier_tail_minimum_time),
        quad_epsrel=float(fourier_quad_epsrel),
        quad_epsabs_dimensionless=float(
            fourier_quad_epsabs_dimensionless
        ),
        quad_limit=int(fourier_quad_limit),
        force_scale_samples=int(fourier_force_scale_samples),
    )

    parameters = {
        "m_dm_kg": m_dm,
        "eps": eps,
        "theta_rad": float(theta),
        "alpha_rad": float(alpha),
        "psi_rad": float(psi),
        "b_m": b,
        "b_um": b * 1.0e6,
        "v_inf_m_s": speed,
        "R_full_m": R_full,
        "R_full_um": R_full * 1.0e6,
        "R_switch_m": R_switch,
        "R_switch_um": R_switch * 1.0e6,
        "R_max_m": float(R_max),
        "fourier_outer_radius_m": resolved_fourier_outer_radius,
        "fourier_energy_method": "full_pulse_fourier",
        "target_E_J": float(target_E),
        "invalid_z_floor_m": invalid_z_floor,
        "x_ion_initial_m": (
            np.zeros(3, dtype=float)
            if x0_ion is None
            else _as_vector(x0_ion, "x0_ion").copy()
        ),
        "v_ion_initial_m_s": (
            np.zeros(3, dtype=float)
            if v0_ion is None
            else _as_vector(v0_ion, "v0_ion").copy()
        ),
        "t_reach_R_far_from_R_far_start_s": 0.0,
        "t_reach_R_switch_from_R_far_start_s": np.nan,
        "t_reach_R_full_from_R_far_start_s": np.nan,
    }

    straight = run_straight_outer_tail_test(
        theta=float(theta),
        alpha=float(alpha),
        psi=float(psi),
        b_m=b,
        v_inf_m_s=speed,
        m_dm=m_dm,
        eps=eps,
        # The outer-tail handoff must not lie inside R_switch because the
        # Fourier prefix is defined from R_far to the incoming switch crossing.
        # The coupled interaction radius R_full remains unchanged below.
        R_full_m=R_switch,
        settings=outer_settings,
        environment=env,
        invalid_z_floor_m=invalid_z_floor,
    )

    result: dict[str, Any] = {
        "parameters": parameters,
        "straight": straight,
        "dm_only": None,
        "full": None,
        "overall_status": "straight_stage_complete",
    }

    resolved_straight_statuses = {
        "straight_safe_to_full_sphere",
        "straight_then_dm_only",
    }
    if straight.get("outer_status") not in resolved_straight_statuses:
        if not (
            straight.get("outer_status") == "needs_larger_R_max"
            and allow_R_far_lower_bound
        ):
            result["overall_status"] = (
                "stopped_after_straight_stage:" + str(straight.get("outer_status"))
            )
            result["total_runtime_s"] = time.perf_counter() - total_started
            if print_summary:
                _print_stage_summary(result)
            if plot:
                plot_simulation(result)
            return result

    x_far = _as_vector(straight["xyz_far_m"], "straight xyz_far_m")
    v_far = _as_vector(
        straight["velocity_far_m_s"],
        "straight velocity_far_m_s",
    )
    parameters.update(
        {
            "R_far_path_coordinate_m": float(straight["R_far_m"]),
            "R_far_actual_m": float(straight["R_far_actual_m"]),
            "R_far_actual_um": float(straight["R_far_actual_m"]) * 1.0e6,
            "v_far_m_s": float(straight["v_far_m_s"]),
            "x_far_m": x_far.copy(),
            "v_far_vector_m_s": v_far.copy(),
            "initial_radial_velocity_at_R_far_m_s": radial_velocity(x_far, v_far),
            "x_dm_initial_at_R_far_m": x_far.copy(),
            "v_dm_initial_at_R_far_m_s": v_far.copy(),
            "nominal_time_R_max_to_R_far_s": max(
                float(np.max(np.asarray(straight.get("s_grid_m", [straight["R_far_m"]]))))
                - float(straight["R_far_m"]),
                0.0,
            ) / speed,
        }
    )

    dm_only = run_dm_trap_only_to_switch(
        x_far,
        v_far,
        m_dm=m_dm,
        eps=eps,
        R_switch_m=R_switch,
        settings=dm_settings,
        environment=env,
        invalid_z_floor_m=invalid_z_floor,
        keep_trajectory=keep_trajectories,
    )
    r_far_actual = float(straight["R_far_actual_m"])
    dm_only["dm_only_min_radius_over_R_far"] = (
        float(dm_only["dm_only_min_radius_m"]) / r_far_actual
        if r_far_actual > 0.0 and np.isfinite(dm_only["dm_only_min_radius_m"])
        else np.nan
    )
    dm_only["moved_inward_from_R_far"] = bool(
        np.isfinite(dm_only["dm_only_min_radius_m"])
        and dm_only["dm_only_min_radius_m"] < r_far_actual
    )
    result["dm_only"] = dm_only
    if dm_only.get("entered_switch", False):
        parameters["t_reach_R_switch_from_R_far_start_s"] = float(
            dm_only.get("dm_only_t_reach_R_switch_s", np.nan)
        )

    if not dm_only.get("entered_switch", False):
        result["overall_status"] = (
            "stopped_after_dm_only_stage:" + str(dm_only.get("dm_only_status"))
        )
        result["total_runtime_s"] = time.perf_counter() - total_started
        if print_summary:
            _print_stage_summary(result)
        if plot:
            plot_simulation(result)
        return result

    incoming_prefix = run_fourier_prefix_R_far_to_switch(
        x_far,
        v_far,
        m_dm=m_dm,
        eps=eps,
        R_switch_m=R_switch,
        dm_settings=dm_settings,
        full_settings=full_settings,
        fourier_settings=fourier_settings,
        environment=env,
        invalid_z_floor_m=invalid_z_floor,
    )
    incoming_outer_tail = run_fourier_incoming_outer_tail(
        x_far,
        v_far,
        prefix_duration_s=float(incoming_prefix.get("duration_s", 0.0)),
        outer_radius_m=resolved_fourier_outer_radius,
        m_dm=m_dm,
        eps=eps,
        full_settings=full_settings,
        fourier_settings=fourier_settings,
        environment=env,
        invalid_z_floor_m=invalid_z_floor,
    )
    result["fourier_incoming_R_far_prefix"] = incoming_prefix
    result["fourier_incoming_outer_tail"] = incoming_outer_tail

    full = run_full_ion_dm_after_switch_fourier(
        dm_only["x_switch_m"],
        dm_only["v_switch_m_s"],
        m_dm=m_dm,
        eps=eps,
        R_full_m=R_full,
        R_switch_m=R_switch,
        R_far_m=float(straight["R_far_actual_m"]),
        incoming_prefix=incoming_prefix,
        incoming_outer_tail=incoming_outer_tail,
        settings=full_settings,
        fourier_settings=fourier_settings,
        environment=env,
        invalid_z_floor_m=invalid_z_floor,
        x0_ion_m=x0_ion,
        v0_ion_m_s=v0_ion,
        target_energy_J=target_E,
        keep_trajectory=keep_trajectories,
    )
    result["full"] = full
    t_switch_from_start = float(
        parameters.get("t_reach_R_switch_from_R_far_start_s", np.nan)
    )
    t_full_after_switch = float(full.get("full_t_reach_R_full_s", np.nan))
    if np.isfinite(t_switch_from_start) and np.isfinite(t_full_after_switch):
        parameters["t_reach_R_full_from_R_far_start_s"] = (
            t_switch_from_start + t_full_after_switch
        )
    result["overall_status"] = str(full.get("full_status"))
    result["total_runtime_s"] = time.perf_counter() - total_started

    if print_summary:
        _print_stage_summary(result)
    if plot:
        plot_simulation(result)
    return result


# =============================================================================
# EXAMPLE
# =============================================================================

# Example after running the trap-model notebook cells:
#
# result = single_simulation(
#     m_dm=2.636651e-25,
#     eps=2.335721,
#     theta=np.pi / 2,
#     alpha=0.0,
#     psi=0.0,
#     b=10.0e-6,
#     speed=1000.0,
#     R_full=61.340612e-6,
#     R_switch=122.681224e-6,
#     plot=True,
# )
#
# E_final = result["full"]["E_final_J"] if result["full"] is not None else np.nan
```

## Potential Energy

```python
# Escape point: (0,0,57um) (refer to verify_potential.ipynb)
# finds local minimum of a 1D array
def find_local_extrema(phi_sliced: np.ndarray):
    delta_phi = [0]
    min_coord = []
    min_val = []
    max_coord = []
    max_val = []
    for i in range(1,len(phi_sliced)):
        delta_phi.append(phi_sliced[i]-phi_sliced[i-1])
        prod = delta_phi[i]*delta_phi[i-1]
        if (prod<0 and delta_phi[i-1]<0):
            min_coord.append(i-1)
            min_val.append(phi_sliced[i-1])
        elif (prod<0 and delta_phi[i-1]>0):
            max_coord.append(i-1)
            max_val.append(phi_sliced[i-1])
    return min_coord,min_val,max_coord,max_val
# Obtain potential energies along (0,0,z), 0<z<100um
z_grid = np.zeros((1000,3))
z_grid[:, 2] = np.arange(0.1, 100.1, 0.1)*1e-6
PE = potential_energy(z_grid,m_dm,eps)
print(PE)
escape_z, escape_energy = find_local_extrema(PE)[2][0],find_local_extrema(PE)[3][0]
# Minimum speed of DM required at infinity
v_min = np.sqrt(2*escape_energy/m_dm)
K0 = 0.5*m_dm*v_min**2
scale = np.sqrt(k_B*T_dm/m_dm)
prob_v_min = 1-maxwell.cdf(v_min,scale=scale)
print(f"The potential energy at the escape point {escape_z*0.1} um is {escape_energy} J.")
print(f"The minimum speed of the DM at infinity is {v_min} m/s and the probability of reaching this speed is {prob_v_min}.")
```

## Scan probability of overcoming potential barrier over (m_dm,eps)

```python
#!/usr/bin/env python3
"""Mass-dependent DM escape-barrier heat maps.

This version fixes the invalid assumption that the z-axis escape point is a
function of epsilon alone.  It supports two efficient search modes:

1. ``ratio`` mode (preferred)
   The code verifies the common Paul-trap scaling

       U(z; s*m, s*eps) = s U(z; m, eps),

   so that the escape-point location depends on eps/m.  For regular log-spaced
   input arrays it constructs the exact log10(eps/m) lattice, evaluates one
   z-profile per lattice point, and maps those results to the full mass-charge
   grid without interpolation.

2. ``sparse2d`` fallback
   If the scaling check fails, the code evaluates a configurable sparse grid in
   (log10(m), log10(eps)) and uses nearest-neighbour interpolation.  This avoids
   an infeasible 5000*5000*1000 direct scan while preserving mass dependence.

For every retained point the module returns and plots:

* Maxwell-Boltzmann probability of exceeding the barrier;
* minimum speed at infinity;
* potential energy at the first positive-z local maximum;
* z coordinate of that local maximum.

Points with probability below ``probability_threshold`` are stored as NaN and
are not shown.  Every heat map has metric-specific labelled contour lines;
non-probability maps can also show dashed probability contours. The heat
maps additionally overlay the five fitted straight-line Mathieu stability
boundaries supplied with the stability diagram.

Required external callable
--------------------------

    potential_energy(positions_m, m_dm_kg, eps, **potential_kwargs)

It must accept pairwise NumPy arrays and return one value per position row.
"""

from __future__ import annotations

from dataclasses import dataclass
from fractions import Fraction
from typing import Any, Callable, Iterable, Mapping, Sequence

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LogNorm, Normalize, SymLogNorm
from matplotlib.patches import Patch
from matplotlib.ticker import MaxNLocator
from scipy.interpolate import RegularGridInterpolator
from scipy.stats import maxwell


K_B = 1.380649e-23

# -----------------------------------------------------------------------------
# Mathieu-stability boundaries from the updated coupled-Mathieu reference plot.
# Each boundary is written as
#
#     m = slope * epsilon,
#
# with slope in kilograms.  The heat maps use epsilon on the horizontal axis
# and mass on the vertical axis, exactly matching the stability diagram.
#
# The reference Floquet scan is stable at the lowest epsilon/m ratio.  As
# epsilon/m increases, stability alternates whenever one of these boundaries is
# crossed.  This gives the visible stable bands used by the overlay below.
# -----------------------------------------------------------------------------
DEFAULT_MATHIEU_BOUNDARIES: tuple[tuple[str, float], ...] = (
    ("Boundary 1", 7.534324e-19),
    ("Boundary 2", 7.399333e-24),
    ("Boundary 3", 7.467762e-26),
)
DEFAULT_MATHIEU_LOW_RATIO_STABLE = True

DEFAULT_PROBABILITY_CONTOURS: tuple[float, ...] = (
    1e-6, 3e-6, 1e-5, 3e-5, 1e-4, 3e-4, 1e-3,
    3e-3, 1e-2, 3e-2, 1e-1, 3e-1, 5e-1, 8e-1,
)


@dataclass
class DynamicEscapeResult:
    """Output from :func:`calculate_dynamic_escape_grid`.

    All four metric arrays and ``peak_found`` have shape ``(n_mass, n_eps)``.
    Metric arrays contain NaN below the probability threshold.
    """

    m_arr_kg: np.ndarray
    eps_arr: np.ndarray
    probability: np.ndarray
    minimum_speed_m_s: np.ndarray
    escape_energy_J: np.ndarray
    z_escape_m: np.ndarray
    peak_found: np.ndarray
    temperature_K: float
    probability_threshold: float
    z_grid_m: np.ndarray
    search_mode: str
    search_metadata: dict[str, Any]

    # Compatibility with the previous file.  It is not physically meaningful
    # for a mass-dependent search and is therefore NaN.
    escape_search_mass_kg: float = np.nan


# =============================================================================
# POTENTIAL EVALUATION
# =============================================================================


def _call_potential_energy(
    potential_energy_fn: Callable[..., Any],
    positions_m: np.ndarray,
    m_values_kg: np.ndarray | float,
    eps_values: np.ndarray | float,
    potential_kwargs: Mapping[str, Any] | None = None,
) -> np.ndarray:
    """Evaluate one potential-energy value per position row."""

    positions = np.asarray(positions_m, dtype=float).reshape(-1, 3)
    kwargs = {} if potential_kwargs is None else dict(potential_kwargs)
    values = np.asarray(
        potential_energy_fn(positions, m_values_kg, eps_values, **kwargs),
        dtype=float,
    )

    if values.ndim == 0:
        if positions.shape[0] != 1:
            raise ValueError(
                "potential_energy returned a scalar for vector inputs. "
                "The scan requires pairwise vectorization."
            )
        values = values.reshape(1)
    else:
        values = np.squeeze(values).reshape(-1)

    if values.size != positions.shape[0]:
        raise ValueError(
            "potential_energy returned the wrong number of values: "
            f"expected {positions.shape[0]}, got {values.size}."
        )
    if not np.all(np.isfinite(values)):
        raise FloatingPointError(
            "potential_energy produced NaN or infinite values."
        )
    return values


def verify_pairwise_vectorization(
    potential_energy_fn: Callable[..., Any],
    potential_kwargs: Mapping[str, Any] | None = None,
) -> None:
    """Fail early if pairwise mass/charge evaluation is unsupported."""

    positions = np.array(
        [[0.0, 0.0, 20e-6], [0.0, 0.0, 50e-6], [0.0, 0.0, 80e-6]],
        dtype=float,
    )
    masses = np.array([1e-27, 1e-24, 1e-20], dtype=float)
    charges = np.array([1e-3, 1.0, 1e3], dtype=float)
    _call_potential_energy(
        potential_energy_fn,
        positions,
        masses,
        charges,
        potential_kwargs,
    )


def _evaluate_pair_profiles(
    potential_energy_fn: Callable[..., Any],
    z_grid_m: np.ndarray,
    masses_kg: np.ndarray,
    eps_values: np.ndarray,
    potential_kwargs: Mapping[str, Any] | None,
) -> np.ndarray:
    """Evaluate z profiles for pairwise ``(mass, epsilon)`` values.

    Returns an array with shape ``(n_pairs, n_z)``.
    """

    z_grid = np.asarray(z_grid_m, dtype=float).reshape(-1)
    masses = np.asarray(masses_kg, dtype=float).reshape(-1)
    charges = np.asarray(eps_values, dtype=float).reshape(-1)
    if masses.size != charges.size:
        raise ValueError("masses_kg and eps_values must have equal length")

    n_pair = masses.size
    n_z = z_grid.size
    base_positions = np.zeros((n_z, 3), dtype=float)
    base_positions[:, 2] = z_grid

    positions = np.tile(base_positions, (n_pair, 1))
    mass_rows = np.repeat(masses, n_z)
    charge_rows = np.repeat(charges, n_z)

    values = _call_potential_energy(
        potential_energy_fn,
        positions,
        mass_rows,
        charge_rows,
        potential_kwargs,
    )
    return values.reshape(n_pair, n_z)


# =============================================================================
# LOCAL-MAXIMUM SEARCH
# =============================================================================


def find_first_local_maximum_batch(
    profiles: np.ndarray,
    z_grid_m: np.ndarray,
    *,
    quadratic_refinement: bool = True,
) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    """Find the first interior positive-to-nonpositive slope transition."""

    y = np.asarray(profiles, dtype=float)
    z = np.asarray(z_grid_m, dtype=float).reshape(-1)

    if y.ndim != 2 or y.shape[1] != z.size:
        raise ValueError("profiles must have shape (n_profile, len(z_grid_m))")
    if z.size < 3:
        raise ValueError("At least three z samples are required")
    if not np.all(np.diff(z) > 0.0):
        raise ValueError("z_grid_m must be strictly increasing")

    slopes = np.diff(y, axis=1)
    local_max = (slopes[:, :-1] > 0.0) & (slopes[:, 1:] <= 0.0)
    found = np.any(local_max, axis=1)

    peak_index = np.argmax(local_max, axis=1) + 1
    rows = np.arange(y.shape[0])
    z_peak = np.full(y.shape[0], np.nan, dtype=float)
    value_peak = np.full(y.shape[0], np.nan, dtype=float)

    valid_rows = rows[found]
    valid_idx = peak_index[found]
    z_peak[found] = z[valid_idx]
    value_peak[found] = y[valid_rows, valid_idx]

    if quadratic_refinement and np.any(found):
        ym = y[valid_rows, valid_idx - 1]
        y0 = y[valid_rows, valid_idx]
        yp = y[valid_rows, valid_idx + 1]

        denominator = ym - 2.0 * y0 + yp
        safe = np.isfinite(denominator) & (np.abs(denominator) > 0.0)
        offset = np.zeros_like(y0)
        offset[safe] = 0.5 * (ym[safe] - yp[safe]) / denominator[safe]
        offset = np.clip(offset, -1.0, 1.0)

        dz_left = z[valid_idx] - z[valid_idx - 1]
        dz_right = z[valid_idx + 1] - z[valid_idx]
        uniform = np.isclose(dz_left, dz_right, rtol=1e-8, atol=0.0)
        dz = 0.5 * (dz_left + dz_right)
        offset = np.where(uniform, offset, 0.0)

        a = 0.5 * (ym + yp - 2.0 * y0)
        b = 0.5 * (yp - ym)
        refined_value = y0 + b * offset + a * offset**2

        z_peak[found] = z[valid_idx] + offset * dz
        value_peak[found] = refined_value

    return z_peak, value_peak, found


# =============================================================================
# EPS/M REDUCTION
# =============================================================================


def _regular_log_spacing(values: np.ndarray) -> tuple[bool, float]:
    logs = np.log10(np.asarray(values, dtype=float).reshape(-1))
    if logs.size < 2:
        return True, 0.0
    diffs = np.abs(np.diff(logs))
    step = float(np.median(diffs))
    regular = np.allclose(diffs, step, rtol=1e-9, atol=1e-12)
    return bool(regular), step


def build_log_ratio_lookup(
    masses: np.ndarray,
    charges: np.ndarray,
    *,
    max_exact_ratio_profiles: int = 100_000,
    approximate_ratio_profiles: int = 20_000,
) -> tuple[np.ndarray, dict[str, Any]]:
    """Build an exact ratio lattice for regular log grids when practical."""

    masses = np.asarray(masses, dtype=float).reshape(-1)
    charges = np.asarray(charges, dtype=float).reshape(-1)
    log_m = np.log10(masses)
    log_e = np.log10(charges)
    lo = float(np.min(log_e) - np.max(log_m))
    hi = float(np.max(log_e) - np.min(log_m))

    m_regular, dm = _regular_log_spacing(masses)
    e_regular, de = _regular_log_spacing(charges)

    if m_regular and e_regular and dm > 0.0 and de > 0.0:
        ratio = de / dm
        fraction = Fraction(ratio).limit_denominator(20_000)
        p = int(fraction.numerator)
        q = int(fraction.denominator)
        base_a = dm / q
        base_b = de / p
        base = 0.5 * (base_a + base_b)
        ratio_error = abs(base_a - base_b) / max(base, np.finfo(float).tiny)
        n_exact = int(np.rint((hi - lo) / base)) + 1

        if ratio_error <= 1e-8 and n_exact <= max_exact_ratio_profiles:
            grid = lo + base * np.arange(n_exact, dtype=float)
            return grid, {
                "lookup_kind": "exact_log_ratio_lattice",
                "log_ratio_min": lo,
                "log_ratio_step": base,
                "n_ratio_profiles": n_exact,
                "mass_log_step": dm,
                "eps_log_step": de,
                "step_ratio_fraction": f"{p}/{q}",
            }

    n = max(3, int(approximate_ratio_profiles))
    return np.linspace(lo, hi, n, dtype=float), {
        "lookup_kind": "uniform_log_ratio_nearest",
        "log_ratio_min": lo,
        "log_ratio_step": (hi - lo) / (n - 1),
        "n_ratio_profiles": n,
    }


def representative_pairs_for_log_ratio(
    log_ratio: np.ndarray,
    masses: np.ndarray,
    charges: np.ndarray,
) -> tuple[np.ndarray, np.ndarray]:
    """Choose feasible representative pairs with the requested eps/m ratio."""

    lr = np.asarray(log_ratio, dtype=float).reshape(-1)
    log_m_min = float(np.log10(np.min(masses)))
    log_m_max = float(np.log10(np.max(masses)))
    log_e_min = float(np.log10(np.min(charges)))
    log_e_max = float(np.log10(np.max(charges)))

    lower = np.maximum(log_m_min, log_e_min - lr)
    upper = np.minimum(log_m_max, log_e_max - lr)
    if np.any(lower > upper + 1e-12):
        raise RuntimeError("Could not construct a feasible representative pair")

    log_m_ref = 0.5 * (lower + upper)
    log_e_ref = lr + log_m_ref
    return 10.0**log_m_ref, 10.0**log_e_ref


def validate_ratio_homogeneity(
    potential_energy_fn: Callable[..., Any],
    masses: np.ndarray,
    charges: np.ndarray,
    z_grid_m: np.ndarray,
    *,
    potential_kwargs: Mapping[str, Any] | None,
    n_ratio_checks: int = 7,
    profile_relative_tolerance: float = 2e-6,
    z_tolerance_grid_steps: float = 2.0,
    progress: bool = True,
) -> tuple[bool, dict[str, float]]:
    """Check ``U/m`` invariance at fixed eps/m.

    This is the correct reduction for a potential containing terms proportional
    to ``m``, ``eps``, and ``eps**2/m``.  It permits a mass-dependent escape
    coordinate while reducing the search to one dimension.
    """

    log_m_min = float(np.log10(np.min(masses)))
    log_m_max = float(np.log10(np.max(masses)))
    log_e_min = float(np.log10(np.min(charges)))
    log_e_max = float(np.log10(np.max(charges)))
    lr_values = np.linspace(
        log_e_min - log_m_max,
        log_e_max - log_m_min,
        max(3, int(n_ratio_checks)),
    )

    worst_profile_error = 0.0
    worst_z_spread = 0.0
    common_peaks = 0

    for lr in lr_values:
        lower = max(log_m_min, log_e_min - lr)
        upper = min(log_m_max, log_e_max - lr)
        if upper < lower:
            continue

        if np.isclose(lower, upper):
            log_m_test = np.array([lower], dtype=float)
        else:
            log_m_test = np.linspace(lower, upper, 3)
        m_test = 10.0**log_m_test
        e_test = 10.0 ** (lr + log_m_test)

        profiles = _evaluate_pair_profiles(
            potential_energy_fn,
            z_grid_m,
            m_test,
            e_test,
            potential_kwargs,
        )
        scaled = profiles / m_test[:, None]
        reference = scaled[0]
        amplitude = max(float(np.nanmax(np.abs(reference))), np.finfo(float).tiny)
        profile_error = float(np.nanmax(np.abs(scaled - reference)) / amplitude)
        worst_profile_error = max(worst_profile_error, profile_error)

        z_peak, _, found = find_first_local_maximum_batch(
            profiles,
            z_grid_m,
            quadratic_refinement=True,
        )
        if np.all(found):
            common_peaks += 1
            spread = float(np.nanmax(z_peak) - np.nanmin(z_peak))
            worst_z_spread = max(worst_z_spread, spread)

    dz = float(np.median(np.diff(z_grid_m)))
    passed = (
        worst_profile_error <= profile_relative_tolerance
        and worst_z_spread <= z_tolerance_grid_steps * dz
    )

    metadata = {
        "worst_scaled_profile_relative_error": worst_profile_error,
        "worst_fixed_ratio_z_spread_m": worst_z_spread,
        "fixed_ratio_common_peak_checks": float(common_peaks),
        "profile_relative_tolerance": float(profile_relative_tolerance),
        "z_tolerance_m": float(z_tolerance_grid_steps * dz),
    }

    if progress:
        verdict = "passed" if passed else "failed"
        print(
            "eps/m scaling validation " + verdict + ": "
            f"profile error={worst_profile_error:.3e}, "
            f"z spread={worst_z_spread * 1e6:.6g} um."
        )
    return passed, metadata


def search_escape_on_ratio_lookup(
    potential_energy_fn: Callable[..., Any],
    log_ratio_lookup: np.ndarray,
    masses: np.ndarray,
    charges: np.ndarray,
    z_grid_m: np.ndarray,
    *,
    pair_profile_chunk_size: int,
    potential_kwargs: Mapping[str, Any] | None,
    quadratic_refinement: bool,
    progress: bool,
) -> dict[str, np.ndarray]:
    """Evaluate one z profile for every eps/m lookup value."""

    m_ref, e_ref = representative_pairs_for_log_ratio(
        log_ratio_lookup,
        masses,
        charges,
    )
    n = log_ratio_lookup.size
    z_peak = np.full(n, np.nan, dtype=float)
    energy_peak = np.full(n, np.nan, dtype=float)
    found = np.zeros(n, dtype=bool)

    total_chunks = (n + pair_profile_chunk_size - 1) // pair_profile_chunk_size
    for chunk_number, i0 in enumerate(
        range(0, n, pair_profile_chunk_size),
        start=1,
    ):
        i1 = min(i0 + pair_profile_chunk_size, n)
        profiles = _evaluate_pair_profiles(
            potential_energy_fn,
            z_grid_m,
            m_ref[i0:i1],
            e_ref[i0:i1],
            potential_kwargs,
        )
        z_chunk, u_chunk, found_chunk = find_first_local_maximum_batch(
            profiles,
            z_grid_m,
            quadratic_refinement=quadratic_refinement,
        )
        z_peak[i0:i1] = z_chunk
        energy_peak[i0:i1] = u_chunk
        found[i0:i1] = found_chunk

        if progress and (
            chunk_number == 1
            or chunk_number == total_chunks
            or chunk_number % max(1, total_chunks // 20) == 0
        ):
            print(
                f"Ratio-profile search {chunk_number:4d}/{total_chunks}: "
                f"indices [{i0}:{i1}), peaks "
                f"{int(np.count_nonzero(found_chunk))}/{i1-i0}"
            )

    energy_per_mass = np.divide(
        energy_peak,
        m_ref,
        out=np.full_like(energy_peak, np.nan),
        where=found & np.isfinite(energy_peak) & (m_ref > 0.0),
    )
    return {
        "log_ratio": np.asarray(log_ratio_lookup, dtype=float),
        "m_reference_kg": m_ref,
        "eps_reference": e_ref,
        "z_escape_m": z_peak,
        "energy_per_mass_J_kg": energy_per_mass,
        "peak_found": found,
    }


def _nearest_lookup_indices(
    values: np.ndarray,
    grid: np.ndarray,
) -> np.ndarray:
    """Return nearest sorted-grid indices for arbitrary values."""

    flat = np.asarray(values, dtype=float)
    lookup = np.asarray(grid, dtype=float).reshape(-1)
    right = np.searchsorted(lookup, flat, side="left")
    right = np.clip(right, 0, lookup.size - 1)
    left = np.clip(right - 1, 0, lookup.size - 1)
    choose_right = np.abs(lookup[right] - flat) < np.abs(flat - lookup[left])
    return np.where(choose_right, right, left)


# =============================================================================
# SPARSE 2D FALLBACK
# =============================================================================


def _subset_indices(size: int, n_requested: int) -> np.ndarray:
    return np.unique(
        np.linspace(0, size - 1, min(size, int(n_requested)), dtype=int)
    )


def search_sparse_2d_escape_grid(
    potential_energy_fn: Callable[..., Any],
    masses: np.ndarray,
    charges: np.ndarray,
    z_grid_m: np.ndarray,
    *,
    n_mass_search: int,
    n_eps_search: int,
    pair_profile_chunk_size: int,
    potential_kwargs: Mapping[str, Any] | None,
    quadratic_refinement: bool,
    progress: bool,
) -> dict[str, Any]:
    """Evaluate a sparse mass-charge search grid for fallback interpolation."""

    mi = _subset_indices(masses.size, n_mass_search)
    ei = _subset_indices(charges.size, n_eps_search)
    m_coarse = masses[mi]
    e_coarse = charges[ei]

    mm, ee = np.meshgrid(m_coarse, e_coarse, indexing="ij")
    m_pairs = mm.reshape(-1)
    e_pairs = ee.reshape(-1)
    n_pair = m_pairs.size

    z_flat = np.full(n_pair, np.nan, dtype=float)
    u_flat = np.full(n_pair, np.nan, dtype=float)
    found_flat = np.zeros(n_pair, dtype=bool)

    total_chunks = (n_pair + pair_profile_chunk_size - 1) // pair_profile_chunk_size
    for chunk_number, i0 in enumerate(
        range(0, n_pair, pair_profile_chunk_size),
        start=1,
    ):
        i1 = min(i0 + pair_profile_chunk_size, n_pair)
        profiles = _evaluate_pair_profiles(
            potential_energy_fn,
            z_grid_m,
            m_pairs[i0:i1],
            e_pairs[i0:i1],
            potential_kwargs,
        )
        z_chunk, u_chunk, found_chunk = find_first_local_maximum_batch(
            profiles,
            z_grid_m,
            quadratic_refinement=quadratic_refinement,
        )
        z_flat[i0:i1] = z_chunk
        u_flat[i0:i1] = u_chunk
        found_flat[i0:i1] = found_chunk

        if progress and (
            chunk_number == 1
            or chunk_number == total_chunks
            or chunk_number % max(1, total_chunks // 20) == 0
        ):
            print(
                f"Sparse 2D search {chunk_number:4d}/{total_chunks}: "
                f"pairs [{i0}:{i1}), peaks "
                f"{int(np.count_nonzero(found_chunk))}/{i1-i0}"
            )

    z_coarse = z_flat.reshape(m_coarse.size, e_coarse.size)
    u_coarse = u_flat.reshape(m_coarse.size, e_coarse.size)
    found_coarse = found_flat.reshape(m_coarse.size, e_coarse.size)

    # Interpolators require ascending axes.
    m_order = np.argsort(m_coarse)
    e_order = np.argsort(e_coarse)
    log_m_axis = np.log10(m_coarse[m_order])
    log_e_axis = np.log10(e_coarse[e_order])
    z_sorted = z_coarse[np.ix_(m_order, e_order)]
    u_sorted = u_coarse[np.ix_(m_order, e_order)]
    f_sorted = found_coarse[np.ix_(m_order, e_order)].astype(float)

    return {
        "m_coarse_kg": m_coarse,
        "eps_coarse": e_coarse,
        "z_escape_m": z_coarse,
        "escape_energy_J": u_coarse,
        "peak_found": found_coarse,
        "z_interpolator": RegularGridInterpolator(
            (log_m_axis, log_e_axis),
            z_sorted,
            method="nearest",
            bounds_error=False,
            fill_value=np.nan,
        ),
        "energy_interpolator": RegularGridInterpolator(
            (log_m_axis, log_e_axis),
            u_sorted,
            method="nearest",
            bounds_error=False,
            fill_value=np.nan,
        ),
        "found_interpolator": RegularGridInterpolator(
            (log_m_axis, log_e_axis),
            f_sorted,
            method="nearest",
            bounds_error=False,
            fill_value=0.0,
        ),
    }


# =============================================================================
# GRID CALCULATION
# =============================================================================


def barrier_probability_from_energy(
    escape_energy_J: np.ndarray,
    temperature_K: float,
) -> np.ndarray:
    """Maxwell-Boltzmann probability of exceeding the barrier."""

    barrier = np.maximum(np.asarray(escape_energy_J, dtype=float), 0.0)
    standardized_speed = np.sqrt(2.0 * barrier / (K_B * float(temperature_K)))
    return maxwell.sf(standardized_speed)


def calculate_dynamic_escape_grid(
    potential_energy_fn: Callable[..., Any],
    m_arr_kg: np.ndarray,
    eps_arr: np.ndarray,
    *,
    temperature_K: float = 300.0,
    probability_threshold: float = 1e-6,
    z_min_m: float = 0.1e-6,
    z_max_m: float = 100.0e-6,
    n_z: int = 1000,
    search_mode: str = "auto",
    max_exact_ratio_profiles: int = 100_000,
    approximate_ratio_profiles: int = 20_000,
    pair_profile_chunk_size: int = 64,
    eps_grid_chunk_size: int = 32,
    fallback_mass_points: int = 161,
    fallback_eps_points: int = 161,
    ratio_profile_relative_tolerance: float = 2e-6,
    ratio_z_tolerance_grid_steps: float = 2.0,
    quadratic_refinement: bool = True,
    potential_kwargs: Mapping[str, Any] | None = None,
    progress: bool = True,
) -> DynamicEscapeResult:
    """Calculate mass-dependent escape metrics on the requested grid."""

    masses = np.asarray(m_arr_kg, dtype=float).reshape(-1)
    charges = np.asarray(eps_arr, dtype=float).reshape(-1)

    if masses.size == 0 or charges.size == 0:
        raise ValueError("m_arr_kg and eps_arr must be nonempty")
    if np.any(masses <= 0.0):
        raise ValueError("All masses must be positive")
    if np.any(charges <= 0.0):
        raise ValueError("All epsilon values must be positive")
    if temperature_K <= 0.0:
        raise ValueError("temperature_K must be positive")
    if not (0.0 < probability_threshold < 1.0):
        raise ValueError("probability_threshold must lie strictly between 0 and 1")
    if not (z_max_m > z_min_m >= 0.0):
        raise ValueError("Require z_max_m > z_min_m >= 0")
    if n_z < 3:
        raise ValueError("n_z must be at least 3")
    if search_mode not in {"auto", "ratio", "sparse2d"}:
        raise ValueError("search_mode must be 'auto', 'ratio', or 'sparse2d'")

    verify_pairwise_vectorization(potential_energy_fn, potential_kwargs)
    z_grid = np.linspace(z_min_m, z_max_m, int(n_z), dtype=float)

    ratio_ok = False
    ratio_validation: dict[str, float] = {}
    if search_mode in {"auto", "ratio"}:
        ratio_ok, ratio_validation = validate_ratio_homogeneity(
            potential_energy_fn,
            masses,
            charges,
            z_grid,
            potential_kwargs=potential_kwargs,
            profile_relative_tolerance=ratio_profile_relative_tolerance,
            z_tolerance_grid_steps=ratio_z_tolerance_grid_steps,
            progress=progress,
        )
        if search_mode == "ratio" and not ratio_ok:
            print(
                "Warning: forced ratio mode failed validation; switching to "
                "the sparse 2D mass-dependent search instead."
            )

    actual_mode = "ratio" if ratio_ok and search_mode != "sparse2d" else "sparse2d"
    metadata: dict[str, Any] = {"ratio_validation": ratio_validation}

    ratio_data: dict[str, np.ndarray] | None = None
    sparse_data: dict[str, Any] | None = None

    if actual_mode == "ratio":
        log_ratio_lookup, lookup_metadata = build_log_ratio_lookup(
            masses,
            charges,
            max_exact_ratio_profiles=max_exact_ratio_profiles,
            approximate_ratio_profiles=approximate_ratio_profiles,
        )
        ratio_data = search_escape_on_ratio_lookup(
            potential_energy_fn,
            log_ratio_lookup,
            masses,
            charges,
            z_grid,
            pair_profile_chunk_size=pair_profile_chunk_size,
            potential_kwargs=potential_kwargs,
            quadratic_refinement=quadratic_refinement,
            progress=progress,
        )
        metadata.update(lookup_metadata)
        metadata["ratio_data"] = ratio_data
    else:
        print(
            "Using sparse 2D mass-dependent escape search.  Increase "
            "fallback_mass_points/fallback_eps_points for a finer fallback."
        )
        sparse_data = search_sparse_2d_escape_grid(
            potential_energy_fn,
            masses,
            charges,
            z_grid,
            n_mass_search=fallback_mass_points,
            n_eps_search=fallback_eps_points,
            pair_profile_chunk_size=pair_profile_chunk_size,
            potential_kwargs=potential_kwargs,
            quadratic_refinement=quadratic_refinement,
            progress=progress,
        )
        metadata.update(
            {
                "fallback_mass_points": int(sparse_data["m_coarse_kg"].size),
                "fallback_eps_points": int(sparse_data["eps_coarse"].size),
                "sparse_data": sparse_data,
            }
        )

    n_m = masses.size
    n_e = charges.size
    shape = (n_m, n_e)

    probability = np.full(shape, np.nan, dtype=np.float32)
    minimum_speed = np.full(shape, np.nan, dtype=np.float32)
    escape_energy = np.full(shape, np.nan, dtype=np.float32)
    z_escape = np.full(shape, np.nan, dtype=np.float32)
    peak_found = np.zeros(shape, dtype=bool)

    log_m = np.log10(masses)
    log_e = np.log10(charges)
    total_chunks = (n_e + eps_grid_chunk_size - 1) // eps_grid_chunk_size

    for chunk_number, j0 in enumerate(
        range(0, n_e, eps_grid_chunk_size),
        start=1,
    ):
        j1 = min(j0 + eps_grid_chunk_size, n_e)
        n_chunk = j1 - j0

        if actual_mode == "ratio":
            assert ratio_data is not None
            log_ratio_block = log_e[None, j0:j1] - log_m[:, None]
            lookup = ratio_data["log_ratio"]
            if metadata["lookup_kind"] == "exact_log_ratio_lattice":
                step = float(metadata["log_ratio_step"])
                start = float(metadata["log_ratio_min"])
                lookup_idx = np.rint((log_ratio_block - start) / step).astype(np.int64)
                lookup_idx = np.clip(lookup_idx, 0, lookup.size - 1)
            else:
                lookup_idx = _nearest_lookup_indices(log_ratio_block, lookup)

            found_block = ratio_data["peak_found"][lookup_idx]
            z_block = ratio_data["z_escape_m"][lookup_idx]
            g_block = ratio_data["energy_per_mass_J_kg"][lookup_idx]
            energy_block = masses[:, None] * g_block
        else:
            assert sparse_data is not None
            lm, le = np.meshgrid(log_m, log_e[j0:j1], indexing="ij")
            points = np.column_stack([lm.reshape(-1), le.reshape(-1)])
            found_block = (
                sparse_data["found_interpolator"](points).reshape(n_m, n_chunk)
                >= 0.5
            )
            z_block = sparse_data["z_interpolator"](points).reshape(n_m, n_chunk)
            energy_block = sparse_data["energy_interpolator"](points).reshape(
                n_m, n_chunk
            )

        barrier_block = np.maximum(energy_block, 0.0)
        vmin_block = np.sqrt(2.0 * barrier_block / masses[:, None])
        probability_block = barrier_probability_from_energy(
            energy_block,
            temperature_K,
        )

        keep = (
            found_block
            & np.isfinite(z_block)
            & np.isfinite(energy_block)
            & np.isfinite(vmin_block)
            & np.isfinite(probability_block)
            & (probability_block >= probability_threshold)
        )

        probability[:, j0:j1] = np.where(keep, probability_block, np.nan).astype(
            np.float32
        )
        minimum_speed[:, j0:j1] = np.where(keep, vmin_block, np.nan).astype(
            np.float32
        )
        escape_energy[:, j0:j1] = np.where(keep, energy_block, np.nan).astype(
            np.float32
        )
        z_escape[:, j0:j1] = np.where(keep, z_block, np.nan).astype(np.float32)
        peak_found[:, j0:j1] = found_block

        if progress:
            print(
                f"Full grid {chunk_number:4d}/{total_chunks}: epsilon "
                f"indices [{j0}:{j1}), retained "
                f"{int(np.count_nonzero(keep))}/{keep.size}"
            )

    return DynamicEscapeResult(
        m_arr_kg=masses,
        eps_arr=charges,
        probability=probability,
        minimum_speed_m_s=minimum_speed,
        escape_energy_J=escape_energy,
        z_escape_m=z_escape,
        peak_found=peak_found,
        temperature_K=float(temperature_K),
        probability_threshold=float(probability_threshold),
        z_grid_m=z_grid,
        search_mode=actual_mode,
        search_metadata=metadata,
    )


# =============================================================================
# CONTOUR LEVELS, MATHIEU BOUNDARIES, AND PLOTTING
# =============================================================================


def _plot_indices(size: int, maximum: int) -> np.ndarray:
    if size <= maximum:
        return np.arange(size, dtype=int)
    return np.unique(np.linspace(0, size - 1, maximum, dtype=int))


def _finite_values(values: np.ndarray) -> np.ndarray:
    array = np.asarray(values, dtype=float)
    return array[np.isfinite(array)]


def _positive_values(values: np.ndarray) -> np.ndarray:
    array = np.asarray(values, dtype=float)
    return array[np.isfinite(array) & (array > 0.0)]


def _strictly_inside_levels(
    levels: Iterable[float],
    values: np.ndarray,
) -> np.ndarray:
    """Keep finite, unique contour levels strictly inside the data range."""

    finite = _finite_values(values)
    if finite.size == 0:
        return np.array([], dtype=float)
    lo = float(np.nanmin(finite))
    hi = float(np.nanmax(finite))
    if not hi > lo:
        return np.array([], dtype=float)

    candidate = np.asarray(tuple(levels), dtype=float)
    candidate = candidate[np.isfinite(candidate)]
    candidate = np.unique(candidate)
    return candidate[(candidate > lo) & (candidate < hi)]


def _log_125_levels(values: np.ndarray, max_levels: int = 9) -> np.ndarray:
    """Return readable 1-2-5 logarithmic contour levels."""

    finite = _positive_values(values)
    if finite.size == 0:
        return np.array([], dtype=float)
    vmin = float(np.nanmin(finite))
    vmax = float(np.nanmax(finite))
    if not vmax > vmin:
        return np.array([], dtype=float)

    candidates: list[float] = []
    e0 = int(np.floor(np.log10(vmin))) - 1
    e1 = int(np.ceil(np.log10(vmax))) + 1
    for exponent in range(e0, e1 + 1):
        for multiplier in (1.0, 2.0, 5.0):
            value = multiplier * 10.0**exponent
            if vmin < value < vmax:
                candidates.append(value)

    levels = np.unique(np.asarray(candidates, dtype=float))
    if levels.size > max_levels:
        select = np.unique(
            np.linspace(0, levels.size - 1, max_levels, dtype=int)
        )
        levels = levels[select]
    return levels


def _linear_levels(values: np.ndarray, max_levels: int = 8) -> np.ndarray:
    """Return readable linearly spaced contour levels."""

    finite = _finite_values(values)
    if finite.size == 0:
        return np.array([], dtype=float)
    vmin = float(np.nanmin(finite))
    vmax = float(np.nanmax(finite))
    if not vmax > vmin:
        return np.array([], dtype=float)

    locator = MaxNLocator(nbins=max_levels, min_n_ticks=3)
    levels = locator.tick_values(vmin, vmax)
    return levels[(levels > vmin) & (levels < vmax)]


def _signed_energy_levels(values: np.ndarray, max_levels: int = 9) -> np.ndarray:
    """Return useful energy levels for positive, negative, or signed data."""

    finite = _finite_values(values)
    if finite.size == 0:
        return np.array([], dtype=float)
    if np.all(finite > 0.0):
        return _log_125_levels(finite, max_levels=max_levels)
    if np.all(finite < 0.0):
        return -_log_125_levels(-finite, max_levels=max_levels)[::-1]

    negative = -_log_125_levels(
        -finite[finite < 0.0],
        max_levels=max(2, max_levels // 2),
    )[::-1]
    positive = _log_125_levels(
        finite[finite > 0.0],
        max_levels=max(2, max_levels // 2),
    )
    return np.unique(np.concatenate([negative, [0.0], positive]))


def _quantile_fallback_levels(
    values: np.ndarray,
    *,
    logarithmic: bool,
    n_levels: int = 6,
) -> np.ndarray:
    """Generate contours from interior quantiles when nice levels are sparse."""

    finite = _finite_values(values)
    if logarithmic:
        finite = finite[finite > 0.0]
        if finite.size:
            finite = np.log10(finite)
    if finite.size < 2:
        return np.array([], dtype=float)

    quantiles = np.linspace(0.12, 0.88, n_levels)
    levels = np.unique(np.quantile(finite, quantiles))
    if logarithmic:
        levels = 10.0**levels
    return _strictly_inside_levels(levels, values)


def _metric_contour_levels(
    values: np.ndarray,
    metric: str,
    requested_levels: Sequence[float] | None = None,
    *,
    max_levels: int = 9,
) -> np.ndarray:
    """Choose meaningful contour levels for the displayed quantity."""

    if requested_levels is not None:
        explicit = _strictly_inside_levels(requested_levels, values)
        if explicit.size:
            return explicit

    if metric == "probability":
        levels = _strictly_inside_levels(DEFAULT_PROBABILITY_CONTOURS, values)
        logarithmic = True
    elif metric == "minimum_speed":
        levels = _log_125_levels(values, max_levels=max_levels)
        logarithmic = True
    elif metric == "escape_energy":
        levels = _signed_energy_levels(values, max_levels=max_levels)
        logarithmic = bool(np.all(_finite_values(values) > 0.0))
    elif metric == "z_escape":
        levels = _linear_levels(values, max_levels=max_levels)
        logarithmic = False
    else:
        raise ValueError(f"Unknown metric: {metric}")

    levels = _strictly_inside_levels(levels, values)
    if levels.size < 3:
        fallback = _quantile_fallback_levels(
            values,
            logarithmic=logarithmic,
            n_levels=min(max_levels, 6),
        )
        levels = np.unique(np.concatenate([levels, fallback]))
        levels = _strictly_inside_levels(levels, values)

    if levels.size > max_levels:
        indices = np.unique(
            np.linspace(0, levels.size - 1, max_levels, dtype=int)
        )
        levels = levels[indices]
    return levels


def _norm_for_metric(values: np.ndarray, metric: str):
    finite = _finite_values(values)
    if finite.size == 0:
        raise RuntimeError(f"No finite values are available for {metric}")
    vmin = float(np.nanmin(finite))
    vmax = float(np.nanmax(finite))
    if vmax <= vmin:
        return Normalize(vmin=vmin - 0.5, vmax=vmax + 0.5)

    if metric in {"probability", "minimum_speed"} and vmin > 0.0:
        return LogNorm(vmin=vmin, vmax=vmax)
    if metric == "escape_energy" and vmin > 0.0:
        return LogNorm(vmin=vmin, vmax=vmax)
    if metric == "escape_energy" and vmin < 0.0 < vmax:
        scale = max(abs(vmin), abs(vmax))
        return SymLogNorm(
            linthresh=max(scale * 1e-6, np.finfo(float).tiny),
            vmin=vmin,
            vmax=vmax,
        )
    return Normalize(vmin=vmin, vmax=vmax)


def _format_contour_value(metric: str) -> Callable[[float], str]:
    if metric == "probability":
        return lambda x: f"P={x:.0e}" if x < 0.1 else f"P={x:.2g}"
    if metric == "minimum_speed":
        return lambda x: f"{x:.3g} m/s"
    if metric == "escape_energy":
        return lambda x: f"{x:.1e} J"
    return lambda x: f"{x:.1f} um"


def _mathieu_stable_mass_bands(
    eps_values: np.ndarray,
    *,
    mass_limits_kg: tuple[float, float],
    boundaries: Sequence[tuple[str, float]] = DEFAULT_MATHIEU_BOUNDARIES,
    low_ratio_stable: bool = DEFAULT_MATHIEU_LOW_RATIO_STABLE,
) -> list[tuple[np.ndarray, np.ndarray]]:
    """Return visible stable mass bands for every epsilon value.

    The system depends on ``epsilon / mass``.  Moving downward at fixed epsilon
    therefore moves from low to high ratio.  Starting from the known low-ratio
    stability state, the classification toggles at every fitted boundary.
    """

    eps = np.asarray(eps_values, dtype=float).reshape(-1)
    mass_lo, mass_hi = sorted(map(float, mass_limits_kg))
    if np.any(eps <= 0.0) or mass_lo <= 0.0:
        raise ValueError("Mathieu stable bands require positive log-axis values")

    valid_boundaries = [
        (str(name), float(slope))
        for name, slope in boundaries
        if np.isfinite(slope) and float(slope) > 0.0
    ]
    # Increasing epsilon/m corresponds to decreasing boundary slope.
    valid_boundaries.sort(key=lambda item: item[1], reverse=True)

    bands: list[tuple[np.ndarray, np.ndarray]] = []
    stable = bool(low_ratio_stable)
    current_upper = np.full_like(eps, mass_hi, dtype=float)

    for _, slope in valid_boundaries:
        boundary_mass = slope * eps
        if stable:
            lower = np.maximum(boundary_mass, mass_lo)
            upper = np.minimum(current_upper, mass_hi)
            visible = upper > lower
            bands.append(
                (
                    np.where(visible, lower, np.nan),
                    np.where(visible, upper, np.nan),
                )
            )
        current_upper = boundary_mass
        stable = not stable

    if stable:
        lower = np.full_like(eps, mass_lo, dtype=float)
        upper = np.minimum(current_upper, mass_hi)
        visible = upper > lower
        bands.append(
            (
                np.where(visible, lower, np.nan),
                np.where(visible, upper, np.nan),
            )
        )

    return bands


def overlay_mathieu_stability_boundaries(
    ax,
    *,
    mass_limits_kg: tuple[float, float],
    eps_limits: tuple[float, float],
    boundaries: Sequence[tuple[str, float]] = DEFAULT_MATHIEU_BOUNDARIES,
    low_ratio_stable: bool = DEFAULT_MATHIEU_LOW_RATIO_STABLE,
    show_stable_regions: bool = True,
    show_region_labels: bool = True,
    stability_color: str = "limegreen",
    stability_alpha: float = 0.13,
    stability_hatch: str = "//",
    linewidth: float = 1.8,
    line_alpha: float = 0.98,
    show_legend: bool = True,
) -> list[Any]:
    """Overlay Mathieu-stable bands and straight stability boundaries.

    Axes follow the requested convention:

        horizontal: charge fraction epsilon
        vertical:   particle mass m
    """

    mass_lo, mass_hi = sorted(map(float, mass_limits_kg))
    eps_lo, eps_hi = sorted(map(float, eps_limits))
    if mass_lo <= 0.0 or eps_lo <= 0.0:
        raise ValueError("Mathieu overlays require positive log-axis limits")

    eps_fill = np.geomspace(eps_lo, eps_hi, 900)
    stable_bands = _mathieu_stable_mass_bands(
        eps_fill,
        mass_limits_kg=(mass_lo, mass_hi),
        boundaries=boundaries,
        low_ratio_stable=low_ratio_stable,
    )

    stable_region_visible = False
    if show_stable_regions:
        for lower, upper in stable_bands:
            visible = np.isfinite(lower) & np.isfinite(upper) & (upper > lower)
            if not np.any(visible):
                continue
            stable_region_visible = True
            ax.fill_between(
                eps_fill,
                lower,
                upper,
                where=visible,
                interpolate=True,
                facecolor=stability_color,
                edgecolor="white",
                linewidth=0.0,
                hatch=stability_hatch,
                alpha=stability_alpha,
                zorder=4,
            )

            if show_region_labels:
                thickness = np.full_like(lower, -np.inf, dtype=float)
                thickness[visible] = np.log10(upper[visible] / lower[visible])
                index = int(np.nanargmax(thickness))
                if np.isfinite(thickness[index]) and thickness[index] > 0.08:
                    ax.text(
                        eps_fill[index],
                        np.sqrt(lower[index] * upper[index]),
                        "Mathieu stable",
                        ha="center",
                        va="center",
                        fontsize=8,
                        fontweight="bold",
                        color="white",
                        bbox={
                            "boxstyle": "round,pad=0.22",
                            "facecolor": "black",
                            "edgecolor": "none",
                            "alpha": 0.52,
                        },
                        zorder=9,
                    )

    line_styles: tuple[Any, ...] = ("--", "-.", ":", (0, (5, 1)))
    color_map = plt.get_cmap("tab10")
    line_handles: list[Any] = []

    for index, (name, slope) in enumerate(boundaries):
        slope = float(slope)
        if not np.isfinite(slope) or slope <= 0.0:
            continue

        visible_eps_lo = max(eps_lo, mass_lo / slope)
        visible_eps_hi = min(eps_hi, mass_hi / slope)
        if not visible_eps_hi > visible_eps_lo:
            continue

        eps_line = np.geomspace(visible_eps_lo, visible_eps_hi, 500)
        mass_line = slope * eps_line
        line, = ax.plot(
            eps_line,
            mass_line,
            linestyle=line_styles[index % len(line_styles)],
            linewidth=linewidth,
            alpha=line_alpha,
            color=color_map(index % 10),
            label=rf"{name}: $m={slope:.4e}\epsilon$ kg",
            zorder=10,
        )
        line_handles.append(line)

    legend_handles: list[Any] = []
    if stable_region_visible:
        legend_handles.append(
            Patch(
                facecolor=stability_color,
                edgecolor="white",
                hatch=stability_hatch,
                alpha=max(stability_alpha, 0.25),
                label="Mathieu stable region",
            )
        )
    legend_handles.extend(line_handles)

    if legend_handles and show_legend:
        ax.legend(
            handles=legend_handles,
            loc="best",
            fontsize=7,
            title="Coupled Mathieu stability",
            title_fontsize=8,
            framealpha=0.90,
        )
    return legend_handles


def plot_metric_heatmap(
    result: DynamicEscapeResult,
    metric: str,
    *,
    metric_contour_levels: Sequence[float] | None = None,
    contour_probabilities: Iterable[float] = DEFAULT_PROBABILITY_CONTOURS,
    overlay_probability_contours: Iterable[float] = (1e-5, 1e-3, 1e-1),
    show_probability_contours_on_other_maps: bool = True,
    show_mathieu_boundaries: bool = True,
    show_mathieu_stable_regions: bool = True,
    show_mathieu_region_labels: bool = True,
    mathieu_boundaries: Sequence[tuple[str, float]] = DEFAULT_MATHIEU_BOUNDARIES,
    mathieu_low_ratio_stable: bool = DEFAULT_MATHIEU_LOW_RATIO_STABLE,
    show_mathieu_legend: bool = True,
    max_metric_contours: int = 9,
    max_plot_points_per_axis: int = 1200,
) -> None:
    """Display one heat map with epsilon horizontal and mass vertical."""

    mass_idx = _plot_indices(result.m_arr_kg.size, max_plot_points_per_axis)
    eps_idx = _plot_indices(result.eps_arr.size, max_plot_points_per_axis)
    mass_idx = mass_idx[np.argsort(result.m_arr_kg[mass_idx])]
    eps_idx = eps_idx[np.argsort(result.eps_arr[eps_idx])]

    # Requested orientation: x = epsilon, y = mass.  The stored metric arrays
    # already have shape (n_mass, n_epsilon), so no transpose is required.
    x = result.eps_arr[eps_idx]
    y = result.m_arr_kg[mass_idx]
    probability = result.probability[np.ix_(mass_idx, eps_idx)].astype(float)

    if metric == "probability":
        values = probability
        label = r"$P(v>v_{\min})$"
        title = "Probability of exceeding the z-axis escape barrier"
        default_requested = tuple(contour_probabilities)
    elif metric == "minimum_speed":
        values = result.minimum_speed_m_s[np.ix_(mass_idx, eps_idx)].astype(float)
        label = r"Minimum speed $v_{\min}$ [m/s]"
        title = "Minimum speed required to exceed the escape barrier"
        default_requested = None
    elif metric == "escape_energy":
        values = result.escape_energy_J[np.ix_(mass_idx, eps_idx)].astype(float)
        label = r"Potential energy at escape point [J]"
        title = "Potential energy at the mass-dependent escape point"
        default_requested = None
    elif metric == "z_escape":
        values = (
            result.z_escape_m[np.ix_(mass_idx, eps_idx)].astype(float) * 1e6
        )
        label = r"Escape coordinate $z_{\rm esc}$ [$\mu$m]"
        title = "z coordinate of the mass-dependent escape point"
        default_requested = None
    else:
        raise ValueError(
            "metric must be probability, minimum_speed, escape_energy, or z_escape"
        )

    if not np.any(np.isfinite(values)):
        raise RuntimeError(f"No retained values are available for {metric}")

    requested = metric_contour_levels
    if requested is None:
        requested = default_requested
    levels = _metric_contour_levels(
        values,
        metric,
        requested_levels=requested,
        max_levels=max_metric_contours,
    )

    norm = _norm_for_metric(values, metric)
    fig, ax = plt.subplots(figsize=(11, 7.5))
    mesh = ax.pcolormesh(
        x,
        y,
        np.ma.masked_invalid(values),
        shading="auto",
        norm=norm,
        rasterized=True,
        zorder=1,
    )

    if show_mathieu_boundaries or show_mathieu_stable_regions:
        overlay_mathieu_stability_boundaries(
            ax,
            mass_limits_kg=(float(np.min(y)), float(np.max(y))),
            eps_limits=(float(np.min(x)), float(np.max(x))),
            boundaries=mathieu_boundaries,
            low_ratio_stable=mathieu_low_ratio_stable,
            show_stable_regions=show_mathieu_stable_regions,
            show_region_labels=show_mathieu_region_labels,
            linewidth=1.8 if show_mathieu_boundaries else 0.0,
            line_alpha=0.98 if show_mathieu_boundaries else 0.0,
            show_legend=show_mathieu_legend,
        )

    if levels.size:
        contours = ax.contour(
            x,
            y,
            np.ma.masked_invalid(values),
            levels=levels,
            colors="black",
            linewidths=1.05,
            zorder=6,
        )
        ax.clabel(
            contours,
            inline=True,
            fontsize=8,
            fmt=_format_contour_value(metric),
        )

    if metric != "probability" and show_probability_contours_on_other_maps:
        p_levels = _strictly_inside_levels(
            overlay_probability_contours,
            probability,
        )
        if p_levels.size:
            p_contours = ax.contour(
                x,
                y,
                np.ma.masked_invalid(probability),
                levels=p_levels,
                colors="white",
                linewidths=1.0,
                linestyles="--",
                alpha=0.95,
                zorder=7,
            )
            ax.clabel(
                p_contours,
                inline=True,
                fontsize=7,
                fmt=lambda value: f"P={value:.0e}",
            )

    ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlim(float(np.min(x)), float(np.max(x)))
    ax.set_ylim(float(np.min(y)), float(np.max(y)))
    ax.set_xlabel(r"Charge fraction $\epsilon$  ($Q=\epsilon e$)")
    ax.set_ylabel(r"DM mass $m_{\rm DM}$ [kg]")
    ax.set_title(title)
    ax.grid(True, which="both", alpha=0.2)

    colorbar = fig.colorbar(mesh, ax=ax)
    colorbar.set_label(label)
    fig.tight_layout()
    plt.show()


def plot_all_heatmaps(
    result: DynamicEscapeResult,
    *,
    metric_contour_levels: Mapping[str, Sequence[float]] | None = None,
    contour_probabilities: Iterable[float] = DEFAULT_PROBABILITY_CONTOURS,
    overlay_probability_contours: Iterable[float] = (1e-5, 1e-3, 1e-1),
    show_probability_contours_on_other_maps: bool = True,
    show_mathieu_boundaries: bool = True,
    show_mathieu_stable_regions: bool = True,
    show_mathieu_region_labels: bool = True,
    mathieu_boundaries: Sequence[tuple[str, float]] = DEFAULT_MATHIEU_BOUNDARIES,
    mathieu_low_ratio_stable: bool = DEFAULT_MATHIEU_LOW_RATIO_STABLE,
    show_mathieu_legend: bool = True,
    max_metric_contours: int = 9,
    max_plot_points_per_axis: int = 1200,
) -> None:
    """Display all four heat maps as separate figures."""

    custom = {} if metric_contour_levels is None else dict(metric_contour_levels)
    for metric in (
        "probability", "minimum_speed", "escape_energy", "z_escape",
    ):
        plot_metric_heatmap(
            result,
            metric,
            metric_contour_levels=custom.get(metric),
            contour_probabilities=contour_probabilities,
            overlay_probability_contours=overlay_probability_contours,
            show_probability_contours_on_other_maps=(
                show_probability_contours_on_other_maps
            ),
            show_mathieu_boundaries=show_mathieu_boundaries,
            show_mathieu_stable_regions=show_mathieu_stable_regions,
            show_mathieu_region_labels=show_mathieu_region_labels,
            mathieu_boundaries=mathieu_boundaries,
            mathieu_low_ratio_stable=mathieu_low_ratio_stable,
            show_mathieu_legend=show_mathieu_legend,
            max_metric_contours=max_metric_contours,
            max_plot_points_per_axis=max_plot_points_per_axis,
        )


# =============================================================================
# DIAGNOSTIC PROFILE PLOTS
# =============================================================================


def plot_selected_potential_slices(
    potential_energy_fn: Callable[..., Any],
    result: DynamicEscapeResult,
    *,
    n_examples: int = 6,
    potential_kwargs: Mapping[str, Any] | None = None,
) -> None:
    """Plot representative z profiles and detected first local maxima."""

    if result.search_mode == "ratio":
        ratio_data = result.search_metadata.get("ratio_data")
        if not isinstance(ratio_data, dict):
            return
        valid = np.flatnonzero(ratio_data["peak_found"])
        if valid.size == 0:
            return
        chosen = valid[
            np.unique(
                np.linspace(0, valid.size - 1, min(n_examples, valid.size), dtype=int)
            )
        ]
        masses = ratio_data["m_reference_kg"][chosen]
        charges = ratio_data["eps_reference"][chosen]
        labels = [
            rf"$\epsilon/m=10^{{{ratio_data['log_ratio'][i]:.2f}}}$"
            for i in chosen
        ]
    else:
        sparse_data = result.search_metadata.get("sparse_data")
        if not isinstance(sparse_data, dict):
            return
        mm, ee = np.meshgrid(
            sparse_data["m_coarse_kg"],
            sparse_data["eps_coarse"],
            indexing="ij",
        )
        found = sparse_data["peak_found"].reshape(-1)
        valid = np.flatnonzero(found)
        if valid.size == 0:
            return
        chosen = valid[
            np.unique(
                np.linspace(0, valid.size - 1, min(n_examples, valid.size), dtype=int)
            )
        ]
        masses = mm.reshape(-1)[chosen]
        charges = ee.reshape(-1)[chosen]
        labels = [
            rf"$m={m:.1e},\ \epsilon={e:.2g}$"
            for m, e in zip(masses, charges)
        ]

    profiles = _evaluate_pair_profiles(
        potential_energy_fn,
        result.z_grid_m,
        masses,
        charges,
        potential_kwargs,
    )
    z_peak, u_peak, found = find_first_local_maximum_batch(
        profiles,
        result.z_grid_m,
        quadratic_refinement=True,
    )

    fig, ax = plt.subplots(figsize=(10, 7))
    for i, label in enumerate(labels):
        ax.plot(result.z_grid_m * 1e6, profiles[i], label=label)
        if found[i]:
            ax.scatter(z_peak[i] * 1e6, u_peak[i], s=35)
    ax.set_xlabel(r"$z$ [$\mu$m]")
    ax.set_ylabel("Potential energy [J]")
    ax.set_title("Representative z-axis potential slices and escape maxima")
    ax.grid(True, alpha=0.25)
    ax.legend()
    fig.tight_layout()
    plt.show()


# =============================================================================
# USER ENTRY POINT
# =============================================================================


def run_dynamic_escape_scan(
    potential_energy_fn: Callable[..., Any],
    *,
    temperature_K: float = 300.0,
    probability_threshold: float = 1e-6,
    n_mass: int = 5000,
    n_eps: int = 5000,
    z_min_m: float = 0.1e-6,
    z_max_m: float = 100.0e-6,
    n_z: int = 1000,
    search_mode: str = "auto",
    max_exact_ratio_profiles: int = 100_000,
    approximate_ratio_profiles: int = 20_000,
    pair_profile_chunk_size: int = 64,
    eps_grid_chunk_size: int = 32,
    fallback_mass_points: int = 161,
    fallback_eps_points: int = 161,
    ratio_profile_relative_tolerance: float = 2e-6,
    ratio_z_tolerance_grid_steps: float = 2.0,
    potential_kwargs: Mapping[str, Any] | None = None,
    show_heatmaps: bool = True,
    show_example_slices: bool = True,
    metric_contour_levels: Mapping[str, Sequence[float]] | None = None,
    contour_probabilities: Iterable[float] = DEFAULT_PROBABILITY_CONTOURS,
    overlay_probability_contours: Iterable[float] = (1e-5, 1e-3, 1e-1),
    show_probability_contours_on_other_maps: bool = True,
    show_mathieu_boundaries: bool = True,
    show_mathieu_stable_regions: bool = True,
    show_mathieu_region_labels: bool = True,
    mathieu_boundaries: Sequence[tuple[str, float]] = DEFAULT_MATHIEU_BOUNDARIES,
    mathieu_low_ratio_stable: bool = DEFAULT_MATHIEU_LOW_RATIO_STABLE,
    show_mathieu_legend: bool = True,
    max_metric_contours: int = 9,
    max_plot_points_per_axis: int = 1200,
) -> DynamicEscapeResult:
    """Run the requested mass-charge scan."""

    m_arr = np.logspace(-20, -27, num=n_mass, endpoint=True, base=10.0)
    eps_arr = np.logspace(-3, 3, num=n_eps, endpoint=True, base=10.0)

    result = calculate_dynamic_escape_grid(
        potential_energy_fn,
        m_arr,
        eps_arr,
        temperature_K=temperature_K,
        probability_threshold=probability_threshold,
        z_min_m=z_min_m,
        z_max_m=z_max_m,
        n_z=n_z,
        search_mode=search_mode,
        max_exact_ratio_profiles=max_exact_ratio_profiles,
        approximate_ratio_profiles=approximate_ratio_profiles,
        pair_profile_chunk_size=pair_profile_chunk_size,
        eps_grid_chunk_size=eps_grid_chunk_size,
        fallback_mass_points=fallback_mass_points,
        fallback_eps_points=fallback_eps_points,
        ratio_profile_relative_tolerance=ratio_profile_relative_tolerance,
        ratio_z_tolerance_grid_steps=ratio_z_tolerance_grid_steps,
        quadratic_refinement=True,
        potential_kwargs=potential_kwargs,
        progress=True,
    )

    retained = int(np.count_nonzero(np.isfinite(result.probability)))
    total = int(result.probability.size)
    peaks = int(np.count_nonzero(result.peak_found))

    print("\nDynamic escape-barrier scan complete")
    print(f"Search mode                  : {result.search_mode}")
    print(f"Mass samples                 : {m_arr.size}")
    print(f"Charge samples               : {eps_arr.size}")
    print(f"z samples per profile        : {result.z_grid_m.size}")
    print(f"Grid points with local max   : {peaks}/{total}")
    print(f"Pairs with P >= threshold    : {retained}/{total}")
    print(f"Probability threshold        : {probability_threshold:.3e}")

    finite_z = result.z_escape_m[np.isfinite(result.z_escape_m)]
    if finite_z.size:
        print(
            "Retained escape-z range      : "
            f"{np.min(finite_z) * 1e6:.6g} to "
            f"{np.max(finite_z) * 1e6:.6g} um"
        )

    if result.search_mode == "ratio":
        print(
            "Ratio lookup                 : "
            f"{result.search_metadata.get('lookup_kind')} with "
            f"{result.search_metadata.get('n_ratio_profiles')} profiles"
        )
    else:
        print(
            "Sparse fallback grid         : "
            f"{result.search_metadata.get('fallback_mass_points')} x "
            f"{result.search_metadata.get('fallback_eps_points')}"
        )

    if show_example_slices:
        plot_selected_potential_slices(
            potential_energy_fn,
            result,
            potential_kwargs=potential_kwargs,
        )
    if show_heatmaps:
        plot_all_heatmaps(
            result,
            metric_contour_levels=metric_contour_levels,
            contour_probabilities=contour_probabilities,
            overlay_probability_contours=overlay_probability_contours,
            show_probability_contours_on_other_maps=(
                show_probability_contours_on_other_maps
            ),
            show_mathieu_boundaries=show_mathieu_boundaries,
            show_mathieu_stable_regions=show_mathieu_stable_regions,
            show_mathieu_region_labels=show_mathieu_region_labels,
            mathieu_boundaries=mathieu_boundaries,
            mathieu_low_ratio_stable=mathieu_low_ratio_stable,
            show_mathieu_legend=show_mathieu_legend,
            max_metric_contours=max_metric_contours,
            max_plot_points_per_axis=max_plot_points_per_axis,
        )
    return result


result = run_dynamic_escape_scan(
    potential_energy,
    temperature_K=T_dm,
    probability_threshold=1e-6,
    n_mass=5000,
    n_eps=5000,
    z_min_m=0.1e-6,
    z_max_m=100.0e-6,
    n_z=1000,
    search_mode="auto",
    pair_profile_chunk_size=64,
    eps_grid_chunk_size=32,
    show_heatmaps=True,
    show_example_slices=True,
    show_mathieu_boundaries=True,
    show_mathieu_stable_regions=True,
    show_mathieu_region_labels=True,
    show_probability_contours_on_other_maps=True,
    max_metric_contours=9,
)
```
